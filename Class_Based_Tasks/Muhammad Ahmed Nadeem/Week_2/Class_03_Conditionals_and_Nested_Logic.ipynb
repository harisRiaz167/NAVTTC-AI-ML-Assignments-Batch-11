```python
# 1. Store 3 users
users = {
    "ali": {"password": "123", "balance": 5000},
    "sara": {"password": "456", "balance": 10000},
    "hamza": {"password": "789", "balance": 8000}
}

print("==========================================")
print("          WELCOME TO PYTHON BANK          ")
print("==========================================")

# 2. Login
username = input("Enter Username: ")
password = input("Enter Password: ")

if username in users and password == users[username]["password"]:
    print("\n------------------------------------------")
    print(f"  Login Successful! Welcome, {username.title()}")
    print("------------------------------------------")

    # 3. Operations Loop
    while True:
        choice = input("\n[1-Deposit | 2-Withdraw | 3-Balance | 4-Exit]: ")

        if choice == "1":
            amount = int(input("Enter amount to deposit: "))
            users[username]["balance"] = users[username]["balance"] + amount
            print("\n------------------------------------------")
            print(f"  Deposit Amount  : +{amount}")
            print(f"  Updated Balance :  {users[username]['balance']}")
            print("------------------------------------------")

        elif choice == "2":
            amount = int(input("Enter amount to withdraw: "))
            if amount > users[username]["balance"]:
                print("\n------------------------------------------")
                print("  [ERROR] Insufficient Balance!")
                print(f"  Available Balance: {users[username]['balance']}")
                print("------------------------------------------")
            else:
                users[username]["balance"] = users[username]["balance"] - amount
                print("\n------------------------------------------")
                print(f"  Withdraw Amount : -{amount}")
                print(f"  Updated Balance :  {users[username]['balance']}")
                print("------------------------------------------")

        elif choice == "3":
            print("\n------------------------------------------")
            print(f"  Current Balance : {users[username]['balance']}")
            print("------------------------------------------")

        elif choice == "4":
            print("\n==========================================")
            print("  Thank you for using our bank! Goodbye.  ")
            print("==========================================")
            break

        else:
            print("\n[!] Invalid Choice! Please enter 1, 2, 3, or 4.")

else:
    print("\n------------------------------------------")
    print("  [!] Wrong Username or Password.")
    print("------------------------------------------")
```

```python
numbers = [1, 2, 3, 4, 5, 6, 7]
print("Available List:", numbers)

number1 = int(input("Enter First No: "))
number2 = int(input("Enter Second No: "))

total = number1 + number2
print("Sum is:", total)

found = False

for i in range(len(numbers)):
    for j in range(i + 1, len(numbers)):
        if numbers[i] == number1 and numbers[j] == number2:
            print("Index is:", (i, j))
            found = True
            break
            
    if found:
        break

if not found:
    print("not found")
```
