```python
number1 = int(input("Enter your First Number"))
number2 = int(input("Enter your Second Number"))
number3 = int(input("Enter your Third Number"))
number4 = int(input("Enter your Forth Number"))

numbers = [number1, number2, number3, number4]

print("Numbers in List", numbers)

maximum = numbers[0]
minimum = numbers[0]

for num in numbers:
    if num > maximum:
       maximum = num

    if num < minimum:
       minimum = number1

print("Maximum Number", maximum)
print("Minimum Number", minimum)
```

Find Length from this - Even or Odd

```python
x = ['alphabets', 'numeric', 'data types', '123', 'wild', 'abc', 'chars']

even = []
odd = []

for i in x:
    if len(i) % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even Lisr:", even)
print("Odd List",odd)
```
