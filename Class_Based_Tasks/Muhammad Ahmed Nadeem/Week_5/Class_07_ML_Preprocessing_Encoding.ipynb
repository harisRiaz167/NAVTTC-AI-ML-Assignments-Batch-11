```python
import pandas as pd

df_houses = pd.DataFrame({
    "City": ["Lahore", "Lahore", "Lahore",
             "Karachi", "Karachi", "Karachi",
             "Islamabad", "Islamabad"],
    "Price_M": [10, 12, 11, 8, 9, 10, 15, 17]
})

city_means = df_houses.groupby("City")["Price_M"].mean()

df_houses["City_Target_Encoded"] = df_houses["City"].map(city_means)

print("Category means:")
print(city_means)
print("\nEncoded data:")
print(df_houses)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Load Data
df = pd.read_csv('sampledata1.csv')

# 2. Label Encoding
le = LabelEncoder()
df['Category_Label'] = le.fit_transform(df['Category'])
df['Region_Label'] = le.fit_transform(df['Region'])

# 3. Target Encoding (Target = Sales)
df['Category_Target_Encoded'] = df['Category'].map(df.groupby('Category')['Sales'].mean())
df['Region_Target_Encoded'] = df['Region'].map(df.groupby('Region')['Sales'].mean())

# Columns compare karein
print(df[['Category', 'Category_Label', 'Category_Target_Encoded', 'Region', 'Region_Label', 'Region_Target_Encoded']].head())
```

```python
import numpy as np
from collections import Counter

activity_points = np.array([[0, 0], [1, 1], [3, 2], [4, 4]])
activity_labels = np.array([0, 0, 1, 1])
activity_query = np.array([2.5, 2.5])

# Step 1: Euclidean distance
distances = np.sqrt(np.sum((activity_points - activity_query) ** 2, axis=1))
print("Distances:", distances)

# Step 2: Sort (indices nikalo)
sorted_idx = np.argsort(distances)
print("Sorted labels:", activity_labels[sorted_idx])

# Step 3: Vote
for k in [1, 3]:
    k_labels = activity_labels[sorted_idx[:k]]
    prediction = Counter(k_labels).most_common(1)[0][0]
    print(f"k={k} -> Prediction: {prediction}")
```
