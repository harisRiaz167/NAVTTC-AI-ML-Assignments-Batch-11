```python
s = "abcabc"

null = ""
count = 0

for char in s:
    if char in null:
        break

    null += char
    count +=1

print("Output", count)
```

```python
vocab = {
'[PAD]': 0, 
'[UNK]': 1, 
'i': 2, 
'love': 3, 
'this': 4, 
'film': 5
}
sentence = "I love this new film"
max_len = 7

unk_id = vocab['[UNK]']
pad_id = vocab['[PAD]']

# Step 1 & 2: Tokenization + Numericalization (1 Line)
token_ids = [vocab.get(w, unk_id) for w in sentence.lower().split()]

# Step 3: Padding + Truncation (1 Line - Bina kisi while loop ke)
padded_input = (token_ids + [pad_id] * max_len)[:max_len]

print("Token IDs:", token_ids)
print("Padded Input for CNN:", padded_input)
```
