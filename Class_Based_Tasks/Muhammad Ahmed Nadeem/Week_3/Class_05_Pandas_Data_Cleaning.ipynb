```python
import pandas as pd
import numpy as np

raw = pd.DataFrame({
    'name': ['Ali', 'Sara', 'Amina', 'Amina', 'Usman'],
    'city': [' lahore', 'KARACHI', 'Lahore ', 'Lahore ', 'Islamabad'],
    'hours': [2, np.nan, 4, 4, 3],                          # Fixed: np.nan (or None)
    'score': ['55', '72', 'missing', 'missing', '64'],
    'date': ['2026-09-01', '2026-09-02', 'bad date', 'bad date', '2026-09-05']
})

# Fixed: drop_duplicates() without extra 'l'
# .copy() prevents SettingWithCopyWarning
clean = raw.drop_duplicates().copy()

clean['city'] = clean['city'].str.strip().str.title()
clean['score'] = pd.to_numeric(clean['score'], errors='coerce')
clean['date'] = pd.to_datetime(clean['date'], errors='coerce')
clean['hours'] = clean['hours'].fillna(clean['hours'].median())

print(clean)
```
