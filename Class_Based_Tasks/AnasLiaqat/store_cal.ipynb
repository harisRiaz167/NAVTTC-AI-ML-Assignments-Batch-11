```python
print("===WELCOME TO MY STORE ===")
print("--------------------------")
print("\nEnter quantity for each item:")
apple = int(input("How many Apples? "))
banana = int(input("How many Bananas? "))
milk = int(input("How many Milk? "))
bread = int(input("How many Bread? "))

apple_total = apple * 300
banana_total = banana * 150
milk_total = milk * 220
bread_total = bread * 150

total = apple_total + banana_total + milk_total + bread_total

print("\n===YOUR TOTAL BILL ===")

print("Apple  :", apple, "x 100 =", apple_total)
print("Banana :", banana, "x 50  =", banana_total)
print("Milk   :", milk, "x 200 =", milk_total)
print("Bread  :", bread, "x 150 =", bread_total)

print("--------------------------")
print("Total Bill = Rs.", total)

print("Thank you for shopping!")
```
