```python
#cnn
import re
def simple_tokenize(text):
    return re.findall(r"[a-z']+", text.lower())
dict = {'PAD': 0, 'UNK': 1, 'i': 2, 'love': 3, 'this' : 4 , 'film': 5}

def encode(text, dict):
    IDs = []
    for token in simple_tokenize(text):
        IDs.append(dict.get(token, dict['UNK']))
    IDs += [dict['PAD']] * (6 - len(IDs))
    return IDs
enc = encode("I love this film", dict)
print(enc)
```
