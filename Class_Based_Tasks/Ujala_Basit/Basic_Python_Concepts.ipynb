```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))
c = int(input("Enter c: "))
d = int(input("Enter d: "))

sum = a+b+c+d

mean = sum / 4
max_num = a
if b > max_num:
    max_num = b
if c > max_num:
    max_num = c
if d > max_num:
    max_num = d

min_num = a
if b < min_num:
    min_num = b
if c < min_num:
    min_num = c
if d < min_num:
    min_num = d

range_value = max_num - min_num
print("Mean =", mean)
print("Range =", range_value)


```

```python
students = [
    {
        "student_id": 101,
        "username": "ali123",
        "password": "1234",
        "name": "Ali",
        "class": "10",
        "section": "A",
        "Math": 85,
        "English": 78,
        "Science": 92,
        "Computer": 88,
        "totalmarks": 343,
        "grade": "A"
    },

    {
        "student_id": 102,
        "username": "sara123",
        "password": "5678",
        "name": "Sara",
        "class": "10",
        "section": "B",
        "Math": 92,
        "English": 89,
        "Science": 95,
        "Computer": 91,
        "totalmarks": 367,
        "grade": "A+"
    },

    {
        "student_id": 103,
        "username": "hamza123",
        "password": "2468",
        "name": "Hamza",
        "class": "10",
        "section": "A",
        "Math": 65,
        "English": 72,
        "Science": 68,
        "Computer": 75,
        "totalmarks": 280,
        "grade": "B"
    }
]


def login(username, password):

    for student in students:

        if student["username"] == username and student["password"] == password:
            return student

    return None


def student_report(student):

    print("\n")
    print("==============================================")
    print("           STUDENT RESULT PORTAL")
    print("==============================================")

    print(f" Student Name : {student['name']}")
    print(f" Class        : {student['class']}")
    print(f" Section      : {student['section']}")

    print("----------------------------------------------")
    print("              SUBJECT RESULTS")
    print("----------------------------------------------")

    print(f" Mathematics  : {student['Math']}")
    print(f" English      : {student['English']}")
    print(f" Science      : {student['Science']}")
    print(f" Computer     : {student['Computer']}")

    print("----------------------------------------------")

    print(f" Total Marks  : {student['totalmarks']} / 400")
    print(f" Grade        : {student['grade']}")

    print("==============================================")
    print("        Thank you for using the portal!")
    print("==============================================")


for i in range(3):

    username = input("Enter Username: ")
    password = input("Enter Password: ")

    student = login(username, password)

    if student:

        print("\nLogin Successful!")
        break

    else:

        print("Wrong Username or Password!")

else:

    print("\nYou have used all 3 attempts.")
    exit()



student_report(student)
```
