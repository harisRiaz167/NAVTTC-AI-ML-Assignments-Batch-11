## 1. FizzBuzz

**Problem:** For numbers from `1` to `n`, print `FizzBuzz` for multiples of both 3 and 5, `Fizz` for multiples of 3, `Buzz` for multiples of 5, and the number otherwise.

```python
def test(l):
    newlist = []
    for i in range(len(l)):
        if l[i] % 5 == 0 and l[i] % 3 == 0:
            newlist.append("FizzBuzz")
        elif l[i] % 3 == 0:
            newlist.append("Fizz")
        elif l[i] % 5 == 0:
            newlist.append("Buzz")
        else:
            newlist.append(str(l[i]))
    print(newlist)

n = 15
l = list(range(1, n + 1))
test(l)
```

## 2. Move Zeroes

```python
l = [0,1,0,2,0,4,0,0]
newl = []

for i in l:
    if i != 0:
        newl.append(i)

for i in l:
    if i == 0:
        newl.append(i)

print(newl)
```

## 3. Majority Element

**Problem:** Find the number that appears more than half the time. The example has a majority element.

```python
l = [1,1,3,3,3,3,3,3,3,3,3,3,3]
length = len(l)//2

for i in range(len(l)):
    count = 0
    for j in range(len(l)):
        if l[i] == l[j]:
            count += 1
    if count > length:
        print(l[i])
        break
```

## 4. First Unique Character in a String

**Problem:** Print the index of the first character that appears only once. Print `-1` if every character repeats.

```python
def unique(string):
    for i in range(len(string)):
        count = 0
        for j in range(len(string)):
            if string[i] == string[j]:
                count += 1
        if count == 1:
            print(i)
            return
    print(-1)

value = "leetcode"
unique(value)
```

## 5. Product of Array Except Self

**Problem:** Make a new list where each value is the product of all the other values.

```python
l = [1, 2, 3, 4, 5]
newl = []

for i in range(len(l)):
    product = 1
    for j in range(len(l)):
        if i == j:
            continue
        product *= l[j]
    newl.append(product)

print(newl)
```

## 7. Remove Minimum and Maximum

**Problem:** Make a new list after removing one occurrence of the minimum and one occurrence of the maximum.

```python
l = [0,1,2,3,4,5]
min = l[0]
max = l[0]

for i in range(len(l)):
    if l[i] < min:
        min = l[i]
    if l[i] > max:
        max = l[i]

new_l = []
min_removed = False
max_removed = False

for i in l:
    if i == min and not min_removed:
        min_removed = True
    elif i == max and not max_removed:
        max_removed = True
    else:
        new_l.append(i)

print(new_l)
```

## 8. Valid Parentheses

**Problem:** Check whether every opening bracket is closed by the correct bracket in the right order.

```python
bracket = "()[]{}"
dic = {")": "(", "]": "[", "}": "{"}
stack = []
valid = True

for i in range(len(bracket)):
    if bracket[i] in "([{":
        stack.append(bracket[i])
    elif bracket[i] in dic:
        if len(stack) == 0 or stack[-1] != dic[bracket[i]]:
            valid = False
            break
        stack.pop()

if len(stack) != 0:
    valid = False

print(valid)
```

## 10. Group Anagrams

Given a list of strings, separate the strings that contain the same characters, regardless of their order. The comparison should be case-insensitive.



```python
l = ["ate", "eat", "tea", "bat", "naA", "tan"]
l_one = []
l_two = []

for word in l:
    found = False

    for other in l:
        if word != other and sorted(word.lower()) == sorted(other.lower()):
            found = True
            break

    if found:
        l_one.append(word)
    else:
        l_two.append(word)

print(l_one)
print(l_two)
```

## 15. Extra Character Counts

**Problem:** Show which characters occur more times in `t` than in `s`, and how many extra copies there are.

```python
s = "abcd"
t = "acbsdde"
newl = {}

for char in t:
    count = t.count(char) - s.count(char)
    if count > 0:
        newl[char] = count

print(newl)
```

## 18. Contains Duplicate

**Problem:** Print `True` if any number appears more than once; otherwise print `False`.
