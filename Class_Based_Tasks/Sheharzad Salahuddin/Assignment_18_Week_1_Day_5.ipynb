## Topic: Python — Continue Statement

**Objective:** Practice using the continue statement.

**Reference:** https://www.geeksforgeeks.org/python-continue-statement/

**Skip the number 3**

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

**Skip even numbers**

```python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

## Observation

- The continue statement skips the current iteration.
- When i equals 3, that number is not printed.
- Even numbers are skipped in the second example.
- The loop continues with the next iteration instead of terminating.
