## What is Pandas?

Pandas is a Python library for working with data sets. It has functions for **analysing, cleaning, exploring and manipulating** data. The name refers to both "Panel Data" and "Python Data Analysis", and it was created by Wes McKinney in 2008.

**Why use Pandas?** It can analyse big data and draw conclusions based on statistics, and it can clean messy data sets so they become readable and relevant.

**What can Pandas do?** It gives answers about the data, such as: Is there a correlation between two or more columns? What is the average value? The maximum? The minimum?

**Data file:** this notebook uses `data_full.csv` (the W3Schools workout data set, 169 rows), which must be in the same folder.

**Diagram:** Pandas supports every step from loading data to using the results.



**Load a CSV file into a DataFrame**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.to_string())
```

**Explanation:** read_csv() loads the file into a DataFrame, and to_string() prints every row. The data records workouts: their duration in minutes, average pulse, maximum pulse and calories burned.

**Diagram:** A Series is one labelled column; a DataFrame is a table made of Series.



**A quick look at the data**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.head())
print("Shape:", df.shape)
print("Columns:", list(df.columns))
```

**Explanation:** head() shows the first 5 rows, shape gives the number of rows and columns, and columns lists the column names.

**Answer questions about the data: average, maximum and minimum**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print("Average duration:", round(df["Duration"].mean(), 1), "minutes")
print("Longest workout:", df["Duration"].max(), "minutes")
print("Lowest pulse:", df["Pulse"].min())
print("Highest calories:", df["Calories"].max())
```

**Explanation:** Each column is a Series with methods such as mean(), max() and min(), so common questions take one line of code.

**Answer questions about the data: correlation**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.corr().round(2))
```

**Explanation:** corr() shows how strongly each pair of columns is related. Duration and Calories have a high value (about 0.92), so longer workouts burn more calories.

**Select and filter data**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

long_workouts = df[df["Duration"] > 100]
print(long_workouts)
print("Number of long workouts:", len(long_workouts))
```

**Explanation:** A condition inside square brackets keeps only the rows where it is True, here the workouts longer than 100 minutes.

**A simple chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data_full.csv")

df.plot(kind="scatter", x="Duration", y="Calories", title="Duration vs Calories")
plt.show()
```

**Explanation:** Pandas can plot data directly with plot(). The scatter plot shows calories rising with duration.

## Observation

- Pandas is a Python library for analysing, cleaning, exploring and manipulating data.
- Its two main structures are the Series (one column) and the DataFrame (a table).
- read_csv() loads a CSV file into a DataFrame in one line.
- Methods such as mean(), max(), min() and corr() answer common questions about the data.
- Rows can be filtered with conditions, and data can be plotted directly.
