## Topic: Statistics — Standard Deviation

**Objective:** Calculate standard deviation.

**Reference:** https://www.w3schools.com/python/python_ml_standard_deviation.asp

Standard deviation is a number that describes how spread out the values are.

- A **low** standard deviation means most of the numbers are close to the mean (average) value.
- A **high** standard deviation means the values are spread out over a wider range.

Example: the speed of 7 cars has been registered.

**Diagram:** Both lists are 7 car speeds, but the second is much more spread out.



**Standard deviation of closely grouped values**

```python
import numpy

speed = [86, 87, 88, 86, 87, 85, 86]

x = numpy.std(speed)

print(x)
```

**Explanation:** The standard deviation is 0.9, which means most values are within 0.9 of the mean (86.4).

**Standard deviation of widely spread values**

```python
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

x = numpy.std(speed)

print(x)
```

**Explanation:** The standard deviation is 37.85, which means most values are within 37.85 of the mean (77.4). The values are far more spread out.

**Diagram:** Variance is the average squared distance from the mean; standard deviation is its square root.



**Variance step by step**

```python
speed = [32, 111, 138, 28, 59, 77, 97]

mean = sum(speed) / len(speed)
print("1. Mean:", round(mean, 2))

differences = [x - mean for x in speed]
print("2. Differences:", [round(d, 2) for d in differences])

squares = [d ** 2 for d in differences]
print("3. Squared differences:", [round(s, 2) for s in squares])

variance = sum(squares) / len(squares)
print("4. Variance:", round(variance, 2))
```

**Explanation:** Variance shows how spread out the values are. To calculate it: find the mean, subtract the mean from each value, square each difference, and take the average of the squares.

**Variance with NumPy**

```python
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

x = numpy.var(speed)

print(x)
```

**Explanation:** NumPy's var() method gives the same variance, 1432.24.

**Standard deviation is the square root of the variance**

```python
import math
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

print(math.sqrt(numpy.var(speed)))
print(numpy.std(speed))
```

**Explanation:** The standard deviation is the square root of the variance: √1432.24 = 37.85. Both lines give the same result.

**Symbols**

- Standard deviation is often written σ (sigma).
- Variance is often written σ² (sigma squared).

**Population vs sample standard deviation**

```python
import numpy
import statistics

speed = [32, 111, 138, 28, 59, 77, 97]

print("Population (ddof=0):", round(numpy.std(speed), 2), "|", round(statistics.pstdev(speed), 2))
print("Sample (ddof=1):    ", round(numpy.std(speed, ddof=1), 2), "|", round(statistics.stdev(speed), 2))
```

**Explanation:** numpy.std() treats the values as a whole population and divides by n. When the values are only a sample of a bigger group, dividing by n − 1 (ddof=1) gives a slightly larger, fairer estimate. Python's statistics module offers both as pstdev() and stdev().

**Values within one standard deviation of the mean**

```python
import numpy

speed = numpy.array([32, 111, 138, 28, 59, 77, 97])

mean, std = speed.mean(), speed.std()
inside = speed[(speed >= mean - std) & (speed <= mean + std)]

print(f"Range: {mean - std:.2f} to {mean + std:.2f}")
print("Values inside:", inside)
print(f"{len(inside)} of {len(speed)} values")
```

**Explanation:** The band from one standard deviation below the mean to one above covers a large share of the values. Here 4 of the 7 speeds fall between 39.58 and 115.27; for data with a bell-shaped (normal) distribution, about 68% of values fall in this band.

## Observation

- Standard deviation measures how spread out values are around the mean.
- [86, 87, 88, 86, 87, 85, 86] has a low standard deviation (0.9); [32, 111, 138, 28, 59, 77, 97] has a high one (37.85).
- Variance is the average of the squared differences from the mean (1432.24), and the standard deviation is its square root.
- NumPy provides std() and var(); ddof=1 gives the sample version.
- The band of one standard deviation either side of the mean holds a large share of the values (4 of 7 here, about 68% for normally distributed data).

## Conclusion

The mean describes the centre of the data, and the standard deviation describes how far the values typically are from it. Together they summarise a data set much better than the mean alone.
