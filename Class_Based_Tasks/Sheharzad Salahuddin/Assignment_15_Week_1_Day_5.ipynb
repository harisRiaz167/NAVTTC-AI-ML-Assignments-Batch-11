## Topic: Python — For Loops

**Objective:** Understand and practice for loops in Python.

**Reference:** https://www.geeksforgeeks.org/python-for-loops/

**Print numbers using a for loop**

```python
for i in range(1, 6):
    print(i)
```

**Explanation:** range(1, 6) generates numbers from 1 to 5. The loop prints each number.

**Iterate through a list**

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
```

**Explanation:** The loop visits each item in the list and prints its value.

```python
total = 0

for i in range(1, 6):
    total += i

print("Sum:", total)
```

**Explanation:** The loop adds the numbers 1, 2, 3, 4, and 5 to total.

**Use a for loop with an if statement**

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

**Explanation:** The % operator returns the remainder. If i % 2 == 0, the number is even.

Use a for loop with **else**

```python
for i in range(1, 4):
    print(i)
else:
    print("Loop completed")
```

## Observation

- A for loop repeats a block of code for each item in a sequence.
- The range() function generates a sequence of numbers.
- A for loop can iterate through lists and other iterable objects.
- An if statement can be used inside a loop to check conditions.
- The else block can execute after a loop finishes normally.
