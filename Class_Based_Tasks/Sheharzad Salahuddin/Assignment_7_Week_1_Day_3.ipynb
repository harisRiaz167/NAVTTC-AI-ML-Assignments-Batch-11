## Topic: Celsius to Fahrenheit Conversion

**Task:** Write a Python program to convert a temperature from Celsius to Fahrenheit.
**Example:** 37.5°C

**Explanation:** We use the formula:
$$
F = \left(C \times \frac{9}{5}\right) + 32
$$

- C represents Celsius.
- F represents Fahrenheit.
- Multiply Celsius by 9/5, then add 32.

```python
celsius = 37.5

fahrenheit = (celsius * 9/5) + 32

print(celsius, "°C is equal to", fahrenheit, "°F")
```
