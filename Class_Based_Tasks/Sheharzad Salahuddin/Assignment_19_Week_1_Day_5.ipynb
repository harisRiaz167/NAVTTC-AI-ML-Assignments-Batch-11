## Topic: Python — Various Looping Techniques

**Objective:** Practice different looping techniques in Python.

**Reference:** https://www.geeksforgeeks.org/looping-techniques-python/

**Use enumerate() to get index and value**

```python
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits):
    print(index, fruit)
```

**Explanation:** enumerate() provides both the index and the item during each iteration.

**Use zip() to loop through two lists**

```python
names = ["Ali", "Sara", "Ahmed"]
marks = [85, 90, 78]

for name, mark in zip(names, marks):
    print(name, mark)
```

**Explanation:** zip() pairs corresponding elements from the two lists.

**Use range() to print a sequence**

```python
for i in range(2, 11, 2):
    print(i)
```

**Explanation:** range(2, 11, 2) starts at 2, stops before 11, and increases by 2 each time.

**Iterate through a dictionary**

```python
student = {
    "name": "Ali",
    "age": 20,
    "grade": "A"
}

for key, value in student.items():
    print(key, ":", value)
```

**Explanation:** The items() method returns each dictionary key and its associated value.

**Use reversed() to reverse a sequence**

```python
numbers = [1, 2, 3, 4, 5]

for number in reversed(numbers):
    print(number)
```

**Explanation:** reversed() iterates over the sequence in reverse order without modifying the original list.

## Observation

- enumerate() returns the index and value of each item.
- zip() pairs elements from multiple iterables.
- range() generates a sequence of numbers.
- dictionary.items() provides key-value pairs.
- reversed() iterates through a sequence in reverse order.

## Conclusion

Python provides several looping techniques to iterate through sequences, process data, access indexes, and work with dictionaries efficiently.
