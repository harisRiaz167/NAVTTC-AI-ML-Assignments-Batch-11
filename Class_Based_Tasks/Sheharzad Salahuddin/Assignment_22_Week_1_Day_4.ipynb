**Diagram:** How a tuple compares with a list. The main difference is that a tuple cannot be changed.



**Create a tuple**

```python
numbers = (1, 2, -5)
print(numbers)

mixed = (2, "Hello", 3.4, True)
print(mixed)

nested = ("Python", [8, 4, 6], (1, 2, 3))
print(nested)

empty = ()
print(empty)
```

**Explanation:** A tuple is created by placing items inside parentheses ( ), separated by commas. Like a list, it can hold items of different types, including lists and other tuples.

**Create a tuple with the tuple() constructor**

```python
tuple_constructor = tuple(("Jack", "Maria", "David"))
print(tuple_constructor)

letters = tuple("abc")
print(letters)
```

**Explanation:** tuple() converts other iterables, such as lists and strings, into tuples.

**Create a tuple with one item**

```python
var1 = ("Hello")
print(type(var1))

var2 = ("Hello",)
print(type(var2))

var3 = "Hello",
print(type(var3))
```

**Explanation:** Parentheses alone do not make a tuple, so ('Hello') is just a string. A one-item tuple needs a trailing comma: ('Hello',). The parentheses are optional.

**Diagram:** Tuple items are accessed by index exactly like list items.



**Access items with an index**

```python
languages = ("Python", "Swift", "C++")

print(languages[0])
print(languages[2])
```

**Explanation:** Indexing starts at 0, so languages[0] is the first item and languages[2] is the third.

**Negative indexing**

```python
languages = ("Python", "Swift", "C++")

print(languages[-1])
print(languages[-3])
```

**Explanation:** Negative indexes count from the end, so -1 is the last item.

**Slicing a tuple**

```python
letters = ("p", "r", "o", "g", "r", "a", "m", "i", "z")

print(letters[1:4])
print(letters[:-7])
print(letters[7:])
print(letters[:])
```

**Explanation:** A slice returns a new tuple with the items from the start index up to, but not including, the stop index.

**Access items in a nested tuple**

```python
nested = ("Python", [8, 4, 6], (1, 2, 3))

print(nested[0][3])
print(nested[1][1])
print(nested[2][0])
```

**Explanation:** The first index selects an item of the outer tuple, and the second index selects an item inside it. nested[0][3] is the fourth character of 'Python'.

**Diagram:** A tuple's items cannot be changed after it is created.



**Tuples cannot be changed**

```python
cars = ("Toyota", "Honda", "Suzuki")

try:
    cars[0] = "BMW"
except TypeError as error:
    print("TypeError:", error)

cars = cars + ("Kia",)
print(cars)
```

**Explanation:** Assigning to an item raises a TypeError because tuples are immutable. To 'add' an item, a new tuple is created by joining the old one with another tuple, and the variable is assigned to it.

**A mutable item inside a tuple can change**

```python
data = (1, 2, [3, 4])

data[2][0] = 30
print(data)
```

**Explanation:** The tuple itself cannot change, but if one of its items is a mutable object such as a list, that list can still be changed.

**Delete a tuple**

```python
animals = ("dog", "cat", "rat")

try:
    del animals[0]
except TypeError as error:
    print("TypeError:", error)

del animals

try:
    print(animals)
except NameError as error:
    print("NameError:", error)
```

**Explanation:** Single items cannot be deleted from a tuple, but the whole tuple can be deleted with del. After that, using the name raises a NameError.

**Tuple methods: count() and index()**

```python
my_tuple = ("a", "p", "p", "l", "e")

print(my_tuple.count("p"))
print(my_tuple.index("l"))
```

**Explanation:** Tuples have only two methods. count() returns how many times a value appears, and index() returns the position of its first occurrence.

**Tuple length**

```python
cars = ("BMW", "Tesla", "Ford", "Toyota")

print("Total items:", len(cars))
```

**Explanation:** len() returns the number of items in a tuple.

**Iterate through a tuple**

```python
fruits = ("apple", "banana", "orange")

for fruit in fruits:
    print(fruit)
```

**Explanation:** A for loop visits each item of the tuple in order.

**Check whether an item exists**

```python
colors = ("red", "orange", "blue")

print("yellow" in colors)
print("red" in colors)
```

**Explanation:** The in operator returns True if the item is in the tuple and False if it is not.

**Diagram:** Packing puts several values into one tuple, and unpacking assigns them to separate variables.



**Tuple packing and unpacking**

```python
student = "Ali", 20, "AI"
print(student)

name, age, course = student
print("Name:", name)
print("Age:", age)
print("Course:", course)
```

**Explanation:** Writing comma-separated values packs them into a tuple. Writing several variables on the left of = unpacks the tuple into them; the number of variables must match the number of items.

**Join and repeat tuples**

```python
t1 = (1, 2, 3)
t2 = (4, 5)

print(t1 + t2)
print(t2 * 3)
```

**Explanation:** + joins two tuples into a new one, and * repeats a tuple a given number of times.

**Convert between tuples and lists**

```python
fruits = ("apple", "banana", "cherry")

fruit_list = list(fruits)
fruit_list.append("mango")

fruits = tuple(fruit_list)
print(fruits)
```

**Explanation:** To change a tuple's items, a common pattern is to convert it to a list, change the list, and convert it back to a tuple.

## Observation

- A tuple stores several items in one variable and is written with parentheses.
- Tuples are ordered and allow duplicates, but they are immutable: items cannot be changed, added or removed.
- A tuple with one item needs a trailing comma.
- Items are accessed by positive or negative index and by slicing, just like lists.
- Tuples have only two methods: count() and index().
- A whole tuple can be deleted with del, and tuples can be joined, repeated, packed and unpacked.

## Conclusion

A tuple is like a list that cannot be changed. This makes it safer for data that should stay fixed, and slightly faster than a list to loop through.
