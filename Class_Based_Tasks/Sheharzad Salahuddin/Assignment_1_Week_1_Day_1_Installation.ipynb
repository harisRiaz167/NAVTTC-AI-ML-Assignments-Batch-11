**Objective:** Install and verify Anaconda3, PyTorch, TensorFlow 2.0, Visual Studio Code (VS Code), and PyCharm.

## 1. Install Anaconda3

1. Visit the official Anaconda Distribution page: https://www.anaconda.com/download
2. Download the installer for your operating system.
3. Run the installer and follow the setup instructions.
4. Open **Anaconda Prompt** (Windows) or a terminal where Conda is available.
5. Verify the installation in the next cell.

```python
# Run this cell inside Jupyter after Anaconda is installed.
# If Conda is not available from the notebook, run `conda --version`
# in Anaconda Prompt / your terminal instead.

import sys
print("Python executable:", sys.executable)
print("Python version:", sys.version)
```

## 2. Create a separate environment for TensorFlow 2.0

Run the following commands in **Anaconda Prompt / terminal**, not in a regular Python cell. This environment is intended for the legacy TensorFlow 2.0 assignment.

```python
conda create -n tf20 python=3.7 -y
conda activate tf20
python -m pip install --upgrade pip
python -m pip install tensorflow==2.0.0
```

If the exact TensorFlow 2.0.0 package is unavailable for your OS/Python combination, record the error and consult the official installation documentation: https://www.tensorflow.org/install. Do not install an arbitrary newer version if your instructor specifically requires TensorFlow 2.0; ask whether a newer compatible version is acceptable.

## 3. Install PyTorch

For the simplest CPU-only installation in the same environment, run this in the activated `tf20` environment. For GPU/CUDA support, use the official PyTorch selector to choose the correct command for your hardware: https://pytorch.org/get-started/locally/.

```python
python -m pip install torch torchvision torchaudio
```

## 4. Verify PyTorch and TensorFlow

Run this cell using the Python kernel associated with the environment where the packages were installed. If the notebook uses a different environment, register the environment as a Jupyter kernel first:

`python -m pip install ipykernel`
`python -m ipykernel install --user --name tf20 --display-name "Python (tf20)"`

Then select **Python (tf20)** as the notebook kernel.

```python
import sys
print("Python:", sys.version)

try:
    import torch
    print("PyTorch version:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())
except Exception as e:
    print("PyTorch import failed:", repr(e))

try:
    import tensorflow as tf
    print("TensorFlow version:", tf.__version__)
except Exception as e:
    print("TensorFlow import failed:", repr(e))
```
