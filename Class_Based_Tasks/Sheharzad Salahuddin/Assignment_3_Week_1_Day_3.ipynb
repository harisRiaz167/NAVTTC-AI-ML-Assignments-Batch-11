**Task:** Write a Python program to add two numbers (num1 = 1.5 and num2 = 6.3) and display the sum.

```python
num1 = 1.5
num2 = 6.3

total = num1 + num2

print("The sum is:", total)
```

What does each line do?
- num1 = 1.5 — stores the first number.
- num2 = 6.3 — stores the second number.
- total = num1 + num2 — adds both numbers and stores the result.
- print() — displays the result on the screen.

The sum is: 7.8

## Explanation

The program stores two decimal numbers in variables named `num1` and `num2`.
It adds them using the `+` operator and stores the result in `total`.
Finally, the `print()` function displays the sum, which is 7.8.
