**View the first 10 rows with head()**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.head(10))
```

**Explanation:** One of the most used methods for a quick overview is head(). It returns the column headers and the given number of rows from the top.

**View the first 5 rows**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.head())
```

**Explanation:** If no number is given, head() returns the top 5 rows.

**View the last 5 rows with tail()**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.tail())
```

**Explanation:** tail() returns the column headers and the given number of rows from the bottom, 5 by default.

**Diagram:** info() summarises the rows, columns, non-null counts and data types.



**Get information about the data with info()**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

df.info()
```

**Explanation:** info() shows that there are 169 rows and 4 columns, the name and data type of each column, and how many non-null values each column has.

**Count the empty values**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.isnull().sum())
print(df[df["Calories"].isnull()])
```

**Explanation:** The Calories column has 164 non-null values out of 169, so 5 rows have no Calories value. isnull().sum() counts the empty cells in each column, and filtering shows exactly which rows they are. Empty values should be handled when cleaning the data.

**Summary statistics with describe()**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.describe().round(2))
```

**Explanation:** describe() gives the count, mean, standard deviation, minimum, quartiles and maximum of each numeric column. For example, the average workout lasted about 64 minutes.

**Value counts and grouping**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df["Duration"].value_counts().head())
print(df.groupby("Duration")["Calories"].mean().round(1).head(8))
```

**Explanation:** value_counts() shows how often each workout length appears, and groupby() gives the average calories for each duration.

**Visual overview with histograms**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data_full.csv")

df.hist(figsize=(10, 6), bins=20, color="#4C78A8")
plt.tight_layout()
plt.show()
```

**Explanation:** Histograms show how the values of each column are spread out. Most workouts last 45 or 60 minutes, and pulse values form a bell-like shape.

## Observation

- head() and tail() show the first and last rows; both default to 5.
- info() shows the number of rows and columns, the data types and the non-null count of each column.
- The Calories column has 5 empty values, which isnull().sum() confirms.
- describe() summarises each numeric column with statistics.
- value_counts(), groupby() and histograms give further insight into the data.

## Conclusion

Analysing a DataFrame starts with a quick overview: looking at a few rows, checking the structure with info() and summarising with describe(). This reveals problems such as missing values before any deeper analysis.
