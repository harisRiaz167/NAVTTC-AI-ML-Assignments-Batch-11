**Join two 1-D arrays with concatenate()**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))

print(arr)
```

**Explanation:** concatenate() joins a sequence of arrays, passed as a tuple. For 1-D arrays, the second array is added after the first.

**Diagram:** The axis argument decides whether 2-D arrays are joined as extra rows or extra columns.



**Join two 2-D arrays along rows (axis=0)**

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=0)

print(arr)
print(arr.shape)
```

**Explanation:** With axis=0, which is the default, the rows of arr2 are placed below the rows of arr1.

**Join two 2-D arrays along columns (axis=1)**

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=1)

print(arr)
print(arr.shape)
```

**Explanation:** With axis=1, the columns of arr2 are placed to the right of arr1, so each row gets longer.

**Arrays must have matching shapes**

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4]])
arr3 = np.array([[5, 6, 7]])

try:
    np.concatenate((arr1, arr3), axis=0)
except ValueError as error:
    print("ValueError:", error)
```

**Explanation:** Arrays can only be joined if their other dimensions match. arr1 rows have 2 columns and arr3 has 3, so joining them as rows fails.

**Diagram:** The stack helper functions join the same two arrays in different directions.



**Join arrays with stack()**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.stack((arr1, arr2), axis=1)

print(arr)
```

**Explanation:** stack() joins arrays along a new axis, so the result has one more dimension. With axis=1, matching elements are paired: [1 4], [2 5], [3 6].

**Stack along rows with hstack()**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.hstack((arr1, arr2))

print(arr)
```

**Explanation:** hstack() stacks arrays horizontally. For 1-D arrays, this gives one longer 1-D array.

**Stack along columns with vstack()**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.vstack((arr1, arr2))

print(arr)
```

**Explanation:** vstack() stacks arrays vertically, so each 1-D array becomes a row of a 2-D array.

**Stack along height (depth) with dstack()**

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.dstack((arr1, arr2))

print(arr)
print(arr.shape)
```

**Explanation:** dstack() stacks arrays along the third axis (depth), giving a 3-D array with shape (1, 3, 2).

**Join more than two arrays**

```python
import numpy as np

a = np.array([1, 2])
b = np.array([3, 4])
c = np.array([5, 6])

print(np.concatenate((a, b, c)))
print(np.vstack((a, b, c)))
```

**Explanation:** Any number of arrays can be joined at once by putting them all in the tuple.

## Observation

- concatenate() joins arrays along an existing axis; axis=0 adds rows and axis=1 adds columns.
- The arrays must match in every dimension except the one they are joined along.
- stack() joins arrays along a new axis, adding a dimension.
- hstack() joins horizontally, vstack() vertically and dstack() along the depth.
- Several arrays can be joined in one call.

## Conclusion

Joining combines separate arrays into one, for example to merge datasets or add new rows or columns of features. Choosing the right function and axis controls exactly how the data is combined.
