**Generate a random integer**

```python
from numpy import random

x = random.randint(100)

print(x)
```

**Explanation:** randint(100) returns a random whole number from 0 up to, but not including, 100.

**Generate a random float**

```python
from numpy import random

x = random.rand()

print(x)
```

**Explanation:** rand() returns a random float between 0 and 1.

**Generate an array of random integers**

```python
from numpy import random

x = random.randint(100, size=(5))
print(x)

y = random.randint(100, size=(3, 5))
print(y)
```

**Explanation:** The size argument sets the shape of the result: size=5 gives a 1-D array of 5 integers, and size=(3, 5) gives a 2-D array with 3 rows of 5 integers.

**Generate an array of random floats**

```python
from numpy import random

x = random.rand(5)
print(x)

y = random.rand(3, 5)
print(y)
```

**Explanation:** rand() takes the shape directly as arguments. rand(5) gives 5 floats and rand(3, 5) gives a 3 × 5 array of floats between 0 and 1.

**Pick a random value from an array**

```python
from numpy import random

x = random.choice([3, 5, 7, 9])

print(x)
```

**Explanation:** choice() returns one randomly selected value from the given array.

**Pick an array of random values from an array**

```python
from numpy import random

x = random.choice([3, 5, 7, 9], size=(3, 5))

print(x)
```

**Explanation:** choice() with size returns an array of the given shape, filled with values picked from the list. Values can repeat.

**Pick values with given probabilities**

```python
import numpy as np
from numpy import random

x = random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=100)

values, counts = np.unique(x, return_counts=True)
print(dict(zip(values.tolist(), counts.tolist())))
```

**Explanation:** p sets the probability of each value. The probabilities must add up to 1. Over 100 picks, 7 appears about 60 times, 5 about 30, 3 about 10 and 9 never, because its probability is 0.

**Diagram:** Using the same seed gives the same sequence of numbers every time.



**Make random results repeatable with a seed**

```python
import numpy as np

np.random.seed(42)
print(np.random.randint(100, size=5))

np.random.seed(42)
print(np.random.randint(100, size=5))

rng = np.random.default_rng(seed=7)
print(rng.integers(0, 100, size=5))
```

**Explanation:** Random numbers on a computer are pseudo-random: an algorithm creates them from a starting value. Setting the same seed gives the same numbers, which is useful for repeatable experiments. default_rng() is NumPy's newer generator and takes a seed too.

**Shuffle and permutation**

```python
import numpy as np
from numpy import random

np.random.seed(1)
arr = np.array([1, 2, 3, 4, 5])

print(random.permutation(arr))
print(arr)

random.shuffle(arr)
print(arr)
```

**Explanation:** permutation() returns a reordered copy and leaves the original unchanged. shuffle() reorders the array itself.

**Random numbers from distributions**

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
normal = np.random.normal(loc=50, scale=10, size=1000)
uniform = np.random.uniform(0, 100, size=1000)

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].hist(normal, bins=30, color="#4C78A8")
axes[0].set_title("Normal distribution (mean 50, std 10)")
axes[1].hist(uniform, bins=30, color="#54A24B")
axes[1].set_title("Uniform distribution (0 to 100)")
plt.tight_layout()
plt.show()
```

**Explanation:** normal() creates values that cluster around a mean in a bell shape, while uniform() spreads values evenly across a range. The histograms show the difference.

## Observation

- randint() creates random integers and rand() creates random floats between 0 and 1.
- The size argument (or the shape passed to rand()) creates arrays of random values.
- choice() picks values from an array, and p sets the probability of each value.
- Computer random numbers are pseudo-random; seed() makes them repeatable.
- permutation() returns a shuffled copy, while shuffle() changes the array itself.
- Functions such as normal() and uniform() generate values that follow a distribution.
