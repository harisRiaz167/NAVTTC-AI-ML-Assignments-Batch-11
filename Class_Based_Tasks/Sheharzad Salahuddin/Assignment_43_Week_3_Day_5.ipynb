**Diagram:** Every value in a Series has a label, either a default number or one you choose.



**Create a simple Series from a list**

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
```

**Explanation:** A Series is like a column in a table: a one-dimensional array that can hold any type of data. The numbers on the left are the labels.

**Access a value by its label**

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar[0])
```

**Explanation:** If no labels are given, the values are labelled with their index number, starting at 0. myvar[0] returns the first value.

**Create your own labels**

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a, index=["x", "y", "z"])

print(myvar)
```

**Explanation:** The index argument gives each value a name.

**Access an item by its custom label**

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a, index=["x", "y", "z"])

print(myvar["y"])
```

**Explanation:** Once labels are set, a value can be accessed by its label.

**Diagram:** Creating a Series from a dictionary, and keeping only selected keys.



**Create a Series from a dictionary**

```python
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

**Explanation:** A dictionary can also be used to create a Series. The keys become the labels.

**Select only some items from the dictionary**

```python
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index=["day1", "day2"])

print(myvar)
```

**Explanation:** The index argument chooses which keys to include, so only day1 and day2 are kept.

**Series operations**

```python
import pandas as pd

calories = pd.Series({"day1": 420, "day2": 380, "day3": 390})

print("Total:", calories.sum())
print("Average:", calories.mean())
print(calories[calories > 385])
print(calories + 10)
```

**Explanation:** A Series supports calculations such as sum() and mean(), filtering with a condition, and arithmetic on every value at once.

**DataFrames are made of Series**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)
print(type(myvar["calories"]))
```

**Explanation:** Data sets in Pandas are usually multi-dimensional tables called DataFrames. A Series is like a column, and a DataFrame is the whole table; selecting one column gives back a Series.

## Observation

- A Pandas Series is a one-dimensional array of values, like one column of a table.
- Each value has a label: its position by default, or a name set with index.
- Values can be accessed by label.
- A dictionary can be turned into a Series, with keys as labels; index selects specific keys.
- A Series supports calculations, filtering and arithmetic, and each column of a DataFrame is a Series.

## Conclusion

The Series is the basic building block of Pandas. Labelled values make data easy to look up by name, and the same operations carry over to the columns of a DataFrame.
