**Create a DataFrame from two Series**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data)

print(df)
```

**Explanation:** A DataFrame is a two-dimensional data structure, like a table with rows and columns. Each key of the dictionary becomes a column.

**DataFrame properties**

```python
import pandas as pd

df = pd.DataFrame({"calories": [420, 380, 390], "duration": [50, 40, 45]})

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Index:", df.index.tolist())
print(df.dtypes)
```

**Explanation:** shape gives (rows, columns), columns and index list the labels, and dtypes shows the data type of each column.

**Diagram:** loc selects rows by their index label; the highlighted rows are the ones returned.



**Locate a row**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data)

print(df.loc[0])
```

**Explanation:** The loc attribute returns one or more rows. loc[0] returns the first row as a Pandas Series.

**Return several rows**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data)

print(df.loc[[0, 1]])
```

**Explanation:** A list of indexes returns those rows. With [ ] inside loc, the result is a DataFrame.

**Named indexes**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df)
```

**Explanation:** The index argument gives the rows names instead of numbers.

**Locate named indexes**

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df.loc["day2"])
```

**Explanation:** With named indexes, loc uses the name to find the row.

**Select columns and single values**

```python
import pandas as pd

df = pd.DataFrame({"calories": [420, 380, 390], "duration": [50, 40, 45]},
                  index=["day1", "day2", "day3"])

print(df["calories"])
print(df[["duration"]])
print(df.loc["day3", "duration"])
print(df.iloc[0, 0])
```

**Explanation:** A column is selected with its name; double brackets keep it as a DataFrame. loc[row, column] gets one value by label, and iloc[row, column] gets it by position.

**Add a column and filter rows**

```python
import pandas as pd

df = pd.DataFrame({"calories": [420, 380, 390], "duration": [50, 40, 45]},
                  index=["day1", "day2", "day3"])

df["cal_per_min"] = (df["calories"] / df["duration"]).round(2)
print(df)
print(df[df["duration"] >= 45])
```

**Explanation:** New columns can be calculated from existing ones, and a condition keeps only the matching rows.

**Load files into a DataFrame**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df)
```

**Explanation:** Data stored in a file can be loaded straight into a DataFrame with read_csv(). This uses data_full.csv, the W3Schools workout data set.

## Observation

- A DataFrame is a two-dimensional table with an index, named columns and values.
- It can be created from a dictionary of lists, or loaded from a file with read_csv().
- loc[0] returns one row as a Series; loc[[0, 1]] returns several rows as a DataFrame.
- The index argument gives rows names, which loc can then use.
- Columns are selected by name, and loc / iloc get single values by label or position.

## Conclusion

The DataFrame is the main object in Pandas. It holds a whole table of data and makes it easy to select, filter and add rows and columns.
