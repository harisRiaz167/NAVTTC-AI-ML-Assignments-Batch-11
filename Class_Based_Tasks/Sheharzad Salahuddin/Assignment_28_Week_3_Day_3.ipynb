**Import NumPy and check the version**

```python
import numpy as np

print(np.__version__)
```

**Explanation:** NumPy is imported with the alias np, which is the common convention. __version__ shows the installed version.

**Diagram:** A list or tuple passed to np.array() becomes a NumPy ndarray.



**Create a NumPy ndarray object**

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(arr)
print(type(arr))
```

**Explanation:** np.array() creates an array object of type ndarray. Here a Python list is passed to it.

**Create an array from a tuple**

```python
import numpy as np

arr = np.array((1, 2, 3, 4, 5))

print(arr)
```

**Explanation:** Any array-like object, such as a tuple, can be passed to np.array() and converted into an ndarray.

**Diagram:** An array's number of dimensions is how many levels of nesting it has.



**0-D arrays**

```python
import numpy as np

arr = np.array(42)

print(arr)
```

**Explanation:** A 0-D array, or scalar, holds a single value. Each value in an array is a 0-D array.

**1-D arrays**

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(arr)
```

**Explanation:** A 1-D array has 0-D arrays as its elements. It is the most common type of array.

**2-D arrays**

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])

print(arr)
```

**Explanation:** A 2-D array has 1-D arrays as its elements. It is often used to represent a matrix of rows and columns.

**3-D arrays**

```python
import numpy as np

arr = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(arr)
```

**Explanation:** A 3-D array has 2-D arrays (matrices) as its elements. This one holds two 2-D arrays, each with two rows of three values.

**Check the number of dimensions**

```python
import numpy as np

a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

**Explanation:** The ndim attribute returns the number of dimensions of an array.

**Higher-dimensional arrays with ndmin**

```python
import numpy as np

arr = np.array([1, 2, 3, 4], ndmin=5)

print(arr)
print("number of dimensions :", arr.ndim)
```

**Explanation:** The ndmin argument sets the minimum number of dimensions. The innermost dimension holds the 4 values, and the other 4 dimensions each hold one element.

**Other ways to create arrays**

```python
import numpy as np

print(np.zeros(4))
print(np.ones((2, 3)))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.full((2, 2), 7))
```

**Explanation:** NumPy also has functions that build arrays directly: zeros() and ones() fill an array with 0s or 1s, arange() works like range(), linspace() makes evenly spaced values, and full() fills an array with one value.

## Observation

- NumPy is imported as np, and np.array() creates an ndarray.
- Lists, tuples and other array-like objects can be converted into arrays.
- Arrays can be 0-D (a scalar), 1-D, 2-D, 3-D or higher.
- ndim returns the number of dimensions, and ndmin sets a minimum number of dimensions.
- Functions such as zeros(), ones(), arange() and linspace() create arrays directly.

## Conclusion

The NumPy ndarray is the main object in NumPy. It stores values of the same type in one or more dimensions, which makes numeric work much faster and simpler than with Python lists.
