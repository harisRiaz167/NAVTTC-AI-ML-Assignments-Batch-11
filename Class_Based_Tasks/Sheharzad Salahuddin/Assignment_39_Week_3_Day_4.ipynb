**Diagram:** Sorting puts elements in an ordered sequence.



**Sort a numeric array**

```python
import numpy as np

arr = np.array([3, 2, 0, 1])

print(np.sort(arr))
print(arr)
```

**Explanation:** np.sort() returns a sorted copy of the array in ascending order. The second print shows the original array is unchanged.

**Sort a string array alphabetically**

```python
import numpy as np

arr = np.array(["banana", "cherry", "apple"])

print(np.sort(arr))
```

**Explanation:** Strings are sorted alphabetically.

**Sort a boolean array**

```python
import numpy as np

arr = np.array([True, False, True])

print(np.sort(arr))
```

**Explanation:** Booleans are sorted with False first, because False counts as 0 and True as 1.

**Diagram:** By default each row of a 2-D array is sorted; axis=0 sorts each column instead.



**Sort a 2-D array**

```python
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])

print(np.sort(arr))
```

**Explanation:** For a 2-D array, np.sort() sorts each row separately.

**Sort the columns of a 2-D array**

```python
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])

print(np.sort(arr, axis=0))
print(np.sort(arr, axis=None))
```

**Explanation:** axis=0 sorts each column. axis=None flattens the array first and returns all the values sorted in one 1-D array.

**Sort in descending order**

```python
import numpy as np

arr = np.array([3, 2, 0, 1])

print(np.sort(arr)[::-1])
print(-np.sort(-arr))
```

**Explanation:** np.sort() only sorts in ascending order. Reversing the result with [::-1], or sorting the negated values, gives descending order.

**Sort in place with the array's sort() method**

```python
import numpy as np

arr = np.array([3, 2, 0, 1])

arr.sort()
print(arr)
```

**Explanation:** The ndarray.sort() method sorts the array itself instead of returning a copy.

**Get the sorting order with argsort()**

```python
import numpy as np

marks = np.array([72, 95, 60, 88])
names = np.array(["Ali", "Sara", "Ahmed", "Zara"])

order = np.argsort(marks)
print(order)
print(names[order])
print(names[order[::-1]])
```

**Explanation:** argsort() returns the indexes that would sort the array. These indexes can reorder a related array, for example to list students from the lowest to the highest mark, or reversed for a ranking.

## Observation

- np.sort() returns a sorted copy and leaves the original array unchanged.
- Numbers sort in ascending order, strings alphabetically and booleans with False first.
- In a 2-D array each row is sorted by default; axis=0 sorts columns and axis=None sorts every value.
- Descending order is obtained by reversing the sorted array.
- arr.sort() sorts in place, and argsort() returns the indexes that would sort the array.

## Conclusion

Sorting arranges data in order, which makes it easier to read, search and rank. NumPy's sort functions work on numbers, strings and booleans and on any number of dimensions.
