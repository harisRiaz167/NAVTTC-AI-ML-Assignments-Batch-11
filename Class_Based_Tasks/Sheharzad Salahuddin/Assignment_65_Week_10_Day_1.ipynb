## 1. Cloud concepts

**Cloud computing** is the delivery of computing services (servers, storage, databases, networking, software, analytics and AI) over the internet. Key ideas:

- **Consumption-based model** – pay only for what you use (operational expenditure, OpEx) instead of buying hardware up front (capital expenditure, CapEx).
- **Benefits** – high availability, scalability (scale up/out), elasticity (automatic scaling), reliability, predictability, security, governance and manageability.
- **Deployment models** – *public cloud* (Azure), *private cloud* (your own data centre) and *hybrid cloud* (both together).

**Diagram:** IaaS, PaaS and SaaS differ in how much the customer has to manage.



**Shared responsibility model:** the more managed the service, the less you are responsible for. Whatever the model, you are always responsible for your **data**, **devices** and **accounts/identities**.

## 2. Azure architecture and services

- **Regions** are geographical areas with one or more data centres; **availability zones** are physically separate data centres within a region; **region pairs** support disaster recovery.
- **Compute** – Virtual Machines, Virtual Machine Scale Sets, App Service, Azure Functions (serverless), Azure Container Instances and Azure Kubernetes Service.
- **Networking** – Virtual Network (VNet), VPN Gateway, ExpressRoute, Azure DNS, Load Balancer.
- **Storage** – Storage accounts with Blob, File, Queue, Table and Disk storage; access tiers (hot, cool, cold, archive); redundancy options (LRS, ZRS, GRS, GZRS).
- **Identity** – Microsoft Entra ID (formerly Azure Active Directory), single sign-on, multi-factor authentication, conditional access, Zero Trust.

**Diagram:** Management groups contain subscriptions, which contain resource groups, which contain resources.



## 3. Management and governance

- **Cost management** – the Pricing Calculator and TCO Calculator estimate costs; Cost Management + Billing tracks spending; **tags** label resources for reporting.
- **Governance** – **Azure Policy** enforces rules (for example, allowed regions); **resource locks** prevent accidental deletion; **Microsoft Purview** governs data.
- **Tools** – Azure portal, Azure Cloud Shell, Azure CLI, Azure PowerShell, Azure Arc and **ARM templates / Bicep** for infrastructure as code.
- **Monitoring** – Azure Advisor (recommendations), Azure Service Health and **Azure Monitor** (Log Analytics, alerts, Application Insights).

## 4. Azure AI fundamentals

Because this course is about AI, the **AI-900** material is especially relevant:

- **Machine learning** on Azure: regression, classification and clustering with **Azure Machine Learning**, Automated ML and the visual Designer.
- **Computer vision**: image analysis, object detection, OCR and face analysis with **Azure AI Vision** and **Document Intelligence**.
- **Natural language processing**: key phrases, sentiment, entity recognition and language detection with **Azure AI Language**; speech-to-text and text-to-speech with **Speech**; translation with **Translator**.
- **Generative AI**: large language models through **Azure OpenAI Service** and **Azure AI Foundry**, copilots and prompt engineering.
- **Responsible AI** principles: fairness, reliability and safety, privacy and security, inclusiveness, transparency and accountability.

**Diagram:** The Azure AI services grouped by the type of problem they solve.
