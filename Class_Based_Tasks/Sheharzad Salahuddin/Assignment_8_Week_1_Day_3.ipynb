## Topic: Area of a Triangle

**Task:** Calculate the area of a triangle using its three sides and the semi-perimeter formula.

**Example:** a = 5, b = 6, c = 7

```python
a = 5
b = 6
c = 7
```

**Calculate the semi-perimeter:**
The semi-perimeter is half the perimeter of the triangle.
$$
s = \frac{a+b+c}{2}
$$

```python
s = (a + b + c) / 2

print("Semi-perimeter:", s)
```

**Calculate the area using Heron's formula**

$$
A = \sqrt{s(s-a)(s-b)(s-c)}
$$

```python

import math

a = 5
b = 6
c = 7

s = (a + b + c) / 2

area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Semi-perimeter:", s)
print("Area of the triangle:", area)
```
