**Create a list**

```python
ages = [19, 26, 29]
print(ages)

student = ["Jack", 32, "Computer Science", [2, 4]]
print(student)

empty_list = []
print(empty_list)
```

**Explanation:** A list is created by placing items inside square brackets [ ], separated by commas. A list can hold items of different types, including another list, or it can be empty.

**Create a list with the list() constructor**

```python
text = "axz"
result = list(text)
print(result)

numbers = list((1, 2, 3))
print(numbers)
```

**Explanation:** list() converts other iterables, such as strings and tuples, into lists.

**List characteristics**

```python
numbers = [1, 2, 2, 3]
print(numbers)

numbers[0] = 10
print(numbers)
print(len(numbers))
```

**Explanation:** Lists are ordered (items keep their position), mutable (items can be changed) and allow duplicate values. len() returns the number of items.

**Diagram:** Each item has a positive index counted from the start and a negative index counted from the end.



**Access items with an index**

```python
languages = ["Python", "Swift", "C++"]

print("languages[0] =", languages[0])
print("languages[2] =", languages[2])
```

**Explanation:** Indexing starts at 0, so languages[0] is the first item and languages[2] is the third.

**Negative indexing**

```python
languages = ["Python", "Swift", "C++"]

print("languages[-1] =", languages[-1])
print("languages[-3] =", languages[-3])
```

**Explanation:** Negative indexes count from the end, so -1 is the last item and -3 is the first item here.

**Diagram:** A slice takes the items from the start index up to, but not including, the stop index.



**Slicing a list**

```python
my_list = ["p", "r", "o", "g", "r", "a", "m"]

print("my_list[2:5] =", my_list[2:5])
print("my_list[5:] =", my_list[5:])
print("my_list[:] =", my_list[:])
print("my_list[::2] =", my_list[::2])
```

**Explanation:** my_list[2:5] returns items at indexes 2, 3 and 4. Leaving out the start or stop means 'from the beginning' or 'to the end', and [:] copies the whole list. A third number is the step, so [::2] takes every second item.

**Diagram:** append(), extend() and insert() add items in different ways. The highlighted boxes are the new items.



**Add an item with append()**

```python
fruits = ["apple", "banana", "orange"]
print("Original List:", fruits)

fruits.append("cherry")
print("Updated List:", fruits)
```

**Explanation:** append() adds one item to the end of the list.

**Add an item at a position with insert()**

```python
fruits = ["apple", "banana", "orange"]
print("Original List:", fruits)

fruits.insert(2, "cherry")
print("Updated List:", fruits)
```

**Explanation:** insert(2, 'cherry') puts 'cherry' at index 2, and the later items move one place to the right.

**Add items from another list with extend()**

```python
numbers = [1, 3, 5]
print("Numbers:", numbers)

even_numbers = [2, 4, 6]
numbers.extend(even_numbers)
print("Updated Numbers:", numbers)
```

**Explanation:** extend() adds every item of another list (or other iterable) to the end of the list.

**Change list items**

```python
colors = ["Red", "Black", "Green"]
print("Original List:", colors)

colors[2] = "Blue"
print("Changed List:", colors)
```

**Explanation:** Because lists are mutable, an item can be replaced by assigning a new value to its index.

**Remove items with remove() and del**

```python
numbers = [2, 4, 7, 9]
numbers.remove(4)
print(numbers)

names = ["John", "Eva", "Laura", "Nick", "Jack"]
del names[1]
print(names)

del names[1:3]
print(names)
```

**Explanation:** remove() deletes the first item with the given value. del deletes items by index and can also delete a whole slice.

**Remove items with pop() and clear()**

```python
cities = ["Lahore", "Karachi", "Islamabad", "Quetta"]

last = cities.pop()
print("Popped:", last, "->", cities)

first = cities.pop(0)
print("Popped:", first, "->", cities)

cities.clear()
print("After clear():", cities)
```

**Explanation:** pop() removes and returns the last item, or the item at the given index. clear() removes every item and leaves an empty list.

**Diagram:** The common list methods, grouped by what they do.



**Search, count, sort, reverse and copy**

```python
marks = [72, 85, 64, 85, 90]

print("index(85):", marks.index(85))
print("count(85):", marks.count(85))

marks.sort()
print("sort():", marks)

marks.reverse()
print("reverse():", marks)

backup = marks.copy()
backup.append(100)
print("original:", marks)
print("copy:", backup)
```

**Explanation:** index() returns the position of the first matching item, and count() returns how many times a value appears. sort() orders the list in place, reverse() reverses it, and copy() makes a separate list, so changing the copy does not change the original.

**Iterate through a list**

```python
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(fruit)
```

**Explanation:** A for loop visits each item of the list in order.

**Check whether an item exists**

```python
fruits = ["apple", "banana", "orange"]

print("orange" in fruits)
print("cherry" in fruits)
print("cherry" not in fruits)
```

**Explanation:** The in operator returns True if the item is in the list, and not in returns True if it is not.

**Nested lists**

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

print(matrix[1])
print(matrix[1][2])

for row in matrix:
    print(row)
```

**Explanation:** A list can contain other lists. matrix[1] is the second inner list, and matrix[1][2] is the third item inside it.

**List comprehension**

```python
numbers = [1, 2, 3, 4, 5]

squares = [n * n for n in numbers]
print(squares)

evens = [n for n in numbers if n % 2 == 0]
print(evens)
```

**Explanation:** A list comprehension builds a new list in one line from another iterable, optionally with a condition.

## Observation

- A list stores several items in one variable and is written with square brackets.
- Lists are ordered, mutable and allow duplicates, and they can hold items of different types.
- Items are accessed by positive or negative index, and slicing returns part of a list.
- append(), insert() and extend() add items, while remove(), pop(), del and clear() remove them.
- index(), count(), sort(), reverse() and copy() search, count, order and copy lists.
- Lists can be looped over, checked with in, nested and built with list comprehensions.

## Conclusion

The list is one of Python's most useful data types. It keeps items in order and can be changed easily, which makes it ideal for storing and processing collections of data.
