**Diagram:** The parts of a function definition.



**Define and call a function**

```python
def greet():
    print("Welcome to NAVTTC")

greet()
```

**Explanation:** A function is defined with the def keyword, a name, parentheses and a colon. The indented body runs only when the function is called by writing its name followed by ().

**Diagram:** Arguments go into the function, and the return value comes back to the caller.



**Function with parameters and a return value**

```python
def add(a, b):
    """Return the sum of a and b."""
    result = a + b
    return result

total = add(5, 3)
print("Sum:", total)
print(add.__doc__)
```

**Explanation:** a and b are parameters. The values 5 and 3 are the arguments passed when the function is called. return sends the result back to the caller. The text in triple quotes is the docstring, which describes the function and can be read with __doc__.

**Function with type hints**

```python
def add(num1: int, num2: int) -> int:
    """Add two numbers."""
    return num1 + num2

print(add(10, 20))
```

**Explanation:** Type hints such as num1: int and -> int show which types the function expects and returns. They make code easier to read but are not enforced by Python.

**Return multiple values**

```python
def calculate(a, b):
    return a + b, a - b, a * b

s, d, p = calculate(10, 4)
print("Sum:", s)
print("Difference:", d)
print("Product:", p)
```

**Explanation:** A function can return several values separated by commas. They are returned as a tuple and can be unpacked into separate variables.

**Check whether a number is even or odd**

```python
def even_odd(x):
    if x % 2 == 0:
        return "Even"
    else:
        return "Odd"

print(2, "is", even_odd(2))
print(3, "is", even_odd(3))
```

**Explanation:** The function can be reused for any number. It returns 'Even' when the remainder after dividing by 2 is 0, and 'Odd' otherwise.

**Diagram:** The different ways arguments can be passed to a function.



**Default arguments**

```python
def my_fun(x, y=50):
    print("x:", x)
    print("y:", y)

my_fun(10)
my_fun(10, 20)
```

**Explanation:** y has a default value of 50, which is used when no value is passed for it. Parameters with defaults must come after parameters without defaults.

**Keyword arguments**

```python
def student(firstname, lastname):
    print(firstname, lastname)

student(firstname="Ali", lastname="Khan")
student(lastname="Khan", firstname="Ali")
```

**Explanation:** With keyword arguments, each value is matched to a parameter by name, so the order does not matter.

**Positional arguments**

```python
def name_age(name, age):
    print("Hi, I am", name)
    print("My age is", age)

print("Case 1:")
name_age("Sara", 22)

print("Case 2:")
name_age(22, "Sara")
```

**Explanation:** Positional arguments are matched to parameters by their order. In Case 2 the order is swapped, so the name and age end up in the wrong places.

**Arbitrary arguments with *args**

```python
def total(*args):
    print("args:", args)
    return sum(args)

print(total(1, 2))
print(total(5, 10, 15, 20))
```

**Explanation:** *args collects any number of positional arguments into a tuple, so the same function works with any number of values.

**Arbitrary keyword arguments with **kwargs**

```python
def show_profile(**kwargs):
    for key, value in kwargs.items():
        print(key, "==", value)

show_profile(name="Ahmed", course="AI", city="Lahore")
```

**Explanation:** **kwargs collects any number of keyword arguments into a dictionary of names and values.

**Nested (inner) functions**

```python
def outer():
    s = "I love Python"

    def inner():
        print(s)

    inner()

outer()
```

**Explanation:** A function defined inside another function is an inner function. It can use the variables of the outer function and can only be called from inside it.

**Anonymous functions (lambda)**

```python
def cube(x):
    return x * x * x

cube_lambda = lambda x: x * x * x

print(cube(7))
print(cube_lambda(7))
```

**Explanation:** lambda creates a small function without a name. It can hold only one expression, whose value is returned automatically. Both versions give the same result.

**Diagram:** What happens to the caller's variable when a function changes its argument.



**Pass by reference and pass by value**

```python
def change_list(x):
    x[0] = 20

lst = [10, 11, 12, 13]
change_list(lst)
print("List after the function:", lst)

def change_number(x):
    x = 20

num = 10
change_number(num)
print("Number after the function:", num)
```

**Explanation:** Python passes a reference to the object. A list is mutable, so changing x[0] changes the caller's list too. Assigning x = 20 only makes x point to a new object, so the caller's num stays 10.

**Diagram:** Each call waits for the smaller call below it, then the results return back up.



**Recursive function**

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print("factorial(4) =", factorial(4))
```

**Explanation:** A recursive function calls itself on a smaller problem. The base case (n is 0 or 1) stops the recursion, then each call multiplies its n by the result of the call below it.

## Observation

- A function is defined with def and runs only when it is called.
- Parameters receive the arguments passed in, and return sends a result back.
- A docstring describes what the function does, and type hints show the expected types.
- Arguments can be default, keyword, positional, *args (tuple) or **kwargs (dictionary).
- Inner functions are defined inside other functions, and lambda creates small anonymous functions.
- Changing a mutable argument inside a function affects the caller, but reassigning a parameter does not.
- A recursive function calls itself and needs a base case to stop.

## Conclusion

User-defined functions group code into reusable blocks with a name. They make programs shorter, easier to read and easier to maintain, because the same logic can be called many times with different inputs.
