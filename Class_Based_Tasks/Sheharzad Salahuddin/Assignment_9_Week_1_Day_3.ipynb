## Topic: Quadratic Equation

**Task:** Solve the quadratic equation ax² + bx + c = 0 using the cmath module.

**Example:** a = 1, b = 5, c = 6

Calculate the discriminant
The discriminant is:

$$
D = b^2 - 4ac
$$

**Calculate both roots**

$$x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$

```python

import cmath

a = 1
b = 5
c = 6

d = (b ** 2) - (4 * a * c)

root1 = (-b + cmath.sqrt(d)) / (2 * a)
root2 = (-b - cmath.sqrt(d)) / (2 * a)

print("Root 1:", root1)
print("Root 2:", root2)
```
