**Diagram:** A set keeps only unique items and has no fixed order.



**Create a set**

```python
student_id = {112, 114, 116, 118, 115}
print("Student ID:", student_id)

vowel_letters = {"a", "e", "i", "o", "u"}
print("Vowel Letters:", vowel_letters)

mixed_set = {"Hello", 101, -2, "Bye"}
print("Set of mixed data types:", mixed_set)
```

**Explanation:** A set is created by placing items inside curly braces { }, separated by commas. It can hold items of different types. The printed order may differ from the order written, because sets are unordered.

**Create an empty set**

```python
empty_set = set()
empty_dictionary = {}

print("Data type of empty_set:", type(empty_set))
print("Data type of empty_dictionary:", type(empty_dictionary))
```

**Explanation:** Empty curly braces { } create an empty dictionary, not a set. An empty set must be created with set().

**Duplicate items are removed**

```python
numbers = {2, 4, 6, 6, 2, 8}
print(numbers)

marks = [90, 85, 90, 70, 85]
unique_marks = set(marks)
print(unique_marks)
```

**Explanation:** A set cannot contain duplicates, so repeated values are kept only once. Converting a list to a set is a quick way to remove its duplicates.

**Set items must be immutable**

```python
valid = {1, "two", (3, 4)}
print(valid)

try:
    invalid = {1, 2, [3, 4]}
except TypeError as error:
    print("TypeError:", error)
```

**Explanation:** A set can contain immutable items such as numbers, strings and tuples, but not mutable items such as lists. Trying to add a list raises a TypeError.

**Sets have no index**

```python
fruits = {"apple", "banana", "mango"}

try:
    print(fruits[0])
except TypeError as error:
    print("TypeError:", error)
```

**Explanation:** Because sets are unordered, items cannot be accessed by index. Use a loop or the in operator instead.

**Diagram:** How items are added to and removed from a set.



**Add items with add()**

```python
numbers = {21, 34, 54, 12}
print("Initial Set:", numbers)

numbers.add(32)
print("Updated Set:", numbers)
```

**Explanation:** add() adds one item to the set. If the item is already there, nothing changes.

**Add several items with update()**

```python
companies = {"Lacoste", "Ralph Lauren"}
tech_companies = ["apple", "google", "apple"]

companies.update(tech_companies)
print(companies)
```

**Explanation:** update() adds every item from another iterable, such as a list, tuple or string. Duplicates such as the second 'apple' are ignored.

**Remove items with remove() and discard()**

```python
languages = {"Swift", "Java", "Python"}
print("Initial Set:", languages)

languages.discard("Java")
print("After discard('Java'):", languages)

languages.discard("C++")
print("After discard('C++'):", languages)

try:
    languages.remove("C++")
except KeyError as error:
    print("KeyError:", error)
```

**Explanation:** Both methods remove an item. discard() does nothing if the item is missing, while remove() raises a KeyError.

**Remove items with pop() and clear()**

```python
colors = {"red", "green", "blue"}

removed = colors.pop()
print("Removed:", removed)
print("Remaining:", colors)

colors.clear()
print("After clear():", colors)
```

**Explanation:** pop() removes and returns an arbitrary item, because a set has no 'last' item. clear() removes every item and leaves an empty set.

**Iterate over a set and find its length**

```python
fruits = {"Apple", "Peach", "Mango"}

for fruit in fruits:
    print(fruit)

print("Total elements:", len(fruits))
print("Mango" in fruits)
```

**Explanation:** A for loop visits each item once, in no guaranteed order. len() returns the number of items, and in checks whether an item is present.

**Built-in functions with sets**

```python
numbers = {4, 9, 1, 7}

print("max:", max(numbers))
print("min:", min(numbers))
print("sum:", sum(numbers))
print("sorted:", sorted(numbers))
print("all:", all(numbers))
print("any:", any(numbers))
```

**Explanation:** Functions such as max(), min(), sum() and sorted() work on sets. sorted() returns a sorted list. all() is True when every item is truthy and any() is True when at least one is.

**Diagram:** The shaded area in each Venn diagram is the result of the operation.



**Union**

```python
A = {1, 3, 5}
B = {1, 2, 3}

print("Union using |:", A | B)
print("Union using union():", A.union(B))
```

**Explanation:** The union contains every item that is in A, in B or in both. It can be written with | or the union() method.

**Intersection**

```python
A = {1, 3, 5}
B = {1, 2, 3}

print("Intersection using &:", A & B)
print("Intersection using intersection():", A.intersection(B))
```

**Explanation:** The intersection contains only the items that are in both A and B.

**Difference**

```python
A = {1, 3, 5}
B = {1, 2, 3}

print("Difference using -:", A - B)
print("Difference using difference():", A.difference(B))
print("B - A:", B - A)
```

**Explanation:** A - B contains the items that are in A but not in B. The order matters, so B - A gives a different result.

**Symmetric difference**

```python
A = {1, 3, 5}
B = {1, 2, 3}

print("Using ^:", A ^ B)
print("Using symmetric_difference():", A.symmetric_difference(B))
```

**Explanation:** The symmetric difference contains the items that are in A or B, but not in both.

**Check whether two sets are equal, subsets or disjoint**

```python
A = {1, 3, 5}
B = {3, 5, 1}
C = {1, 3}
D = {7, 8}

print("A == B:", A == B)
print("C is a subset of A:", C.issubset(A))
print("A is a superset of C:", A.issuperset(C))
print("A and D share no items:", A.isdisjoint(D))
```

**Explanation:** Two sets are equal when they have the same items, whatever the order. issubset(), issuperset() and isdisjoint() compare how two sets overlap.

**Frozen sets**

```python
frozen = frozenset([1, 2, 3, 2])
print(frozen)

try:
    frozen.add(4)
except AttributeError as error:
    print("AttributeError:", error)
```

**Explanation:** A frozenset is a set that cannot be changed after it is created, so it has no add() or remove() methods.

## Observation

- A set stores unique items inside curly braces and has no fixed order or index.
- An empty set must be created with set(), because { } creates a dictionary.
- Set items must be immutable, so lists cannot be stored in a set.
- add() and update() add items; remove(), discard(), pop() and clear() remove them.
- Union (|), intersection (&), difference (-) and symmetric difference (^) combine sets.
- A frozenset is a set that cannot be changed.

## Conclusion

Sets are useful whenever only unique values matter, such as removing duplicates, checking membership quickly, or comparing two groups of items with set operations.
