## Installation of Pandas

If Python and pip are already installed, Pandas is installed with:

```
pip install pandas
```

If that fails, a Python distribution that already includes Pandas, such as Anaconda or Spyder, can be used instead.

**Install Pandas from inside Jupyter (only if it is missing)**

````python
# Uncomment the next line to install Pandas from inside the notebook:
# %pip install pandas

import importlib.util
print("Pandas installed:", importlib.util.find_spec("pandas") is not None)
````

**Explanation:** %pip runs pip from inside Jupyter. The check shows whether Pandas is already available, so the install line can stay commented out.

**Import Pandas**

````python
import pandas

mydataset = {
    "cars": ["BMW", "Volvo", "Ford"],
    "passings": [3, 7, 2]
}

myvar = pandas.DataFrame(mydataset)

print(myvar)
````

**Explanation:** After installing, Pandas is imported with the import keyword, and its functions are used with the pandas. prefix.

**Diagram:** Each dictionary key becomes a column name and each list becomes that column's values.



**Import Pandas with the pd alias**

````python
import pandas as pd

mydataset = {
    "cars": ["BMW", "Volvo", "Ford"],
    "passings": [3, 7, 2]
}

myvar = pd.DataFrame(mydataset)

print(myvar)
````

**Explanation:** Pandas is usually imported with the alias pd, so pd.DataFrame() can be written instead of pandas.DataFrame().

**Check the Pandas version**

````python
import pandas as pd

print(pd.__version__)
````

**Explanation:** The version string is stored in the __version__ attribute.

**Show versions of Pandas and its dependencies**

````python
import pandas as pd
import numpy as np
import sys

print("Python:", sys.version.split()[0])
print("Pandas:", pd.__version__)
print("NumPy:", np.__version__)
````

**Explanation:** Pandas is built on NumPy, so it is useful to know both versions when following a tutorial or reporting a problem.

**A first look at the DataFrame**

````python
import pandas as pd

mydataset = {
    "cars": ["BMW", "Volvo", "Ford"],
    "passings": [3, 7, 2]
}

df = pd.DataFrame(mydataset)

print(df.shape)
print(df.columns.tolist())
print(df["passings"].sum())
print(df.sort_values("passings", ascending=False))
````

**Explanation:** Once data is in a DataFrame, its size, column names, totals and sorted views are available immediately.
