## 2. Navigation commands: `pwd`, `cd`, `ls`

- `pwd` — print the current working directory.
- `cd` — change directory.
- `ls` — list files and folders. `-l` shows details; `-a` includes hidden files.

```bash
pwd
cd ~
pwd
cd ~/linux_day2
ls
ls -la
```

## 3. Create files and folders: `mkdir`, `touch`

- `mkdir` creates a directory.
- `touch` creates an empty file if it does not exist, or updates its timestamp.

```bash
mkdir notes
touch notes/day2.txt
ls -l notes
```

## 4. Write and read files: `cat`, `head`, `more`, `less`, `reading`

- `cat` prints file contents and can combine files.
- `head` displays the first lines (`-n 5` means five lines).
- `more` and `less` are terminal pagers. In `less`, use arrow keys/Page Up/Page Down; press `q` to quit.
- **`reading` is not a standard Linux command.** The likely intended command may be `read` (a shell builtin) or “reading files” generally. `read` reads user input in a shell.

```bash
printf 'Linux commands\nLine 2\nLine 3\nLine 4\nLine 5\nLine 6\n' > notes/day2.txt
cat notes/day2.txt
head -n 3 notes/day2.txt
more notes/day2.txt
less notes/day2.txt
```

Try the shell `read` builtin in a terminal:

```bash
read -r -p "Type your name: " student_name
echo "Hello, $student_name"
```

## 5. Help commands: `man`

`man` opens the manual page for a command. Press `q` to quit.

```bash
man ls
man cp
```

If a manual page is not installed, try `ls --help` or `help cd` (for shell builtins).

## 6. Redirection and pipes

- `>` writes output to a file, replacing existing contents.
- `>>` appends output to a file.
- `<` takes input from a file.
- `2>` redirects error output.
- `|` sends one command's output to another command.

```bash
echo "First line" > output.txt
echo "Second line" >> output.txt
cat output.txt
wc -l < output.txt
ls missing_file 2> errors.txt
cat errors.txt
cat output.txt | wc -l
```

## 7. Copy, move, and delete: `cp`, `mv`, `rm`, `rmdir`

- `cp` copies files.
- `mv` moves or renames files.
- `rm` removes files; deleted files may not be recoverable.
- `rmdir` removes an **empty** directory only.

```bash
cp notes/day2.txt notes/day2-copy.txt
mv notes/day2-copy.txt notes/backup.txt
ls -l notes
rm notes/backup.txt
mkdir empty_folder
rmdir empty_folder
```

**Safety:** Avoid `rm -rf` while learning. Check your current directory with `pwd` and the target with `ls` before deleting anything.

## 8. Identify a file: `file`

`file` reports the type of a file based on its contents.

```bash
file notes/day2.txt
file /bin/ls
```

## 9. Run commands with elevated privileges: `sudo`

`sudo` runs a command with administrator privileges, if your account is authorized.

```bash
sudo -v
```

This validates your sudo credentials. You may be asked for your password. For safety, do not practise by changing or deleting system files. Only use `sudo` when you understand why administrator access is needed.

## 10. Create shortcuts: `alias`

An alias gives a command a short name. This example creates a temporary alias for the current shell session.

```bash
alias ll='ls -lah'
ll
unalias ll
```

Aliases usually do not persist after closing the terminal unless added to a shell startup file such as `~/.bashrc`.

## 11. Shells and scripts: `bash`, `sh`

- `bash` starts a Bash shell.
- `sh` starts the system's POSIX-style shell (which may be linked to another shell).
- Use `exit` to leave a nested shell.

Create and run a simple script:

```bash
cat > hello.sh <<'EOF'
#!/bin/sh
echo "Hello from a shell script"
EOF
sh hello.sh
bash hello.sh
```

## 12. Permissions: `chmod`

`chmod` changes file permissions. This example makes a script executable by its owner, group, and others.

```bash
chmod u+x hello.sh
ls -l hello.sh
./hello.sh
```

Numeric example: `chmod 755 hello.sh` gives the owner read/write/execute and group/others read/execute. Use permissions thoughtfully, especially on shared systems.

## 13. Processes: `ps`, `kill`

- `ps` displays process information.
- `kill` sends a signal to a process; by default, it requests termination.

```bash
ps
ps -ef
sleep 300 &
ps
```

Find the `sleep` process PID from the output, then replace `<PID>` with that number:

```bash
kill <PID>
```

Do not type the angle brackets literally. Avoid killing processes you do not recognize. You can also use `jobs` and `kill %1` if the background `sleep` command is the first job in your current shell.

## 14. Package management: `dpkg`

On Debian/Ubuntu systems, `dpkg` manages `.deb` packages.

```bash
dpkg --version
dpkg -l | head
dpkg -s bash
```

To list installed packages, use `dpkg -l`. Installing or removing packages changes the system and generally requires administrator privileges; do not remove packages for this exercise.

## 15. Shutdown and restart

The commands below can shut down or reboot the computer. **Do not run them on your main computer or a shared/remote server just for practice.** Learn the syntax only, or use a disposable virtual machine with permission.

```bash
# Syntax examples — do not execute unless you intend to:
sudo shutdown -h now    # shut down now
sudo reboot             # restart now
```

Some Linux systems also accept `sudo shutdown -r now` to restart.
