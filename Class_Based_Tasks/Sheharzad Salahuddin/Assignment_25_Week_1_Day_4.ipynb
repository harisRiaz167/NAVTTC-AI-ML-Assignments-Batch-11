**Create a dictionary**

```python
country_capitals = {
    "Germany": "Berlin",
    "Canada": "Ottawa",
    "England": "London"
}

print(country_capitals)
print(type(country_capitals))
```

**Explanation:** A dictionary stores data as key: value pairs inside curly braces { }. Each key is linked to one value.

**Create a dictionary with dict()**

```python
student = dict(name="Ali", age=20, course="AI")
print(student)

pairs = dict([("a", 1), ("b", 2)])
print(pairs)
```

**Explanation:** dict() can build a dictionary from keyword arguments or from a list of (key, value) pairs.

**Keys must be immutable**

```python
numbers = {1: "One", 2: "Two", 3: "Three"}
print(numbers)

coordinates = {(0, 0): "origin", (1, 2): "point"}
print(coordinates)

try:
    invalid = {[1, 2]: "list key"}
except TypeError as error:
    print("TypeError:", error)
```

**Explanation:** Keys can be immutable types such as numbers, strings and tuples. A list cannot be a key, so it raises a TypeError. Values can be of any type.

**Keys must be unique**

```python
hogwarts_houses = {
    "Harry Potter": "Gryffindor",
    "Hermione Granger": "Gryffindor",
    "Ron Weasley": "Gryffindor",
    "Harry Potter": "Slytherin"
}

print(hogwarts_houses)
```

**Explanation:** If the same key appears twice, the later value replaces the earlier one. Values, however, can repeat.

**Access values with [] and get()**

```python
country_capitals = {
    "Germany": "Berlin",
    "Canada": "Ottawa",
    "England": "London"
}

print(country_capitals["Germany"])
print(country_capitals.get("England"))
print(country_capitals.get("Japan"))
print(country_capitals.get("Japan", "Not found"))

try:
    print(country_capitals["Japan"])
except KeyError as error:
    print("KeyError:", error)
```

**Explanation:** A value is accessed by putting its key in square brackets. get() does the same but returns None, or a default value, instead of raising a KeyError when the key is missing.

**Add and change items**

```python
country_capitals = {
    "Germany": "Berlin",
    "Canada": "Ottawa"
}

country_capitals["Italy"] = "Rome"
print(country_capitals)

country_capitals["Germany"] = "Bonn"
print(country_capitals)

country_capitals.update({"Germany": "Berlin", "Japan": "Tokyo"})
print(country_capitals)
```

**Explanation:** Assigning to a new key adds an item, and assigning to an existing key changes its value. update() adds or changes several items at once.

**Remove items with del, pop(), popitem() and clear()**

```python
country_capitals = {
    "Germany": "Berlin",
    "Canada": "Ottawa",
    "Italy": "Rome",
    "Japan": "Tokyo"
}

del country_capitals["Germany"]
print(country_capitals)

removed = country_capitals.pop("Canada")
print("pop():", removed, "->", country_capitals)

last = country_capitals.popitem()
print("popitem():", last, "->", country_capitals)

country_capitals.clear()
print("clear():", country_capitals)
```

**Explanation:** del and pop() remove the item with the given key; pop() also returns its value. popitem() removes and returns the last item added. clear() removes every item.

**Diagram:** keys(), values() and items() give the keys, the values, or (key, value) pairs.



**keys(), values() and items()**

```python
student = {"name": "Ali", "age": 20}

print(student.keys())
print(student.values())
print(student.items())
```

**Explanation:** keys() returns all the keys, values() returns all the values and items() returns each key-value pair as a tuple.

**Iterate through a dictionary**

```python
country_capitals = {
    "United States": "Washington D.C.",
    "Italy": "Rome"
}

for country in country_capitals:
    print(country)

print()

for country, capital in country_capitals.items():
    print(country, "->", capital)
```

**Explanation:** Looping over a dictionary gives its keys. Looping over items() gives both the key and the value of each pair.

**Length and membership test**

```python
file_types = {
    ".txt": "Text File",
    ".pdf": "PDF Document",
    ".jpg": "JPEG Image"
}

print("Length:", len(file_types))
print(".pdf" in file_types)
print(".mp3" in file_types)
print("Text File" in file_types)
print("Text File" in file_types.values())
```

**Explanation:** len() returns the number of key-value pairs. The in operator checks keys only, so 'Text File' is not found until values() is checked.

**Diagram:** The common dictionary methods, grouped by what they do.



**copy() and setdefault()**

```python
original = {"a": 1, "b": 2}

backup = original.copy()
backup["c"] = 3
print("original:", original)
print("copy:", backup)

original.setdefault("b", 99)
original.setdefault("z", 0)
print(original)
```

**Explanation:** copy() makes a separate dictionary, so changing the copy does not change the original. setdefault() returns the value of a key, and adds the key with a default value only if it is missing.

**Diagram:** A dictionary can hold other dictionaries as values.



**Nested dictionaries**

```python
students = {
    1: {"name": "Ali", "age": 20},
    2: {"name": "Sara", "age": 22}
}

print(students[1]["name"])
print(students[2]["age"])

for student_id, info in students.items():
    print(student_id, info["name"], info["age"])
```

**Explanation:** A nested dictionary stores dictionaries as values. The first key selects the inner dictionary, and the second key selects a value inside it.

**Dictionary comprehension**

```python
squares = {x: x * x for x in range(1, 6)}
print(squares)

prices = {"apple": 100, "banana": 40, "mango": 150}
expensive = {fruit: price for fruit, price in prices.items() if price > 50}
print(expensive)
```

**Explanation:** A dictionary comprehension builds a dictionary in one line, optionally keeping only the items that meet a condition.

## Observation

- A dictionary stores data as key: value pairs inside curly braces.
- Keys must be unique and immutable; values can be of any type and can repeat.
- Values are read with [] or get(); get() avoids a KeyError when a key is missing.
- Items are added or changed by assigning to a key or with update().
- del, pop(), popitem() and clear() remove items.
- keys(), values() and items() return the keys, the values and the pairs; in checks keys.
- Dictionaries can be nested and built with dictionary comprehensions.

## Conclusion

Dictionaries store data as labelled pairs, so values can be looked up quickly by a meaningful key instead of by position. This makes them ideal for records, settings and lookup tables.
