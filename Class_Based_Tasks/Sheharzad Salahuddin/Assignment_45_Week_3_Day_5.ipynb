A simple way to store big data sets is in CSV files (comma-separated values). They hold plain text and can be read by almost any program, including Pandas. This notebook uses `data_full.csv`, the W3Schools workout data set, which must be in the same folder.

**Diagram:** Each line of the CSV becomes a row, and the first line gives the column names.



**Load a CSV file and print the whole DataFrame**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df.to_string())
```

**Explanation:** read_csv() loads the CSV file into a DataFrame. to_string() prints the entire DataFrame, all 169 rows.

**Diagram:** Pandas shortens long output unless it is told otherwise.



**Print without to_string()**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

print(df)
```

**Explanation:** If the DataFrame has many rows, a plain print() shows only the first 5 and the last 5 rows, with ... in between.

**Check the maximum number of rows displayed**

```python
import pandas as pd

print(pd.options.display.max_rows)
```

**Explanation:** pd.options.display.max_rows sets how many rows can be printed before Pandas shortens the output. The default is 60.

**Increase the maximum number of rows**

```python
import pandas as pd

pd.options.display.max_rows = 9999

df = pd.read_csv("data_full.csv")

print(df)

pd.reset_option("display.max_rows")
```

**Explanation:** After raising max_rows to 9999, print(df) shows every row. reset_option() puts the default back so later cells are not affected.

**Useful read_csv() options**

```python
import pandas as pd

df = pd.read_csv("data_full.csv", usecols=["Duration", "Calories"], nrows=5)
print(df)

df2 = pd.read_csv("data_full.csv", dtype={"Pulse": "float64"})
print(df2.dtypes)
```

**Explanation:** usecols loads only the named columns and nrows loads only the first rows, which helps with very large files. dtype sets a column's data type while reading.

**Write a DataFrame back to a CSV file**

```python
import pandas as pd

df = pd.read_csv("data_full.csv")

long_workouts = df[df["Duration"] >= 120]
long_workouts.to_csv("long_workouts.csv", index=False)

print(pd.read_csv("long_workouts.csv"))
```

**Explanation:** to_csv() saves a DataFrame as a CSV file. index=False leaves out the row numbers. Reading the new file back shows it was saved correctly.

## Observation

- CSV files store tables as plain text, with commas between values.
- pd.read_csv() loads a CSV file into a DataFrame, using the first line as column names.
- to_string() prints the whole DataFrame; plain print() shortens it when there are more rows than max_rows.
- pd.options.display.max_rows controls that limit (60 by default).
- read_csv() options such as usecols, nrows and dtype control what is loaded, and to_csv() saves a DataFrame.

## Conclusion

Reading and writing CSV files is the most common way to get data into and out of Pandas. read_csv() handles the parsing in one line, and display options control how much of a large table is shown.
