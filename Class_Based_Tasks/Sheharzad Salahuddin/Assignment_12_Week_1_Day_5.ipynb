## Topic: Python
Run and examine the if-else example using i = 20 and the condition i < 15.

```python
i = 20

if i < 15:
    print("i is less than 15")
else:
    print("i is greater than or equal to 15")
```

## Observation
- The variable i is assigned the value 20.
- The condition i < 15 evaluates to False.
- Therefore, the else block executes.
- The output is "i is greater than or equal to 15".
