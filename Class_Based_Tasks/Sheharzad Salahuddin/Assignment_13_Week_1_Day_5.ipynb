Run and examine the nested-if example using i = 10 and nested conditions i < 10 and i < 12.

```python
i = 10

if i < 10:
    if i < 12:
        print("i is less than 12")
```

## Observation

- The variable i is assigned the value 10.
- The outer condition i < 10 is False because 10 is not less than 10.
- The inner condition i < 12 is not evaluated because the outer condition is False.
- Therefore, the print statement is not executed.
