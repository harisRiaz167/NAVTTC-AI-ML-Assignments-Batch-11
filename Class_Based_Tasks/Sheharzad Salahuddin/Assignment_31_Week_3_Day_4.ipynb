**Python data types for comparison**

**Explanation:** Python has strings, integers, floats, booleans and complex numbers. NumPy has these and more, each named with a character such as i, f or U.

**Check the data type of an integer array**

```python
import numpy as np

arr = np.array([1, 2, 3, 4])

print(arr.dtype)
```

**Explanation:** The dtype property returns the data type of an array. An array of whole numbers is int64 on most systems.

**Check the data type of a string array**

```python
import numpy as np

arr = np.array(["apple", "banana", "cherry"])

print(arr.dtype)
```

**Explanation:** <U6 means a Unicode string with a maximum length of 6 characters, the length of the longest item, 'banana' and 'cherry'.

**Check the data type of float and boolean arrays**

```python
import numpy as np

print(np.array([1.5, 2.0, 3.25]).dtype)
print(np.array([True, False]).dtype)
print(np.array([1, 2.5]).dtype)
```

**Explanation:** Decimal numbers give float64 and True/False values give bool. Mixing integers and floats gives float64, because an array holds one type and float can store both.

**Create an array with a set data type**

```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype="S")

print(arr)
print(arr.dtype)
```

**Explanation:** The dtype argument chooses the data type when the array is created. 'S' makes byte strings, shown with a b prefix, and S1 means each one is 1 byte long.

**Set the size of the data type**

```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype="i4")

print(arr)
print(arr.dtype)
```

**Explanation:** For i, u, f, S and U a size can be given. 'i4' creates 4-byte integers, so the dtype is int32.

**A value that cannot be converted raises an error**

```python
import numpy as np

try:
    arr = np.array(["a", "2", "3"], dtype="i")
except ValueError as error:
    print("ValueError:", error)
```

**Explanation:** 'a' cannot be turned into an integer, so NumPy raises a ValueError. Strings such as '2' and '3' could be converted, but every element must be convertible.

**Diagram:** astype() returns a copy of the array in a new data type.



**Convert float to integer with astype('i')**

```python
import numpy as np

arr = np.array([1.1, 2.1, 3.1])

newarr = arr.astype("i")

print(newarr)
print(newarr.dtype)
```

**Explanation:** astype() makes a copy of the array in a new data type. Converting to 'i' cuts off the decimal part.

**Convert float to integer with astype(int)**

```python
import numpy as np

arr = np.array([1.1, 2.1, 3.1])

newarr = arr.astype(int)

print(newarr)
print(newarr.dtype)
```

**Explanation:** A Python type such as int can be used instead of a character. Here it gives int64.

**Convert integer to boolean**

```python
import numpy as np

arr = np.array([1, 0, 3])

newarr = arr.astype(bool)

print(newarr)
print(newarr.dtype)
```

**Explanation:** When converting to bool, 0 becomes False and every other number becomes True.

**Memory used by different data types**

```python
import numpy as np

for dtype in ["i1", "i2", "i4", "i8", "f4", "f8"]:
    arr = np.array([1, 2, 3], dtype=dtype)
    print(dtype, "->", arr.dtype, "| bytes per item:", arr.itemsize)
```

**Explanation:** itemsize shows how many bytes each element uses. Smaller types save memory but can store a smaller range of values.

## Observation

- NumPy data types are named by characters such as i (integer), f (float), b (boolean), S and U (strings).
- The dtype property shows an array's data type.
- The dtype argument sets the type when creating an array, and a size such as i4 can be added.
- A ValueError is raised if a value cannot be converted to the requested type.
- astype() creates a copy of the array in a new data type.
- Different types use different amounts of memory per element.

## Conclusion

Every NumPy array has one data type for all its elements. Choosing and converting data types controls how values are stored, how much memory they use and how calculations behave.
