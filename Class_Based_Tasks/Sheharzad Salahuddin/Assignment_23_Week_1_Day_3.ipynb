## Topic: Python — String Data Type

**Objective:** Complete the exercise on the string data type in Python.

**Reference:** https://www.programiz.com/python-programming/string

**Create a string**

```python
string1 = "Python programming"
string2 = 'Hello'

print(string1)
print(string2)
print(type(string1))
```

**Explanation:** A string is a sequence of characters. It can be written inside double quotes or single quotes; both create the same str type.

**Multiline strings**

```python
message = """
Never gonna give you up
Never gonna let you down
"""

print(message)
```

**Explanation:** Triple quotes (""" or ''') create a string that spans several lines, keeping the line breaks.

**Diagram:** Each character has a positive index from the start and a negative index from the end.



**Access characters with an index**

```python
greet = "hello"

print(greet[1])
print(greet[-4])
```

**Explanation:** Indexing starts at 0, so greet[1] is the second character, 'e'. Negative indexes count from the end, so greet[-4] is also 'e'.

**Slicing a string**

```python
greet = "Python"

print(greet[1:4])
print(greet[:2])
print(greet[2:])
print(greet[::-1])
```

**Explanation:** A slice returns the characters from the start index up to, but not including, the stop index. A step of -1 reverses the string.

**Diagram:** A string's characters cannot be changed after it is created.



**Strings are immutable**

```python
message = "Hola Amigos"

try:
    message[0] = "H"
except TypeError as error:
    print("TypeError:", error)

message = "Hello Friends"
print(message)
```

**Explanation:** Changing a single character raises a TypeError because strings are immutable. A variable can still be assigned a completely new string.

**Delete a string**

```python
text = "temporary"
del text

try:
    print(text)
except NameError as error:
    print("NameError:", error)
```

**Explanation:** Characters cannot be deleted from a string, but the whole string can be deleted with del. After that, using the name raises a NameError.

**Compare two strings**

```python
str1 = "Hello, world!"
str2 = "I love Swift."
str3 = "Hello, world!"

print(str1 == str2)
print(str1 == str3)
```

**Explanation:** The == operator returns True when two strings contain exactly the same characters.

**Join (concatenate) two strings**

```python
greet = "Hello, "
name = "Jack"

result = greet + name
print(result)
print("Ha" * 3)
```

**Explanation:** + joins two strings into a new one, and * repeats a string a given number of times.

**Iterate through a string**

```python
greet = "Hello"

for letter in greet:
    print(letter)
```

**Explanation:** A for loop visits each character of the string in order.

**String length**

```python
greet = "Hello"

print(len(greet))
```

**Explanation:** len() returns the number of characters in the string.

**String membership test**

```python
print("a" in "program")
print("at" not in "battle")
```

**Explanation:** in returns True if the text appears inside the string. not in returns True if it does not; here 'at' does appear in 'battle', so the result is False.

**Diagram:** Common string methods, grouped by what they do. Because strings are immutable, each method returns a new value instead of changing the original.



**Change the case**

```python
text = "python is FUN"

print(text.upper())
print(text.lower())
print(text.title())
print(text)
```

**Explanation:** upper(), lower() and title() return new strings in upper case, lower case and title case. The last line shows the original string is unchanged.

**Search with find(), index() and count()**

```python
text = "I love Python and Python loves me"

print(text.find("Python"))
print(text.find("Java"))
print(text.index("loves"))
print(text.count("Python"))
```

**Explanation:** find() returns the index of the first match, or -1 if there is none. index() works the same way but raises an error if there is no match. count() returns the number of matches.

**Test strings with startswith(), isnumeric() and isalpha()**

```python
print("Python".startswith("Py"))
print("12345".isnumeric())
print("12a45".isnumeric())
print("Hello".isalpha())
```

**Explanation:** These methods return True or False. startswith() checks the beginning of the string, isnumeric() checks that every character is a digit, and isalpha() checks that every character is a letter.

**Edit with replace(), strip() and rstrip()**

```python
text = "   I like cats   "

print(text.replace("cats", "dogs"))
print("[" + text.strip() + "]")
print("[" + text.rstrip() + "]")
```

**Explanation:** replace() swaps one piece of text for another. strip() removes spaces from both ends and rstrip() removes them only from the right.

**Diagram:** split() breaks a string into a list, and join() puts a list back together with a separator.



**split(), partition() and join()**

```python
sentence = "I love Python"

words = sentence.split(" ")
print(words)

print("-".join(words))

print("Python is fun".partition("is"))
```

**Explanation:** split() breaks the string at each separator and returns a list. join() connects the items of a list using the string it is called on. partition() splits at the first match and returns a tuple of three parts: before, the match and after.

**Escape sequences**

```python
example = "He said, \"What's there?\""
print(example)

print("Line one\nLine two")
print("Name:\tAli")
print("C:\\Users\\Ali")
```

**Explanation:** A backslash starts an escape sequence. `\"` prints a double quote inside a double-quoted string, `\n` starts a new line, `\t` inserts a tab and `\\` prints one backslash.

**Raw strings**

```python
print("C:\\new\\folder")
print(r"C:\new\folder")
```

**Explanation:** Putting r before a string makes it a raw string, so backslashes are printed as they are instead of starting escape sequences.

**f-strings (formatted strings)**

```python
name = "Cathy"
country = "UK"
age = 20
marks = 92.456

print(f"{name} is from {country}")
print(f"Next year {name} will be {age + 1}")
print(f"Marks: {marks:.2f}")
```

**Explanation:** An f-string starts with f, and anything inside { } is evaluated and inserted into the string, including expressions such as age + 1. :.2f formats a number to two decimal places.

## Observation

- A string is a sequence of characters written in single, double or triple quotes.
- Characters are accessed by positive or negative index, and slicing returns part of a string.
- Strings are immutable: characters cannot be changed, so methods return new strings.
- Strings can be compared with ==, joined with +, repeated with *, looped over and checked with in.
- Methods such as upper(), find(), replace(), strip(), split() and join() change case, search, edit and split text.
- Escape sequences add special characters, raw strings ignore them, and f-strings insert values into text.

## Conclusion

Strings are used to store and work with text. Python's indexing, slicing, operators and built-in methods make it easy to read, search, format and transform text.
