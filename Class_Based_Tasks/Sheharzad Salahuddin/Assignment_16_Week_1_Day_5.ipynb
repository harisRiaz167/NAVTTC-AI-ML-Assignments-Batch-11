## Topic: Python — While Loops

**Objective:** Understand and practice while loops in Python.

**Reference:** https://www.geeksforgeeks.org/python-while-loop/

**Print numbers from 1 to 5**

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

**Calculate the sum of numbers from 1 to 5**

```python
i = 1
total = 0

while i <= 5:
    total += i
    i += 1

print("Sum:", total)
```

## Observation

- A while loop repeats while its condition is True.
- The variable i increases by 1 in each iteration.
- The loop stops when i becomes 6.
- The sum of numbers from 1 to 5 is 15.
