## Topic: Python — Break Statement

**Objective:** Practice using the break statement.

**Reference:** https://www.geeksforgeeks.org/python-break-statement/

**Stop a loop using break**

```python
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

**How break works**

- The loop starts with i = 1.
- Numbers are printed until i reaches 5.
- When i == 6, the break statement terminates the loop before printing 6.

## Observation

- The break statement terminates a loop immediately.
- When i becomes 6, the loop stops.
- Numbers 1 through 5 are printed.
- No further iterations are executed.
