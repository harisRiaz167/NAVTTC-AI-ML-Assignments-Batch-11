```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("sampledata1.csv")
```

```python
x_axis = data['Sales'].mean()
y_axis = data['Profit'].mean()
plt.plot(x_axis,y_axis, marker = 'o')
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
x_axis = data['Discount'].mean()
y_axis = data['Quantity'].mean()
plt.plot(x_axis,y_axis, marker = 'o')
plt.xlabel("Discount")
plt.ylabel("Quantity")
plt.show()
```

```python
y_axis = data['Sales'][:50]
x_axis = data['Region'][:50]
plt.bar(x_axis,y_axis)
plt.ylabel("Sales")
plt.xlabel("Region")
plt.xticks(rotation=90)
plt.show()
```

```python
y_axis = data['Sales'][:50]
x_axis = data['State'][:50]
plt.bar(x_axis,y_axis)
plt.ylabel("Sales")
plt.xlabel("State")
plt.xticks(rotation=90)
plt.show()
```

```python
y_axis = data['Quantity'][:50]
x_axis = data['Profit'][:50]
plt.bar(x_axis,y_axis,width = 20)
plt.ylabel("Quantity")
plt.xlabel("Profit")
plt.xticks(rotation=90)
plt.show()
```

```python
top_5 = data.sort_values(by='Sales', ascending=False).head(5)

y_axis = top_5['Sales']
mylabels = top_5['Region']

plt.pie(y_axis, labels=mylabels)
plt.show()
```
