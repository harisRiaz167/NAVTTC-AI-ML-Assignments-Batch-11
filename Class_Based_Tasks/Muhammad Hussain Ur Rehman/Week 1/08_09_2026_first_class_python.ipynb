# virtual store slip generation

```python
# copy and RUN in .py file else will give error
print("============welcome to store==============")
name = input("Please enter your name : ")

items = []
while True:
    item = input("please enter item name : ")
    items.append(item)
    yes = input("do you want to add another item (Y/N) : ")
    if yes.lower() == "y":
        continue
    else:
        break

quantities = []
for item in items:
    qty = int(input(f"Please enter quantity for {item} : "))
    quantities.append(qty)

sum_qut = sum(quantities)

prices = []
total_price = 0


for i in range(len(items)):
    price = float(input(f"Please enter price for {items[i]} : "))
    prices.append(price)
    total_price += price * quantities[i]
print("\n\n\n")
print("------------Welcome to our store-------------")
print("\n===================Bill=================")
print(f"\nCustomer Name : {name}")
print(f"Total Quantity : {sum_qut}")
print(f"Subtotal : ${total_price:.2f}")
print("\n====================end=================")
```

## finding Mean and Range without usin python min() and max() funs

```python
num_1 = int(input("please enter first number : "))
num_2 = int(input("please enter second number : "))
num_3 = int(input("please enter third number : "))
num_4 = int(input("please enter forth number : "))

mean = num_1 + num_2 + num_3 + num_4/4
nums = []
nums.append(num_1)
nums.append(num_2)
nums.append(num_3)
nums.append(num_4)
nums.sort()
min_num = nums[0]
max_num = nums[-1]
range =  max_num - min_num
print(f"Mean is : {mean}")
print(f"Range is : {range}")
```

#Finding if input number is odd or even or zero

```python
num = int(input("please enter number : "))
if(num % 2 == 0 and num !=0):
  print("number is even")
elif(num == 0):
  print("Number is zero")
else:
  print("number is odd")
```
