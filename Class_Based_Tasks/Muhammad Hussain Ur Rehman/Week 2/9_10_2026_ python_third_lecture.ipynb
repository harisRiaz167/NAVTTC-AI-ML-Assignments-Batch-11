##Guess the number Game


```python
num = 4
print("=====================================")
print("          Guess the Number")
print("=====================================")
while(True):

    num_user = int(input("\nGuess the number between 1 - 6 : "))
    if num_user > num:
        print("\n*********Number is greater, Try again*********")
        continue
    elif num_user < num:
        print("\n*********Number is less, Try again*********")
        continue
    else:
        print(f"\n===============    Your find the number, its {num}    =================")
        print("\n========================END=====================")
        break


```

##find word length if even or odd

```python
x = ['alphabets','numerics', 'data types', '123', 'wild', 'abc', 'chars']

even =[]
odd =[]

for i in range(len(x)):
    if(len(x[i]) % 2 == 0 ):
        even.append(x[i])
    else:
        odd.append(x[i])
print(f"Even : {even}" )
print(f"Odd : {odd}")
```

#Functions

```python
def sum_two_numbers(x, y):                  # defining functions
  sum = x + y
  return sum                                # if we don't return something, then function will by default return None

x = sum_two_numbers(10,20)                  # calling function with positional argumnets
y = sum_two_numbers(x = 10, y = 20)         # calling function with key word arguments
print(x)
print(y)
```

##Python rule : Positional arguments first, then key word arguments

```python
def welcome(f_name, l_name):
  print(f"welcome {f_name}, {l_name}")

x = welcome('haris','riaz')                            # calling function with positional argumnets
y = welcome(l_name = 'riaz', f_name='Haris')           # calling function with key word arguments
```
