```python
para = input("Enter brackets: ")
lis = []

for i in para:

    if i == '(':
        lis.append(')')
    
    elif i == '{':
        lis.append('}')

    elif i == '[':
        lis.append(']')

    elif i == ')' or i == '}' or i == ']':
        lis.append('(')
        
        if lis[-1] == i:
            lis.pop()
        
        else:
            print("Wrong or missing opening bracket for:", i)
            lis.pop()

if len(lis) > 0:
    print("Missing:", lis)
```
