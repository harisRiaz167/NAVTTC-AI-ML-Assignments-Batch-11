```python
import pandas as pd

df_colors = pd.DataFrame({
     'Color': ['Red', 'Green', 'Red', 'Blue']
})

one_hot = pd.get_dummies(df_colors, columns=['Color'], dtype=int)

print(one_hot)

print(df_colors)
```

```python
from sklearn.preprocessing import OrdinalEncoder

education = [['Bachelors'],['Masters'],['PhD'],['Bachelors'],['PhD']]

OrdinalEncoder = OrdinalEncoder(
    categories =[['Bachelors','Masters','PhD']]
)

encoded_education = OrdinalEncoder.fit_transform(education)
print(encoded_education)
```
