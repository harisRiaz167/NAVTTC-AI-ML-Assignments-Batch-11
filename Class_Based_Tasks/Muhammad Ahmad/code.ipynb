```python
items = {
    'laptop': 1000,
    'mobile': 500,
    'tablet': 200,
    'desktop': 800
}

cus_name = input('Enter your name: ')
print('Welcome', cus_name, 'to our store!')

product = []          
grand_total = 0

while True:
    input_item = input('Enter the item you want to buy (or "#" to finish): ')

    if input_item == '#':
        break

    if input_item not in items:
        print('Item entered is not available.')
        continue

    try:
        item_quantity = int(input('Enter the quantity: '))
    except ValueError:
        print('Please enter a valid number for quantity.')
        continue

    if item_quantity <= 0:
        print('Quantity must be positive.')

    price = items[input_item]
    line_total = price * item_quantity
    product.append((input_item, item_quantity, line_total))
    grand_total += line_total

    print(f'Added: {item_quantity} x {input_item} = {line_total}')

# Final bill
print('========================================')
print('BILL')
print('Customer:', cus_name)

if not product:
    print('No items purchased.')
else:
    for name, qty, total in product:
        print(f'{name:<10} x {qty:<3} = {total}')
    print('Grand Total:', grand_total)

print('Thank you for shopping with us,', cus_name )
print('========================================')
```
