```python
X = ['alphabets', 'numbers', 'symbols', 'data types', 'functions', 'loops', 'conditionals', 'classes', 'modules', 'exceptions']

even = []
odd = []

for i in range(len(X)):
    if len(X[i]) % 2 == 0:
        even.append(X[i])
    else:
        odd.append(X[i])

print("Even elements:", even)
print("Odd elements:", odd)
```

```python
car = False
credit = False

car_price = 250000

input_car = input("Do you have a car? yes/no: ").strip().lower()
input_credit = input("Do you have credit? yes/no: ").strip().lower()

if input_car == "yes":
    car = True
if input_credit == "yes":
    credit = True

if car and credit:
    payment = car_price * 0.1
    print(f"Pay 10% of the car price: {payment}")

if car and not credit:
    payment = car_price * 0.2
    print(f"Pay 20% of the car price: {payment}")
```
