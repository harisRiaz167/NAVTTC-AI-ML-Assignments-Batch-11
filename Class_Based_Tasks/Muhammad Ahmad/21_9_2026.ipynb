```python
def Single_Check(s):
    for i in range(len(s)):
        if s.count(s[i]) == 1:
            print(i)

string = input("Enter a string :")
Single_Check(string)
```

```python
l = [1,2,3,4]
new_l = []

for i in range(len(l)):
    first = 1

    for j in range(len(l)):
        if i != j:
          first *= l[j] 

    new_l.append(first) 

print(new_l)
```
