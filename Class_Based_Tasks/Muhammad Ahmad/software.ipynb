```python
from datetime import datetime

balance = 0
history = []

def log(action, amount):
    now = datetime.now()
    history.append(f"[{now}] {action}: ${amount:} | Balance: ${balance:}")

def deposit(amount):
    global balance
    if amount <= 0:
        print("Amount has to be positive man")
        return

    balance = balance + amount
    log("Deposit", amount)
    print(f"Deposited ${amount:}, new balance is ${balance:}")

def withdraw(amount):
    global balance
    if amount <= 0:
        print("Amount has to be positive man")
        return
    if amount > balance:
        print("You don't have enough money for that")
        return

    balance = balance - amount
    log("Withdraw", amount)
    print(f"Withdrew ${amount:}, new balance is ${balance:}")

def check_balance():
    print("Your balance is: $" + str(balance))

def view_history():
    if len(history) == 0:
        print("No transactions yet")
        return

    print("--- Transaction History ---")
    for h in history:
        print(h)
    print("---------------------------")


def main():
    while True:
        print("1. Deposit")
        print("2. Withdraw")
        print("3. Check balance")
        print("4. Transaction history")
        print("5. Quit")

        choice = input("What do you want to do? ")

        if choice == "1":
            amt = int(input("How much you wanna deposit? "))
            deposit(amt)
        elif choice == "2":
            amt = int(input("How much you wanna withdraw? "))
            withdraw(amt)
        elif choice == "3":
            check_balance()
        elif choice == "4":
            view_history()
        elif choice == "5":
            print("Thnx for coming !")
            break
        else:
            print("that's not a valid option, try again")


main()
```
