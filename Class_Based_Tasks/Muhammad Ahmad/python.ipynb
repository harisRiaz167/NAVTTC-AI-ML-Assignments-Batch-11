```python

x1=input('Enter your num1: ')
x2=input('Enter your num2: ')
x3=input('Enter your num3: ')
x4=input('Enter your num4: ')

avg = (int(x1) + int(x2) + int(x3) + int(x4)) / 4
print('Average is: ', avg)

#for max num
max_num = max(int(x1), int(x2), int(x3), int(x4))
print('Max number is: ', max_num)

#for min num
min_num = min(int(x1), int(x2), int(x3), int(x4))
print('Min number is: ', min_num)

ran = max_num - min_num
print('Range is: ', ran)

    
```

```python
num = []
input_num = input('Enter a number (or "#" to finish): ')
while input_num != "#":
    try:
        num.append(int(input_num))
    except ValueError:
        print('Please enter a valid number.')
    input_num = input('Enter a number (or "#" to finish): ')

if num:
    avg = sum(num) / len(num)
    print('Average:', avg)

max_num = max(num)
min_num = min(num)
print('Maximum:', max_num)
print('Minimum:', min_num)

ran = max_num - min_num
print('Range:', ran)
```
