```python
def miss_brackets(text):
    parenthesis = [('(', ')'), ('{', '}'), ('[', ']')]
    miss = []

    for open_b, close_b in parenthesis:
        diff = text.count(open_b) - text.count(close_b)
        if diff > 0:
            miss += [close_b] * diff      # more opening -> closing missing
        elif diff < 0:
            miss += [open_b] * (-diff)    # more closing -> opening missing

    if miss:
        print("Input",para)
        print("Missing brackets:", " ".join(miss))
    else:
        print("Input",para)
        print("No brackets are missing")


para = input("Enter brackets: ")
miss_brackets(para)
```
