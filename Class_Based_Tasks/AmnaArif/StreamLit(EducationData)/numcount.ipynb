```python
sentence = "i love this movie"

tokens = sentence.split()

vocab = {
    "this": 2,
    "movie": 3,
    "love": 4,
    "i": 5
}

unk = 1
padding = 0

ids = []

for word in tokens:
    if word in vocab:
        ids.append(vocab[word])
    else:
        ids.append(unk)

print("Tokens:", tokens)
print("IDs:", ids)
```
