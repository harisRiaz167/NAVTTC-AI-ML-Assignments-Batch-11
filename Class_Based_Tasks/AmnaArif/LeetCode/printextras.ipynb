```python
l = ['a', 'b', 'c', 'd']
l2 = ['a', 'c', 'e', 'd', 'd']

store = {}

for i in l:
    if i in store:
        store[i] = store[i] + 1
    else:
        store[i] = 1

for i in l2:
    if i in store:
        store[i] = store[i] - 1
    else:
        store[i] = -1

for i in store:
    if store[i] < 0:
        print(i)
```
