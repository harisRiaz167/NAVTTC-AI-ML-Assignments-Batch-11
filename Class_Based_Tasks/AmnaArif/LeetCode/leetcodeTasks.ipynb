TASK-03 Write a Python program that checks whether all brackets/parentheses in a sentence are properly opened and closed.

```python
def check_parentheses(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    opening = "([{"

    for i, char in enumerate(s):
        if char in opening:
            stack.append((char, i))

        elif char in pairs:
            if not stack:
                print("Missing opening bracket:", pairs[char])
                return s

            if stack[-1][0] != pairs[char]:
                wrong_opening, index = stack[-1]
                print("Incorrect bracket:", char)

                correct_closing = {
                    '(': ')',
                    '[': ']',
                    '{': '}'
                }

                s = s[:i] + correct_closing[wrong_opening] + s[i + 1:]
                return s

            stack.pop()

    # Check for missing closing brackets
    if stack:
        result = s

        for bracket, index in reversed(stack):
            correct_closing = {
                '(': ')',
                '[': ']',
                '{': '}'
            }

            result += correct_closing[bracket]

        print("Missing closing bracket(s)")
        return result

    return s


sentence = input("Enter a sentence: ")

corrected = check_parentheses(sentence)

print("Corrected sentence:")
print(corrected)
```

TASK-04
Write a Python program that merges two lists into a single list and sorts the resulting list in ascending order without using Python’s built-in sort() or sorted() functions.
Given:

```python

l1 = [1, 1, 2, 3, 4]
l2 = [1, 2, 5, 6, 7]


def swap(newlist, i, j):
    newlist[i], newlist[j] = newlist[j], newlist[i]


def mergelist():
    newlist = l1 + l2

    for i in range(len(newlist)):
        for j in range(i + 1, len(newlist)):
            if newlist[i] > newlist[j]:
                swap(newlist, i, j)

    return newlist


print(mergelist())
```

Task-06 find length of list, then divide by 2, and print the number that comes more then that no.
  eg list len is 6 , 6/2 is 3
print no that comes more than 3

```python

l = [2, 2, 3, 4, 4, 5, 5, 5, 3, 3, 1, 1]

n = len(l) 
store = {}

for i in l:
    if i in store:
        store[i] = store[i] + 1
    else:
        store[i] = 1

print(store)
print("n =", n)

for i in store:
    if store[i] > n:
        print(i)
```
