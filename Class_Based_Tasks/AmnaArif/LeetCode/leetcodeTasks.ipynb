Write a Python program that checks whether all brackets/parentheses in a sentence are properly opened and closed.

```python
def check_parentheses(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    opening = "([{"

    for i, char in enumerate(s):
        if char in opening:
            stack.append((char, i))

        elif char in pairs:
            if not stack:
                print("Missing opening bracket:", pairs[char])
                return s

            if stack[-1][0] != pairs[char]:
                wrong_opening, index = stack[-1]
                print("Incorrect bracket:", char)

                correct_closing = {
                    '(': ')',
                    '[': ']',
                    '{': '}'
                }

                s = s[:i] + correct_closing[wrong_opening] + s[i + 1:]
                return s

            stack.pop()

    # Check for missing closing brackets
    if stack:
        result = s

        for bracket, index in reversed(stack):
            correct_closing = {
                '(': ')',
                '[': ']',
                '{': '}'
            }

            result += correct_closing[bracket]

        print("Missing closing bracket(s)")
        return result

    return s


sentence = input("Enter a sentence: ")

corrected = check_parentheses(sentence)

print("Corrected sentence:")
print(corrected)
```
