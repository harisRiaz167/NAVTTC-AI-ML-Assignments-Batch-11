Complete the exercise on user-defined functions in Python.  
Reference: https://www.geeksforgeeks.org/functions-in-python/

```python
# A function with no parameters
def greet():
    print('Hello, welcome to Python!')

greet()

# A function with parameters and a return value
def add(a, b):
    return a + b

print('Sum:', add(5, 3))

# A function with a default parameter
def power(base, exp=2):
    return base ** exp

print('Square of 4:', power(4))
print('4 to the power 3:', power(4, 3))

# A function to check even or odd
def check_even_odd(n):
    if n % 2 == 0:
        return 'Even'
    return 'Odd'

print('7 is', check_even_odd(7))
```


Complete the exercise on the list data type in Python.  
Reference: https://www.programiz.com/python-programming/list

```python
# Create a list
fruits = ['apple', 'banana', 'mango']
print('List:', fruits)

# Access items using index
print('First item:', fruits[0])
print('Last item:', fruits[-1])

# Add items
fruits.append('orange')
fruits.insert(1, 'grapes')
print('After adding:', fruits)

# Remove items
fruits.remove('banana')
print('After removing banana:', fruits)

# Change an item
fruits[0] = 'pear'
print('After changing first item:', fruits)

# Slicing and length
print('First two items:', fruits[:2])
print('Length of list:', len(fruits))

# Sort the list
fruits.sort()
print('Sorted list:', fruits)
```


Complete the exercise on the tuple data type in Python.  
Reference: https://www.programiz.com/python-programming/tuple

```python
# Create a tuple
numbers = (10, 20, 30, 40, 50)
print('Tuple:', numbers)

# Access items
print('First item:', numbers[0])
print('Last item:', numbers[-1])

# Slicing
print('Items from index 1 to 3:', numbers[1:4])

# Tuples are immutable (cannot be changed)
try:
    numbers[0] = 99
except TypeError:
    print('Error: tuple items cannot be changed')

# Useful methods
t = (1, 2, 2, 3, 2)
print('Count of 2:', t.count(2))
print('Index of 3:', t.index(3))

# Tuple unpacking
a, b, c = (1, 2, 3)
print('a =', a, ', b =', b, ', c =', c)
```


Complete the exercise on the string data type in Python.  
Reference: https://www.programiz.com/python-programming/string

```python
text = 'Hello Python'

# Access characters and slicing
print('First character:', text[0])
print('Last character:', text[-1])
print('Slice [0:5]:', text[0:5])

# Common string methods
print('Length:', len(text))
print('Upper:', text.upper())
print('Lower:', text.lower())
print('Replace:', text.replace('Python', 'World'))
print('Split:', text.split())
print('Starts with Hello?', text.startswith('Hello'))

# Concatenation and repetition
print('Hello' + ' ' + 'World')
print('Hi! ' * 3)

# String formatting
name = 'Amna'
print(f'My name is {name}')

# Loop through a string
for ch in 'abc':
    print(ch)
```


Complete the exercise on the set data type in Python.  
Reference: https://www.programiz.com/python-programming/set

```python
# Create a set (duplicates are removed automatically)
s = {1, 2, 3, 3, 4}
print('Set:', s)

# Add and remove items
s.add(5)
print('After add(5):', s)
s.remove(1)
print('After remove(1):', s)

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print('Union:', a | b)
print('Intersection:', a & b)
print('Difference (a - b):', a - b)
print('Symmetric difference:', a ^ b)

# Check membership
print('Is 3 in a?', 3 in a)
```


Complete the exercise on the dictionary data type in Python.  
Reference: https://www.programiz.com/python-programming/dictionary

```python
# Create a dictionary
student = {'name': 'Amna', 'class': 'C2C', 'age': 20}
print('Dictionary:', student)

# Access values
print('Name:', student['name'])
print('Age:', student.get('age'))

# Add and update items
student['city'] = 'Lahore'
student['age'] = 21
print('After changes:', student)

# Remove an item
student.pop('city')
print('After pop:', student)

# Keys, values and items
print('Keys:', list(student.keys()))
print('Values:', list(student.values()))

# Loop through a dictionary
for key, value in student.items():
    print(key, ':', value)
```


Complete the exercise on exception handling in Python.  
Reference: https://www.programiz.com/python-programming/exception-handling

```python
# try / except for division by zero
try:
    result = 10 / 0
except ZeroDivisionError:
    print('Error: cannot divide by zero')

# Handling a wrong value type
try:
    number = int('abc')
except ValueError:
    print('Error: invalid number')

# try / except / else / finally
try:
    x = 10 / 2
except ZeroDivisionError:
    print('Division error')
else:
    print('No error, result is', x)
finally:
    print('This always runs')
```


Complete the exercise on user-defined exception handling in Python.  
Reference: https://www.programiz.com/python-programming/user-defined-exception

```python
# Create our own exception class
class AgeTooLowError(Exception):
    def __init__(self, age):
        self.age = age
        super().__init__(f'Age {age} is too low. Must be 18 or older.')

# Function that raises the exception
def check_age(age):
    if age < 18:
        raise AgeTooLowError(age)
    print('Age', age, 'is valid')

# Using the custom exception
try:
    check_age(20)
    check_age(15)
except AgeTooLowError as e:
    print('Caught error:', e)
```
