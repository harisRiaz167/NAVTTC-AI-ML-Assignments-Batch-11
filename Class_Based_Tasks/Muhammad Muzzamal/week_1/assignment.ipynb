# Python: Calculate the Area of a Triangle

**Week:** 1

Write a Python program to calculate the area of a triangle using its three sides and Heron's formula.


```python

import math

a = 5
b = 6
c = 7

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Area of the triangle:", area)
```

# Python: Solve a Quadratic Equation

**Week:** 1

Write a Python program using the `cmath` module to solve the quadratic equation \(ax^2 + bx + c = 0\) for `a = 1`, `b = 5`, and `c = 6`.


```python

import cmath

a = 1
b = 5
c = 6

d = (b**2) - (4 * a * c)

root1 = (-b + cmath.sqrt(d)) / (2 * a)
root2 = (-b - cmath.sqrt(d)) / (2 * a)

print("Root 1:", root1)
print("Root 2:", root2)
```

# Python: Looping Techniques

**Week:** 1

Practice Python looping techniques using `enumerate()`, `zip()`, `sorted()`, and `reversed()` to iterate through collections efficiently.

**Reference:** https://www.geeksforgeeks.org/looping-techniques-python/


```python

# Using enumerate()
fruits = ["Apple", "Banana", "Mango"]

for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)

# Using zip()
names = ["Ali", "Ahmed", "Sara"]
marks = [85, 90, 95]

for name, mark in zip(names, marks):
    print(name, mark)

# Using sorted()
numbers = [5, 2, 8, 1, 3]

for number in sorted(numbers):
    print(number)

# Using reversed()
for fruit in reversed(fruits):
    print(fruit)
```
