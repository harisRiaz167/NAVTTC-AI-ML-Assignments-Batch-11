```python
import pandas as pd
import numpy as np
np.random.seed(0)
df = pd.DataFrame({
    "Duration": np.random.randint(30, 120, 20),
    "Pulse": np.random.randint(90, 130, 20),
    "Maxpulse": np.random.randint(130, 180, 20),
    "Calories": np.random.uniform(200, 500, 20).round(1)})
print(df.head(10)); print(df.tail()); print(df.shape)
print(df.info()); print(df.describe())
print(df.mean(), df.median(), df.std())
print(df.sort_values("Calories", ascending=False).head())
print(df[df.Calories > 400]); print(df.groupby(pd.cut(df.Duration, [0, 60, 90, 120]))["Calories"].mean())
print(df.nunique(), df["Pulse"].value_counts().head())
```
