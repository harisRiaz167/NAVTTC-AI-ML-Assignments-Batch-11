```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Duration": [60, 60, 45, np.nan, 60, 45, 300],
    "Date": ["2020/12/01", "2020/12/02", "2020/12/03", "2020/12/04", "2020/12/02", "20201205", "2020/12/07"],
    "Pulse": [110, 117, 103, 109, 117, 102, 110],
    "Calories": [409.1, 479.0, 340.0, np.nan, 479.0, 282.4, 450.0]})
print(df.isnull().sum())
# empty cells
print(df.dropna())
df_fill = df.fillna({"Duration": df.Duration.mean(), "Calories": df.Calories.median()}); print(df_fill)
# wrong format
df_fill["Date"] = pd.to_datetime(df_fill["Date"], format="mixed")
print(df_fill)
# wrong data - replace outlier / drop
df_fill.loc[df_fill["Duration"] > 120, "Duration"] = 120
# duplicates
print(df_fill.duplicated())
df_fill = df_fill.drop_duplicates().reset_index(drop=True)
print(df_fill)
```
