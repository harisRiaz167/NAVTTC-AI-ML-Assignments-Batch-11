```python
import pandas as pd
import numpy as np
np.random.seed(1)
dur = np.random.randint(30, 120, 30)
df = pd.DataFrame({
    "Duration": dur,
    "Pulse": 80 + dur * 0.3 + np.random.randn(30) * 5,
    "Maxpulse": np.random.randint(130, 180, 30),
    "Calories": dur * 4 + np.random.randn(30) * 20})
corr = df.corr()
print(corr.round(2))
print(df["Duration"].corr(df["Calories"]))
print(df["Duration"].corr(df["Maxpulse"]))
import matplotlib.pyplot as plt
plt.imshow(corr, cmap="coolwarm"); plt.colorbar()
plt.xticks(range(4), corr.columns, rotation=45); plt.yticks(range(4), corr.columns)
plt.title("Correlation matrix"); plt.show()
```
