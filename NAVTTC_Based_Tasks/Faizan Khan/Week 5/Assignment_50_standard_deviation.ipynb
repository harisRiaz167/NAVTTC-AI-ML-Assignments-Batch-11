```python
import numpy as np
speed = [86, 87, 88, 86, 87, 85, 86]
print("Std Dev:", np.std(speed), "Variance:", np.var(speed))
speed = [32, 111, 138, 28, 59, 77, 97]
print("Std Dev:", np.std(speed), "Variance:", np.var(speed))
print("Sample std:", np.std(speed, ddof=1))
# manual
m = np.mean(speed)
print("Manual:", (sum((x - m) ** 2 for x in speed) / len(speed)) ** 0.5)
```
