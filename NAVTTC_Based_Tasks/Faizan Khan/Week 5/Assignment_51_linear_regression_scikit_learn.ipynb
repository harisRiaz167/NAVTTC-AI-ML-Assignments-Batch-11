```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

# Simple linear regression: hours studied vs score
np.random.seed(0)
hours = np.random.uniform(1, 10, 100)
score = 10 + 8 * hours + np.random.randn(100) * 5
df = pd.DataFrame({"Hours": hours, "Scores": score})
X, y = df[["Hours"]], df["Scores"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

reg = LinearRegression().fit(X_train, y_train)
print("Intercept:", reg.intercept_, "Slope:", reg.coef_[0])
y_pred = reg.predict(X_test)
print(pd.DataFrame({"Actual": y_test.values, "Predicted": y_pred}).head())
print("MAE :", metrics.mean_absolute_error(y_test, y_pred))
print("MSE :", metrics.mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, y_pred)))
print("R2  :", metrics.r2_score(y_test, y_pred))
plt.scatter(X_test, y_test); plt.plot(X_test, y_pred, color="red")
plt.xlabel("Hours"); plt.ylabel("Score"); plt.show()
```

```python
# Multiple linear regression (diabetes dataset)
from sklearn.datasets import load_diabetes
d = load_diabetes(as_frame=True)
X, y = d.data, d.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
mlr = LinearRegression().fit(X_train, y_train)
print(pd.DataFrame(mlr.coef_, X.columns, columns=["Coefficient"]))
p = mlr.predict(X_test)
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, p)), "R2:", metrics.r2_score(y_test, p))
```
