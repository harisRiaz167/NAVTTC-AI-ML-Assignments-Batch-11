```python
count = 0
while count < 3:
    count += 1
    print("Hello Geek")
# while-else
i = 0
while i < 3:
    print(i); i += 1
else:
    print("while condition false")
# factorial
n, fact = 5, 1
while n > 0:
    fact *= n; n -= 1
print("5! =", fact)
# sum of digits
num, s = 12345, 0
while num:
    s += num % 10; num //= 10
print("Digit sum:", s)
# reverse number
num, rev = 1234, 0
while num:
    rev = rev * 10 + num % 10; num //= 10
print("Reversed:", rev)
```
