```python
# enumerate()
for i, v in enumerate(['tic', 'tac', 'toe']): print(i, v)
# zip()
q = ['name', 'quest', 'favorite color']; a = ['lancelot', 'the holy grail', 'blue']
for x, y in zip(q, a): print('What is your {0}? It is {1}.'.format(x, y))
# reversed()
for i in reversed(range(1, 10, 2)): print(i, end=" ")
print()
# sorted()
basket = ['apple', 'orange', 'apple', 'pear', 'orange', 'banana']
for f in sorted(set(basket)): print(f)
# items() for dict
knights = {'gallahad': 'the pure', 'robin': 'the brave'}
for k, v in knights.items(): print(k, v)
# list comprehension
print([x ** 2 for x in range(6)])
# itertools
import itertools
for x in itertools.chain([1, 2], [3, 4]): print(x, end=" ")
print()
for p in itertools.product("AB", repeat=2): print(p, end=" ")
```
