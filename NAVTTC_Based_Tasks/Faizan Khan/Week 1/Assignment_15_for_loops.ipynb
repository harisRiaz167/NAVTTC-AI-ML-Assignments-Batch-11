```python
# basic
for i in range(5):
    print(i, end=" ")
print()
# over list, string, tuple, dict
for x in ["apple", "banana", "cherry"]: print(x)
for ch in "Python": print(ch, end="-")
print()
d = {"a": 1, "b": 2}
for k, v in d.items(): print(k, v)
# range with start/stop/step
for i in range(1, 10, 2): print(i, end=" ")
print()
# enumerate & zip
for i, v in enumerate(["x", "y", "z"]): print(i, v)
for a, b in zip([1, 2, 3], ["one", "two", "three"]): print(a, b)
# nested loop - pattern
for i in range(1, 5):
    print("*" * i)
# for-else
for i in range(3): print(i)
else: print("Loop finished")
# sum 1..100
total = 0
for i in range(1, 101): total += i
print("Sum:", total)
```
