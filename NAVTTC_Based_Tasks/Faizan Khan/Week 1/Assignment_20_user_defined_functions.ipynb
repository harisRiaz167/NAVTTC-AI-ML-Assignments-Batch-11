```python
def evenOdd(x):
    return "even" if x % 2 == 0 else "odd"
print(evenOdd(2), evenOdd(3))

# default args
def greet(name, msg="Hello"):
    return f"{msg}, {name}!"
print(greet("Ali"), greet("Sara", "Hi"))

# keyword args
def student(firstname, lastname):
    print(firstname, lastname)
student(lastname='Khan', firstname='Ali')

# *args and **kwargs
def myFun(*args, **kwargs):
    print("args:", args, "| kwargs:", kwargs)
myFun('Hello', 'to', first='Geeks', mid='for', last='Geeks')

# pass by reference
def myFun2(x): x[0] = 20
lst = [10, 11, 12]; myFun2(lst); print(lst)

# lambda & recursion
square = lambda x: x * x
print(square(7))
def fact(n): return 1 if n <= 1 else n * fact(n - 1)
print(fact(6))

# multiple returns
def fun():
    return 1, 2, 3
a, b, c = fun(); print(a, b, c)
```
