```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy(); arr[0] = 42
print("orig:", arr, "copy:", x)
arr = np.array([1, 2, 3, 4, 5])
y = arr.view(); arr[0] = 42
print("orig:", arr, "view:", y)
y[1] = 99; print("orig after view change:", arr)
print(x.base, y.base)
print(arr.copy().base is None, arr.view().base is arr)
```
