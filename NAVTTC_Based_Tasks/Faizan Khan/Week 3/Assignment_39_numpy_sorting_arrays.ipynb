```python
import numpy as np
print(np.sort(np.array([3, 2, 0, 1])))
print(np.sort(np.array(['banana', 'cherry', 'apple'])))
print(np.sort(np.array([True, False, True])))
print(np.sort(np.array([[3, 2, 4], [5, 0, 1]])))
print(np.sort(np.array([3, 1, 2]))[::-1], np.argsort(np.array([3, 1, 2])))
print(np.sort(np.array([[3, 2, 4], [5, 0, 1]]), axis=0))
```
