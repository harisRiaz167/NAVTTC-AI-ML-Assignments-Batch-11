```python
import numpy as np
arr = np.array([1, 2, 3, 4]); print(arr.dtype)
print(np.array(['apple', 'banana', 'cherry']).dtype)
print(np.array([1, 2, 3, 4], dtype='S'), np.array([1, 2, 3, 4], dtype='i4').dtype)
print(np.array([1.1, 2.1, 3.1]).astype('i'))
print(np.array([1, 0, 3]).astype(bool))
print(np.array([1, 2, 3], dtype=np.float64).dtype, np.array([1, 2], dtype=complex))
try: np.array(['a', '2', '3'], dtype='i')
except ValueError as e: print("ValueError:", e)
```
