```python
import numpy as np
a = np.arange(1, 13)
print(a.reshape(4, 3)); print(a.reshape(2, 3, 2))
print(a.reshape(2, -1).shape, a.reshape(-1, 4).shape)
print(a.reshape(3, 4).base is not None)   # reshape returns a view
print(a.reshape(3, 4).flatten(), a.reshape(3, 4).ravel())
print(a.reshape(3, 4).T)
```
