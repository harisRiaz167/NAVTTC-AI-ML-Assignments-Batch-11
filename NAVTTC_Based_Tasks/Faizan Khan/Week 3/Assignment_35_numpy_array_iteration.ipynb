```python
import numpy as np
for x in np.array([1, 2, 3]): print(x)
a2 = np.array([[1, 2, 3], [4, 5, 6]])
for row in a2:
    for x in row: print(x, end=" ")
print()
for x in np.nditer(a2): print(x, end=" ")
print()
for x in np.nditer(a2[:, ::2]): print(x, end=" ")
print()
for x in np.nditer(np.array([1, 2, 3]), flags=['buffered'], op_dtypes=['S']): print(x)
for idx, x in np.ndenumerate(a2): print(idx, x)
```
