```python
import numpy as np
from numpy import random
np.random.seed(42)
print(random.randint(100), random.rand(), random.rand(3), random.rand(2, 3))
print(random.randint(100, size=5), random.randint(100, size=(2, 3)))
print(random.choice([3, 5, 7, 9]), random.choice([3, 5, 7, 9], size=(2, 3)))
print(random.choice([1, 2, 3, 4], p=[0.1, 0.3, 0.6, 0.0], size=10))
x = np.array([1, 2, 3, 4, 5]); random.shuffle(x); print(x)
print(random.permutation(np.array([1, 2, 3, 4, 5])))
print(random.normal(size=(2, 3)), random.binomial(n=10, p=0.5, size=5), random.poisson(2, 5))
print(random.uniform(size=3), random.exponential(size=3))
```
