```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5]); print(arr, type(arr), np.__version__)
print(np.array((1, 2, 3)))
print(np.array(42).ndim, np.array([1, 2, 3]).ndim,
      np.array([[1, 2], [3, 4]]).ndim, np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]]).ndim)
print(np.array([1, 2, 3, 4], ndmin=5))
print(np.zeros((2, 3)), np.ones((2, 2)), np.full((2, 2), 7), np.eye(3))
print(np.arange(0, 10, 2), np.linspace(0, 1, 5), np.empty((2, 2)).shape)
```
