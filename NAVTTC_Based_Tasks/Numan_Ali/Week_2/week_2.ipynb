User-Defined Functions in Python

```python
# Function of calculating area
def calculate_area(length, width=5):
    """Calculates the area of a rectangle."""
    return length * width

# Function accepting arbitrary arguments
def print_summary(*scores, **student_info):
    name = student_info.get("name", "Student")
    avg_score = sum(scores) / len(scores) if scores else 0
    return f"{name}'s average score across {len(scores)} subjects: {avg_score:.2f}"

print("Area (default width):", calculate_area(10))        
print("Area (custom width):", calculate_area(10, 8))      
print(print_summary(85, 90, 78, name="Alice"))
```

Exception Handling in Python

```python
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
        return None
    except TypeError:
        print("Error: Inputs must be numerical values.")
        return None
    else:
        print("Division successful.")
        return result
    finally:
        print("Execution complete.\n")

print("Result 1:", safe_divide(10, 2))
print("Result 2:", safe_divide(10, 0))
print("Result 3:", safe_divide("10", 2))
```

Assignment_27 Day_3

User-Defined Exception Handling in Python

```python
class InsufficientFundsError(Exception):
    # Raised when an account withdrawal exceeds the available balance.
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw ${amount}. Current balance is ${balance}.")

class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)
        self.balance -= amount
        return self.balance

account = BankAccount(100)

try:
    account.withdraw(150)
except InsufficientFundsError as e:
    print(f"Caught Custom Exception: {e}")
```

Statistics: Mean, Median, and Mode


```python
import statistics

data = [15, 18, 20, 20, 22, 25, 30]

mean_val = statistics.mean(data)
median_val = statistics.median(data)
mode_val = statistics.mode(data)

print(f"Mean:   {mean_val:.2f}")
print(f"Median: {median_val}")
print(f"Mode:   {mode_val}")
```

Statistics: Standard Deviation

```python
import statistics

data = [15, 18, 20, 20, 22, 25, 30]

pop_std = np.std(data)

sample_std = statistics.stdev(data)

print(f"Population Standard Deviation: {pop_std:.2f}")
print(f"Sample Standard Deviation:     {sample_std:.2f}")
```
