Assignment 53: Decision Tree Classification

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Synthetic Dataset representing Diabetes diagnostic measurements
data = {
    'Pregnancies': [6, 1, 8, 1, 0, 5, 3, 10, 2, 8, 4, 10],
    'Glucose': [148, 85, 183, 89, 137, 116, 78, 115, 197, 125, 110, 168],
    'BloodPressure': [72, 66, 64, 66, 40, 74, 50, 0, 70, 96, 92, 74],
    'BMI': [33.6, 26.6, 23.3, 28.1, 43.1, 25.6, 31.0, 35.3, 30.5, 0.0, 37.6, 38.0],
    'Age': [50, 31, 32, 21, 33, 30, 26, 29, 53, 54, 30, 38],
    'Outcome': [1, 0, 1, 0, 1, 0, 0, 0, 1, 1, 0, 1]
}

df = pd.DataFrame(data)

# Features (X) and Target (y)
feature_cols = ['Pregnancies', 'Glucose', 'BloodPressure', 'BMI', 'Age']
X = df[feature_cols]
y = df['Outcome']

# Train-Test Split (80% Train, 20% Test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Instantiate and fit Decision Tree Classifier using Gini criterion
clf = DecisionTreeClassifier(criterion='gini', random_state=42)
clf.fit(X_train, y_train)

# Predict test set results
y_pred = clf.predict(X_test)

# Model Accuracy
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%")
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

Assignment 54: Support Vector Machines (SVM) with scikit-learn

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Synthetic dataset representing banknote features: [Variance, Skewness, Curtosis, Entropy]
data = {
    'Variance': [3.6216, 4.5459, 3.8660, 3.4566, 0.3292, -2.3978, -3.2000, -1.8000],
    'Skewness': [8.6661, 8.1674, -2.6383, 9.5228, -4.4552, -9.3000, -8.1000, -6.2000],
    'Curtosis': [-2.8073, -2.4586, 1.9242, -4.0112, 4.5718, 9.6000, 8.4000, 7.1000],
    'Entropy': [-0.44699, -1.46210, 0.10645, -3.59440, -0.98880, -0.1500, -0.2000, -0.5000],
    'Class': [0, 0, 0, 0, 1, 1, 1, 1]
}

df = pd.DataFrame(data)

# Features and Target
X = df[['Variance', 'Skewness', 'Curtosis', 'Entropy']]
y = df['Class']

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Standardize features (Crucial step for SVM performance)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Instantiate and fit Linear SVM
svclassifier = SVC(kernel='linear')
svclassifier.fit(X_train, y_train)

# Predict test set
y_pred = svclassifier.predict(X_test)

# Model evaluation
print(f"Accuracy: {accuracy_score(y_test, y_pred) * 100:.2f}%\n")
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
