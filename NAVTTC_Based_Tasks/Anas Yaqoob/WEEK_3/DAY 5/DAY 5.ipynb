```python
import pandas as pd

a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)
print(myvar[0])             # access by position/label
```

```python
# Custom labels with index
myvar = pd.Series(a, index=["x", "y", "z"])
print(myvar)
print(myvar["y"])
```

```python
# A Series from a dictionary: keys become labels
calories = {"day1": 420, "day2": 380, "day3": 390}
myvar = pd.Series(calories)
print(myvar)

# Keep only some of the items
print(pd.Series(calories, index=["day1", "day2"]))
```

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data)
print(df)
```

```python
# loc selects rows by label
print(df.loc[0])             # one row (returns a Series)
print(df.loc[[0, 1]])        # several rows (returns a DataFrame)
```

```python
# Named index
df = pd.DataFrame(data, index=["day1", "day2", "day3"])
print(df)
print(df.loc["day2"])
```

```python
print("shape:", df.shape)       # (rows, columns)
df.info()                      # column names, types and non-null counts
```
