```python
# 1. A simple function: define it once, call it many times
def greet():
    print("Hello from a function!")

greet()
greet()
```

```python
# 2. Parameters and a return value
def add(a, b):
    return a + b

result = add(5, 7)
print("5 + 7 =", result)
```

```python
# 3. Default parameter values
def power(base, exponent=2):
    return base ** exponent

print(power(5))        # uses the default exponent 2
print(power(2, 5))     # overrides the default
```

```python
# 4. Keyword arguments (the order does not matter)
def student_info(name, age, city):
    print(f"{name} is {age} years old and lives in {city}")

student_info(age=22, city="Lahore", name="Ali")
```

```python
# 5. *args (any number of values) and **kwargs (any number of name=value pairs)
def total(*numbers):
    return sum(numbers)

def show_profile(**details):
    for key, value in details.items():
        print(key, "->", value)

print(total(1, 2, 3, 4))
show_profile(name="Ali", course="AI", city="Lahore")
```

```python
# 6. Returning more than one value
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([4, 9, 1, 7])
print("Min:", low, "Max:", high)
```

```python
# 7. Lambda: a small anonymous function
square = lambda x: x * x
print(square(6))
print(list(map(lambda x: x * 2, [1, 2, 3])))
```

```python
# 8. Recursion: a function that calls itself
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print("5! =", factorial(5))
```

```python
# 9. Local and global variables
x = "global"

def show():
    x = "local"          # a separate variable that exists only inside the function
    print("inside :", x)

show()
print("outside:", x)
```

```python
# 10. Docstring and type hints
def area_of_circle(radius: float) -> float:
    """Return the area of a circle with the given radius."""
    return 3.14159 * radius ** 2

print(area_of_circle(3))
print(area_of_circle.__doc__)
```

```python
# 1. try / except: handle a division by zero
try:
    result = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero")
```

```python
# 2. Handle different exceptions separately
for value in ["25", "abc", None]:
    try:
        number = int(value)
        print("Converted:", number)
    except ValueError:
        print(repr(value), "is not a valid number")
    except TypeError:
        print(repr(value), "has the wrong type")
```

```python
# 3. else and finally
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: division by zero")
    else:
        print("Result:", result)             # runs only when no exception happened
    finally:
        print("Done (finally always runs)")

divide(10, 2)
divide(10, 0)
```

```python
# 4. Catch the exception object to see its message
try:
    numbers = [1, 2, 3]
    print(numbers[5])
except IndexError as e:
    print("Caught:", e)
```

```python
# 5. A file that does not exist
try:
    with open("missing_file.txt") as f:
        print(f.read())
except FileNotFoundError:
    print("The file does not exist")
```

```python
# 6. Raise an exception yourself with raise
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    return age

try:
    check_age(-5)
except ValueError as e:
    print("Error:", e)
```

```python
# 7. Several exception types in one handler
try:
    x = int("10")
    y = 5 / 0
except (ValueError, ZeroDivisionError) as e:
    print("Handled:", type(e).__name__, "-", e)
```

```python
# 1. The simplest user-defined exception
class MyError(Exception):
    pass

try:
    raise MyError("Something went wrong")
except MyError as e:
    print("Caught MyError:", e)
```

```python
# 2. A custom exception that carries extra data (salary example)
class SalaryNotInRangeError(Exception):
    """Raised when the salary is outside the allowed range."""
    def __init__(self, salary, message="Salary is not in range (5000 to 20000)"):
        self.salary = salary
        self.message = message
        super().__init__(self.message)

    def __str__(self):
        return f"{self.salary} -> {self.message}"

salary = 3000
try:
    if not 5000 < salary < 20000:
        raise SalaryNotInRangeError(salary)
except SalaryNotInRangeError as e:
    print("Error:", e)
```

```python
# 3. Bank example: InsufficientBalanceError
class InsufficientBalanceError(Exception):
    def __init__(self, balance, amount):
        self.balance = balance
        self.amount = amount
        super().__init__(f"Cannot withdraw {amount}; balance is only {balance}")

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError(balance, amount)
    return balance - amount

try:
    print(withdraw(1000, 400))
    print(withdraw(600, 900))
except InsufficientBalanceError as e:
    print("Error:", e)
```

```python
# 4. A hierarchy: catching the base class also catches its children
class AppError(Exception):
    pass

class DatabaseError(AppError):
    pass

class NetworkError(AppError):
    pass

for err in (DatabaseError("DB is down"), NetworkError("No internet")):
    try:
        raise err
    except AppError as e:
        print(type(e).__name__, ":", e)
```

```python
# Data: speeds of 13 cars
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
print(len(speed), "values")
```

```python
# Mean = the average value
import numpy as np

print("Mean (by hand):", sum(speed) / len(speed))
print("Mean (NumPy)  :", np.mean(speed))
```

```python
# Median = the middle value after sorting
print("Median (NumPy):", np.median(speed))

ordered = sorted(speed)
print(ordered)
print("Middle value :", ordered[len(ordered) // 2])    # 13 values, so the 7th one
```

```python
# Mode = the value that appears most often
from statistics import mode, multimode

print("Mode     :", mode(speed))
print("All modes:", multimode(speed))     # the value(s) that appear the most
print("86 appears", speed.count(86), "times")
```

```python
# Standard deviation tells how spread out the values are
import numpy as np

speed = [86, 87, 88, 86, 87, 85, 86]
print("Std:", np.std(speed))
```

```python
# A more spread-out set of values gives a larger standard deviation
speed2 = [32, 111, 138, 28, 59, 77, 97]
print("Std:", np.std(speed2))
```

```python
# By hand: variance is the average of the squared distances from the mean,
# and standard deviation is the square root of the variance
mean = sum(speed) / len(speed)
variance = sum((x - mean) ** 2 for x in speed) / len(speed)

print("Mean     :", mean)
print("Variance :", variance)
print("Std dev  :", variance ** 0.5)
print("NumPy var:", np.var(speed))
```

```python
# Population vs sample standard deviation
print("Population (default, ddof=0):", np.std(speed))
print("Sample     (ddof=1)         :", np.std(speed, ddof=1))
```
