```python
# Data: speeds of 13 cars
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
print(len(speed), "values")
```

```python
# Mean = the average value
import numpy as np

print("Mean (by hand):", sum(speed) / len(speed))
print("Mean (NumPy)  :", np.mean(speed))
```

```python
# Median = the middle value after sorting
print("Median (NumPy):", np.median(speed))

ordered = sorted(speed)
print(ordered)
print("Middle value :", ordered[len(ordered) // 2])    # 13 values, so the 7th one
```

```python
# Mode = the value that appears most often
from statistics import mode, multimode

print("Mode     :", mode(speed))
print("All modes:", multimode(speed))     # the value(s) that appear the most
print("86 appears", speed.count(86), "times")
```
