## 1. mkdir, cd, pwd, ls

Create a practice folder, go inside it, check where you are and list its contents.

```python
%%bash
rm -rf /tmp/linux_practice          # start clean
mkdir /tmp/linux_practice           # mkdir: create a folder
cd /tmp/linux_practice              # cd: go inside it
pwd                                 # pwd: where am I?
ls -la                              # ls -la: list everything, including hidden entries
```

## 2. touch, redirection, cat

`>` writes to a file (replacing it), `>>` appends, `<` reads input from a file, and `|` pipes the output of one command into another.

```python
%%bash
cd /tmp/linux_practice
touch empty.txt                     # touch: create an empty file
echo "Hello Linux" > a.txt          # > : write (overwrites)
echo "Second line" >> a.txt         # >> : append
cat a.txt                           # cat: show the file
echo "--- sort < a.txt (input redirection) ---"
sort < a.txt
echo "--- ls | wc -l (pipe) ---"
ls | wc -l                          # how many files are in this folder
```

## 3. cp, mv, ls -l, file

Copy a file, rename the copy, list with details and check the file type.

```python
%%bash
cd /tmp/linux_practice
cp a.txt b.txt                      # cp: copy
mv b.txt c.txt                      # mv: rename (or move)
ls -l                               # long listing
file a.txt                          # file: what type is it?
```

## 4. head (and more / less)

`head` shows the first lines. `more` and `less` open a file page by page. They are interactive, so run them yourself in a terminal: `more numbers.txt` (Space = next page) and `less numbers.txt` (arrow keys to scroll, `q` to quit).

```python
%%bash
cd /tmp/linux_practice
seq 1 20 > numbers.txt              # create a file containing 1..20
head -n 5 numbers.txt               # head: first 5 lines
```

## 5. reading input (read)

`read` stores a line of input in a variable. Here the input is supplied by a pipe so the cell does not wait for you.

```python
%%bash
echo "Ali" | { read name; echo "Hello $name"; }
```

## 6. alias

An alias is a shortcut for a longer command. (`shopt -s expand_aliases` is needed only because notebook cells run non-interactively; in a normal terminal just type the `alias` line.)

```python
%%bash
shopt -s expand_aliases
cd /tmp/linux_practice
alias ll='ls -l'
ll
```

## 7. chmod, bash, sh

Create a small script, make it executable with `chmod`, and run it three different ways.

```python
%%bash
cd /tmp/linux_practice
echo 'echo "Script is running"' > script.sh
ls -l script.sh                     # no execute permission yet
chmod +x script.sh                  # chmod: add execute permission
ls -l script.sh                     # now has x
./script.sh                         # run directly
bash script.sh                      # run with bash
sh script.sh                        # run with sh
```

## 8. ps and kill

Start a harmless background process (`sleep`), find it with `ps`, stop it with `kill`.

```python
%%bash
sleep 300 &                         # start a background process
PID=$!
echo "Started process $PID"
ps -p $PID                          # ps: is it running?
kill $PID                           # kill: stop it
sleep 1
ps -p $PID || echo "Process $PID has stopped"
```

## 9. dpkg

`dpkg` manages `.deb` packages on Debian/Ubuntu (it will not exist on macOS). Here we only *query* it, nothing is installed or removed.

```python
%%bash
dpkg --version | head -n 1          # which dpkg version
echo "--- first 5 files installed by the bash package ---"
dpkg -L bash | head -n 5
echo "--- package details ---"
dpkg -s bash | head -n 4
```

## 10. rm and rmdir (cleanup)

`rmdir` removes only an **empty** folder. `rm -r` removes a folder and everything inside it, so use it carefully.

```python
%%bash
cd /tmp/linux_practice
mkdir emptydir
rmdir emptydir                      # rmdir: works because it is empty
rm c.txt                            # rm: delete one file
ls
cd /tmp
rm -r /tmp/linux_practice           # rm -r: delete the folder and its contents
ls /tmp/linux_practice 2>&1 | head -n 1
```

```python
import cmath

a = 1
b = 5
c = 6

d = (b ** 2) - (4 * a * c)                   # discriminant

sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print("The solutions are {0} and {1}".format(sol1, sol2))
```

```python
# 1. Skip the letters 'e' and 's'
for letter in "geeksforgeeks":
    if letter == "e" or letter == "s":
        continue
    print(letter, end=" ")
print()
```

```python
# 2. Print only odd numbers from 1 to 10
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

```python
# 3. continue in a while loop (skip 3)
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i, end=" ")
print()
```

```python
fruits = ["apple", "banana", "cherry"]
prices = [120, 60, 200]
```

```python
# 1. enumerate(): index + value
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
# 2. zip(): loop over two lists together
for fruit, price in zip(fruits, prices):
    print(fruit, "costs", price)
```

```python
# 3. reversed(): loop backwards
for fruit in reversed(fruits):
    print(fruit, end=" ")
print()
```

```python
# 4. sorted(): loop in sorted order
for p in sorted(prices):
    print(p, end=" ")
print()
```
