```python
num = 8

sqrt = num ** 0.5          # power of 1/2 gives the square root
print("The square root of %0.3f is %0.3f" % (num, sqrt))
```

```python
import cmath

num = 1 + 2j
sqrt = cmath.sqrt(num)

print("The square root of {0} is {1:0.3f}+{2:0.3f}j".format(num, sqrt.real, sqrt.imag))
```

```python
celsius = 37.5

# Formula: F = (C x 9/5) + 32
fahrenheit = (celsius * 1.8) + 32
print("%0.1f degree Celsius is equal to %0.1f degree Fahrenheit" % (celsius, fahrenheit))
```

```python
a = 5
b = 6
c = 7

s = (a + b + c) / 2                          # semi-perimeter
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print("The area of the triangle is %0.2f" % area)
```

```python
import cmath

a = 1
b = 5
c = 6

d = (b ** 2) - (4 * a * c)                   # discriminant

sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print("The solutions are {0} and {1}".format(sol1, sol2))
```

```python
kilometers = float(input("Enter value in kilometers: "))

conv_fac = 0.621371
miles = kilometers * conv_fac

print("%0.2f kilometers is equal to %0.2f miles" % (kilometers, miles))
```

```python
text = "Hello, Python"

print(text)                       # Hello, Python
print(len(text))                  # length of the string
print(text[0], text[-1])          # first and last character
print(text[0:5])                  # slicing -> Hello
print(text[7:])                   # slicing from index 7
print(text[::-1])                 # reverse the string
print(text.upper(), text.lower())
print(text.replace("Python", "World"))
print(text.split(","))            # split into a list
print("-".join(["a", "b", "c"]))  # join a list into a string
print("   spaces   ".strip())     # remove leading/trailing spaces
print(text.find("Python"))        # index where the word starts
print(text.count("l"))            # how many times "l" appears
print(text.startswith("Hello"), text.endswith("!"))
print("Py" in text)               # membership test

name = "Ali"
age = 22
print(f"My name is {name} and I am {age} years old.")   # f-string

for ch in "abc":                  # iterate over characters
    print(ch, end=" ")
print()
print("Line1\nLine2")             # escape characters
```

**How it works:** Strings are immutable sequences of characters. Methods such as `upper()` and `replace()` return a new string.
