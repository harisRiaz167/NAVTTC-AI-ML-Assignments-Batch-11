**Assignment 8**

Topic: Python
Write a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).

```python
a, b, c = 5, 6, 7
s = (a + b + c) / 2
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print(f"Semi-perimeter is: {s}")
print(f"Area of the triangle is: {area}")
```

**Assignment 9**

Topic: Python
Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).

```python
import cmath

a = 1
b = 5
c = 6

d = (b**2) - (4 * a * c)

root1 = (-b - cmath.sqrt(d)) / (2 * a)
root2 = (-b + cmath.sqrt(d)) / (2 * a)

print(f"First root is: {root1}")
print(f"Second root is: {root2}")
```

**Assignment 19**

Topic: Python
Complete the exercise on various looping techniques in Python. Reference: https://www.geeksforgeeks.org/looping-techniques-python/
Objective: Practice different ways of iterating over Python collections.

```python
# 1. Loop through a list
numbers = [10, 20, 30]
for n in numbers:
    print(n)

# 2. Loop using enumerate()
fruits = ["Apple", "Mango", "Banana"]
for index, fruit in enumerate(fruits):
    print(index, fruit)

# 3. Loop through two lists using zip()
names = ["Ali", "Sara"]
marks = [85, 90]
for name, mark in zip(names, marks):
    print(name, mark)

# 4. Loop in reverse order
for n in reversed(numbers):
    print(n)

# 5. Loop through a dictionary using items()
student_info = {"name": "Mahnoor", "batch": 11, "program": "AI/ML"}
for key, value in student_info.items():
    print(key, value)

# 6. Loop in sorted order
scores = [45, 12, 89, 33]
for s in sorted(scores):
    print(s)
```

**Assignment 24**

Topic: Python
Complete the exercise on the set data type in Python. Reference: https://www.programiz.com/python-programming/set
Objective: Create sets and perform basic set operations.

```python
set_a = {1, 2, 3, 4, 5, 2, 3}
set_b = {4, 5, 6, 7, 8}

print("Set A:", set_a)
print("Set B:", set_b)

# 2. Adding and removing elements
set_a.add(6)
set_a.remove(1)
print("(Updated Set A after add/remove):", set_a)

# 3. Basic Set Operations
# Union
print("Union:", set_a | set_b)

# Intersection
print("Intersection:", set_a & set_b)

# Difference 
print("Difference:", set_a - set_b)

# Symmetric Difference
print("Symmetric Difference:", set_a ^ set_b)
```
