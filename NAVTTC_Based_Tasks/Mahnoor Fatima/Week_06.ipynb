**Assignment 53**

Topic: Machine Learning (Decision Trees)
Objective: Implement a Decision Tree classifier using scikit-learn to predict categories based on input features.

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X = np.array([
    [2, 50],
    [3, 60],
    [5, 80],
    [6, 85],
    [8, 95],
    [1, 40],
    [7, 90]
])
y = np.array([0, 0, 1, 1, 1, 0, 1])

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

new_students = np.array([[4, 70], [7, 92]])
predictions = model.predict(new_students)

print("Decision Tree Predictions:", predictions)
print("Accuracy:", round(accuracy_score(y, model.predict(X)), 2))
```

**Assignment 54**

Machine LearningTask statement: Complete the exercise on Support Vector Machines (SVM).Reference: https://stackabuse.com/implementing-svm-and-kernel-svm-with-pythons-scikit-learn/

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.datasets import load_iris

iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

svclassifier = SVC(kernel='linear', random_state=42)
svclassifier.fit(X_train, y_train)

y_pred = svclassifier.predict(X_test)

print("--- Confusion Matrix ---")
print(confusion_matrix(y_test, y_pred))

print("\n--- Classification Report ---")
print(classification_report(y_test, y_pred))
```

Topic: Time-Series Analysis
Objective: Create a time-series index, resample data over time intervals, and compute rolling averages using Pandas.

```python
import pandas as pd
import numpy as np

dates = pd.date_range(start="2026-10-01", periods=10, freq="D")
sales_data = {
    "Sales": [120, 135, 125, 140, 160, 155, 170, 185, 180, 195]
}

df = pd.DataFrame(sales_data, index=dates)
print("--- Original Time-Series DataFrame ---")
print(df)

df["Rolling_Avg_5D"] = df["Sales"].rolling(window=5).mean()

print("\n--- After Adding 5-Day Rolling Average ---")
print(df)
```
