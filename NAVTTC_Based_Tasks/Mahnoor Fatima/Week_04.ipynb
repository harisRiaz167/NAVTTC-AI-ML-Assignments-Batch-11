**Assignment 47**

Topic: Pandas
Complete the exercise on Pandas data-cleaning techniques. Reference: https://www.w3schools.com/python/pandas_cleaning.asp
Objective: To identify and handle missing values, remove duplicate rows, and correct data in a DataFrame.

```python
import pandas as pd
import numpy as np



data = {
    "Student_ID": [101, 102, 103, 103, 105, 106, 107], # Duplicate ID 103
    "Name": ["Aaliyah", "Bilal", None, "Bilal", "Eman", "Fahad", "Hassan"], # Missing Name (NaN)
    "Age": [20, 21, 19, 21, 150, 20, 21], # Incorrect Age '150' (outlier/typo)
    "Score": [88, 74, 92, 74, 65, None, 95] # Missing Score and a duplicate row
}

df = pd.DataFrame(data)
print("--- 1. ORIGINAL MESSY DATAFRAME ---")
print(df)


print("\n--- 2. DETECTING NULL VALUES ---")
print(df.isnull())

df["Name"] = df["Name"].fillna("Unknown")
mean_score = df["Score"].mean()
df["Score"] = df["Score"].fillna(mean_score)

print("\n--- After Handling Missing Values ---")
print(df)

print(f"\nNumber of duplicate rows found: {df.duplicated().sum()}")
df = df.drop_duplicates()

print("\n--- After Removing Duplicates ---")
print(df)


df.loc[df["Age"] > 100, "Age"] = 22

print("\n--- After Correcting Wrong Data (Age > 100 fixed) ---")
print(df)
```

Topic: Pandas
Complete the exercise on data correlation with Pandas. Reference: https://www.w3schools.com/python/pandas_correlations.asp
Objective: To calculate the correlation between numerical columns in a DataFrame and understand how two variables are related.

```python
import pandas as pd


data = {
    "Study_Hours": [2, 4, 6, 8, 10, 12, 3, 5, 7, 9],
    "Attendance_Rate": [50, 60, 75, 85, 95, 98, 55, 70, 80, 90],
    "Exam_Score": [55, 63, 76, 88, 94, 99, 58, 71, 82, 91],
    "Random_Values": [42, 19, 85, 33, 91, 10, 67, 54, 22, 78] 
}

df = pd.DataFrame(data)

print("--- 1. ORIGINAL DATAFRAME (First 5 rows) ---")
print(df.head())


correlation_matrix = df.corr()

print("\n--- 2. CORRELATION MATRIX ---")
print(correlation_matrix)

study_score_corr = df["Study_Hours"].corr(df["Exam_Score"])
print(f"\n--- 3. SPECIFIC CORRELATION ---")
print(f"Correlation between Study Hours and Exam Score: {study_score_corr:.2f}")
```
