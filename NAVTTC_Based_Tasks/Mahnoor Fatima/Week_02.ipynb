**Assignment 27**

Topic: Python
Complete the exercise on user-defined exception handling in Python. Reference: https://www.programiz.com/python-programming/user-defined-exception
Objective: Create a custom exception and raise it when a user's age is below 18.

```python
class InvalidAgeError(Exception):
    pass

age = 15
try:
    if age < 18:
        raise InvalidAgeError("Age must be at least 18.")
except InvalidAgeError as e:
    print("Custom Exception:", e)
```

**Assignment 49**

Topic: Statistics
Calculate mean, median, and mode. Reference: https://www.w3schools.com/python/python_ml_mean_median_mode.asp
Objective: Calculate the mean, median, and mode of a dataset using Python's statistics module.

```python
import statistics

data = [10, 20, 20, 30, 40]
print("Mean:", statistics.mean(data))
print("Median:", statistics.median(data))
print("Mode:", statistics.mode(data))
```

**Assignmnet 50**

Topic: Statistics
Calculate standard deviation. Reference: https://www.w3schools.com/python/python_ml_standard_deviation.asp
Objective: Calculate the standard deviation of a dataset using Python.

```python
import statistics
data = [10, 20, 20, 30, 40, 50]
std_deviation = statistics.pstdev(data)
print("Data:", data)
print("Standard deviation:", std_deviation)
```
