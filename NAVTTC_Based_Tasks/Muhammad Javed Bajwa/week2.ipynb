Task 20: User-Defined Functions in Python

```python
# Task 20: Implementation of User-Defined Functions with default args and docstrings

def calculate_metrics(y_true, y_pred):
    """
    Computes basic Mean Absolute Error (MAE) and accuracy.
    """
    total = len(y_true)
    correct = sum(yt == yp for yt, yp in zip(y_true, y_pred))
    mae = sum(abs(yt - yp) for yt, yp in zip(y_true, y_pred)) / total
    accuracy = correct / total
    return mae, accuracy

# Example usage
actual = [1, 0, 1, 1, 0]
predicted = [1, 0, 0, 1, 0]

mae, accuracy = calculate_metrics(actual, predicted)
print(f"MAE: {mae:.2f}")
print(f"Accuracy: {accuracy * 100:.1f}%")
```

Task 26: Exception Handling in Python

```python
# Task 26: Exception handling using try, except, else, and finally blocks

def safe_division(numerator, denominator):
    try:
        val1 = float(numerator)
        val2 = float(denominator)
        result = val1 / val2
    except ValueError as ve:
        print(f"Error: Invalid literal provided - {ve}")
        return None
    except ZeroDivisionError as zde:
        print(f"Error: Division by zero is undefined - {zde}")
        return None
    else:
        print("Division completed successfully.")
        return result
    finally:
        print("Execution completed.\n")

# Tests
print("Test 1 Result:", safe_division(10, 2))
print("Test 2 Result:", safe_division(10, 0))
print("Test 3 Result:", safe_division("abc", 5))
```

Task 27: User-Defined Exception Handling

```python
# Task 27: Custom Exception definition and handling

class NegativeValueError(Exception):
    """Raised when an input value is strictly negative."""
    def __init__(self, message="Value cannot be negative"):
        self.message = message
        super().__init__(self.message)

def configure_epochs(epochs):
    try:
        if epochs <= 0:
            raise NegativeValueError(f"Invalid epoch count: {epochs}. Must be greater than 0.")
        print(f"Model will train for {epochs} epochs.")
    except NegativeValueError as err:
        print(f"Caught Custom Exception: {err}")

# Tests
configure_epochs(25)
configure_epochs(-5)
```

Task 49: Statistics — Mean, Median, and Mode

```python
# Task 49: Calculating Central Tendency Measures using NumPy and SciPy
import numpy as np
from scipy import stats

# Dataset: e.g., Speed of vehicles or exam scores
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

mean_val = np.mean(speed)
median_val = np.median(speed)
mode_res = stats.mode(speed, keepdims=True)

print(f"Dataset: {speed}")
print(f"Mean   : {mean_val:.2f}")
print(f"Median : {median_val}")
print(f"Mode   : {mode_res.mode[0]} (count = {mode_res.count[0]})")
```

Task 50: Statistics — Standard Deviation and Variance

```python
# Task 50: Calculating Dispersion Measures using NumPy
import numpy as np

# Dataset: e.g., Speed of vehicles
speed = [86, 87, 88, 86, 87, 85, 86]

std_dev = np.std(speed)
variance = np.var(speed)

print(f"Dataset            : {speed}")
print(f"Standard Deviation : {std_dev:.4f}")
print(f"Variance           : {variance:.4f}")
```
