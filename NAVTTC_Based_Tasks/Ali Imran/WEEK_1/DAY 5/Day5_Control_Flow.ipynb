```python
i = 10

if i == 10:
    # first nested if
    if i < 10:
        print("i is smaller than 10")

    # second nested if
    if i < 12:
        print("i is smaller than 12 too")
else:
    print("i is not equal to 10")
```

```python
# 1. Loop over a list
for fruit in ["apple", "banana", "cherry"]:
    print(fruit)
```

```python
# 2. Loop over a string
for ch in "Python":
    print(ch, end=" ")
print()
```

```python
# 3. range(): 0 to 4
for i in range(5):
    print(i, end=" ")
print()
```

```python
# 4. range() with start, stop and step
for i in range(2, 11, 2):
    print(i, end=" ")
print()
```

```python
# 5. Sum of numbers 1 to 10
total = 0
for i in range(1, 11):
    total += i
print("Sum =", total)
```

```python
# 6. Multiplication table of 5
for i in range(1, 6):
    print("5 x", i, "=", 5 * i)
```

```python
# 7. Nested loop (star pattern)
for i in range(1, 4):
    for j in range(i):
        print("*", end="")
    print()
```

```python
# 8. for-else (else runs when the loop ends without break)
for i in range(3):
    print(i)
else:
    print("Loop finished")
```

```python
# 1. Count from 1 to 5
count = 1
while count <= 5:
    print(count, end=" ")
    count += 1
print()
```

```python
# 2. Countdown
n = 5
while n > 0:
    print(n, end=" ")
    n -= 1
print("Liftoff!")
```

```python
# 3. Sum of first 10 numbers
i, total = 1, 0
while i <= 10:
    total += i
    i += 1
print("Sum =", total)
```

```python
# 4. Factorial of 5
num, fact = 5, 1
while num > 0:
    fact *= num
    num -= 1
print("5! =", fact)
```

```python
# 5. while-else
x = 0
while x < 3:
    print("x =", x)
    x += 1
else:
    print("Condition became False")
```

```python
# 1. Stop a for loop when the letter 'e' is found
for letter in "geeksforgeeks":
    if letter == "e":
        break
    print(letter, end=" ")
print()
```

```python
# 2. Stop a while loop at 5
i = 1
while True:
    if i > 5:
        break
    print(i, end=" ")
    i += 1
print()
```

```python
# 3. Search a list and stop at the first match
numbers = [4, 8, 15, 16, 23, 42]
for n in numbers:
    if n % 5 == 0:
        print("First multiple of 5:", n)
        break
```

```python
# 1. Skip the letters 'e' and 's'
for letter in "geeksforgeeks":
    if letter == "e" or letter == "s":
        continue
    print(letter, end=" ")
print()
```

```python
# 2. Print only odd numbers from 1 to 10
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()
```

```python
# 3. continue in a while loop (skip 3)
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i, end=" ")
print()
```

```python
fruits = ["apple", "banana", "cherry"]
prices = [120, 60, 200]
```

```python
# 1. enumerate(): index + value
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

```python
# 2. zip(): loop over two lists together
for fruit, price in zip(fruits, prices):
    print(fruit, "costs", price)
```

```python
# 3. reversed(): loop backwards
for fruit in reversed(fruits):
    print(fruit, end=" ")
print()
```

```python
# 4. sorted(): loop in sorted order
for p in sorted(prices):
    print(p, end=" ")
print()
```

```python
# 5. Dictionary: items()
student = {"name": "Ali", "age": 22}
for key, value in student.items():
    print(key, "=", value)
```

```python
# 6. set with sorted() to get unique values in order
for x in sorted(set([3, 1, 2, 3, 1])):
    print(x, end=" ")
print()
```

```python
# 7. List comprehension
print([x * 2 for x in range(5)])
```
