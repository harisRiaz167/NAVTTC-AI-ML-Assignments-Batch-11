```python
fruits = ["apple", "banana", "cherry"]
print(fruits)

print(fruits[0], fruits[-1])          # indexing (positive / negative)
print(fruits[0:2])                    # slicing

fruits.append("mango")                # add at the end
fruits.insert(1, "orange")            # add at a position
fruits.extend(["kiwi", "grape"])      # add many items
print(fruits)

fruits.remove("banana")               # remove by value
last = fruits.pop()                   # remove and return the last item
del fruits[0]                         # remove by index
print(fruits, "| popped:", last)

numbers = [5, 2, 9, 1, 7]
numbers.sort()                        # ascending order
print(numbers)
numbers.reverse()
print(numbers)

print(len(numbers), max(numbers), min(numbers), sum(numbers))
print(numbers.count(9), numbers.index(7))
print(7 in numbers)                   # membership

squares = [x ** 2 for x in range(1, 6)]   # list comprehension
print(squares)

matrix = [[1, 2], [3, 4]]             # nested list
print(matrix[1][0])
```

```python
colors = ("red", "green", "blue")
print(colors)

single = ("only",)                    # one-element tuple needs a comma
print(type(single))

print(colors[0], colors[-1], colors[0:2])

try:
    colors[0] = "black"               # tuples are immutable
except TypeError as e:
    print("Error:", e)

more = colors + ("yellow",)           # concatenation creates a new tuple
print(more)
print(("a", "b") * 2)                 # repetition

nums = (1, 2, 2, 3, 2)
print(nums.count(2), nums.index(3))

x, y, z = colors                      # unpacking
print(x, y, z)

nested = (1, (2, 3), [4, 5])
print(nested[1][0])

as_list = list(colors)                # convert to list to change items
as_list[0] = "black"
print(tuple(as_list))
```

```python
nums = {1, 2, 3, 3, 2}                # duplicates are removed automatically
print(nums)

nums.add(4)
nums.update([5, 6])
print(nums)

nums.remove(6)                        # error if missing
nums.discard(100)                     # no error if missing
print(nums)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric difference:", a ^ b)

print({1, 2}.issubset(a))
print(3 in a)
print(len(a))

frozen = frozenset([1, 2, 3])         # immutable set
print(frozen)
```

```python
student = {"name": "Ali", "age": 22, "city": "Lahore"}
print(student)

print(student["name"])
print(student.get("marks", "not found"))   # safe access with default

student["age"] = 23                   # update
student["course"] = "AI"              # add a new key
print(student)

removed = student.pop("city")         # remove a key
print("Removed:", removed)

print(student.keys())
print(student.values())
print(student.items())

for key, value in student.items():    # loop through key-value pairs
    print(key, "->", value)

print("name" in student)              # membership checks keys

squares = {x: x ** 2 for x in range(1, 5)}   # dict comprehension
print(squares)

nested = {"a": {"x": 1}, "b": {"x": 2}}
print(nested["b"]["x"])
```
