```python
import pandas as pd
from io import StringIO

dirty_csv = """Duration,Date,Pulse,Maxpulse,Calories
60,2020/12/01,110,130,409.1
60,2020/12/02,117,145,479.0
60,2020/12/03,103,135,340.0
45,2020/12/04,109,175,282.4
45,2020/12/05,117,148,406.0
60,2020/12/06,102,127,300.0
60,2020/12/07,110,136,374.0
450,2020/12/08,104,134,253.3
30,2020/12/09,109,133,195.1
60,2020/12/10,98,124,269.0
60,2020/12/11,103,147,
60,2020/12/12,100,120,250.7
60,2020/12/12,100,120,250.7
45,20201213,90,112,
60,2020/12/14,103,123,323.0
"""

def load_dirty():
    return pd.read_csv(StringIO(dirty_csv))

df = load_dirty()
print(df)
```

```python
# 1. Find the problems first
df = load_dirty()
print("missing values per column:")
print(df.isnull().sum())
print("duplicate rows:", df.duplicated().sum())
```

```python
# 2. Empty cells: remove the rows ...
df = load_dirty()
no_empty = df.dropna()
print("rows before:", len(df), "| after dropna():", len(no_empty))
```

```python
# ... or fill them with a value (mean, median or mode)
df = load_dirty()
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
print(df["Calories"].isnull().sum(), "missing values left")
print(df[["Calories"]].tail(5))
```

```python
# 3. Wrong format: convert the Date column; values that do not match become NaT
df = load_dirty()
df["Date"] = pd.to_datetime(df["Date"], format="%Y/%m/%d", errors="coerce")
print(df[["Date"]].tail(4))
df = df.dropna(subset=["Date"])          # drop the rows whose date could not be read
print("rows left:", len(df))
```

```python
# 4. Wrong data: 450 minutes is clearly a mistake; replace values above a limit
df = load_dirty()
for row in df.index:
    if df.loc[row, "Duration"] > 120:
        df.loc[row, "Duration"] = 120
print(df["Duration"].max())
```

```python
# 5. Duplicates: find them and remove them
df = load_dirty()
print(df.duplicated())                    # True marks the repeated row
df = df.drop_duplicates()
print("rows left:", len(df))
```

```python
# All the steps together
df = load_dirty()
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
df["Date"] = pd.to_datetime(df["Date"], format="%Y/%m/%d", errors="coerce")
df = df.dropna(subset=["Date"])
df.loc[df["Duration"] > 120, "Duration"] = 120
df = df.drop_duplicates()
print(df)
print()
print("missing values:", int(df.isnull().sum().sum()), "| duplicates:", int(df.duplicated().sum()))
```
