```python
#Area of a Triangle Using Three Sides
import math

a = 5
b = 6
c = 7

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Area =", area)
```

```python
#Solving a Quadratic Equation
import cmath
a=1
b=8
c=9
d=b**2-4*a*c
root1 = (-b + cmath.sqrt(d)) / (2*a)
root2= (-b-cmath.sqrt(d)) /(2*a)
print("Root 1 =", root1)
print("Root 2 =", root2)
```
