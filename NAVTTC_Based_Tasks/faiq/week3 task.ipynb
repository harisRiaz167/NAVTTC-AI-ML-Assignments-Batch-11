```python
#NumPy Data Type
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([1.5, 2.5, 3.5])
arr3 = np.array([True, False, True])
arr4 = np.array([1, 2, 3], dtype="float64")

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
print(arr4)
print(arr4.dtype)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40])

copy_arr = arr.copy()
view_arr = arr.view()

arr[0] = 100

print("Original:", arr)
print("Copy:", copy_arr)
print("View:", view_arr)
```

```python
#NumPy Array Shape
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
arr3 = np.array([[[1, 2], [3, 4]]])

print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

arr2 = arr.reshape(2, 3)
arr3 = arr.reshape(3, 2)
arr4 = arr.reshape(2, 3, 1)

print(arr2)
print(arr3)
print(arr4)
```

```python
import numpy as np

arr = np.array([10, 20, 30])

for x in arr:
    print(x)

arr2 = np.array([[1, 2, 3], [4, 5, 6]])

for row in arr2:
    print(row)

for x in np.nditer(arr2):
    print(x)
```

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))
print(result)

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print(np.concatenate((a, b), axis=0))
print(np.concatenate((a, b), axis=1))
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 8, 6])

result = np.array_split(arr, 3)

for part in result:
    print(part)
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 20])

print("Index of 20:", np.where(arr == 20))
print("Values greater than 25:", np.where(arr > 25))
```

```python
import numpy as np

arr = np.array([40, 10, 30, 20, 50])

print("Ascending:", np.sort(arr))
print("Descending:", np.sort(arr)[::-1])

arr2 = np.array([[3, 2, 1], [6, 5, 4]])

print(np.sort(arr2))
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Usman"],
    "Age": [20, 22, 21, 23],
    "Marks": [85, 90, 88, 78]
}

df = pd.DataFrame(data)

print(df)
print(df.head())
print(df.info())
print(df.describe())
print(df["Name"])
print(df[df["Marks"] > 80])
```

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 22],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)

print(df)
print("First row:")
print(df.iloc[0])

print("Marks column:")
print(df["Marks"])

print("Students with marks above 80:")
print(df[df["Marks"] > 80])
```
