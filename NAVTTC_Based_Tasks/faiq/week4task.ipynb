```python
#Pandas Data Analysis
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Age": [20, 22, 21, 23, 20],
    "Marks": [85, 92, 67, 78, 88],
    "Attendance": [90, 95, 80, 85, 92]
}

df = pd.DataFrame(data)

print("Complete Dataset:")
print(df)

print("\nFirst 3 Rows:")
print(df.head(3))

print("\nLast 2 Rows:")
print(df.tail(2))

print("\nDataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nStatistical Summary:")
print(df.describe())

print("\nStudents with Marks Above 80:")
print(df[df["Marks"] > 80])

print("\nAverage Marks:")
print(df["Marks"].mean())
```

```python
#Pandas Data Cleaning
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Ali"],
    "Age": [20, 22, None, 23, 20],
    "Marks": [85, 92, 67, None, 85],
    "Date": ["2026-10-01", "2026-10-02", "2026-10-03",
             "wrong_date", "2026-10-01"]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

df = df.drop_duplicates()

print("\nCleaned Data:")
print(df)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

```python
#Pandas Data Correlation
import pandas as pd

data = {
    "StudyHours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80],
    "SleepHours": [8, 7, 8, 6, 7]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nCorrelation Matrix:")
print(df.corr())

print("\nStudy Hours and Marks Correlation:")
print(df["StudyHours"].corr(df["Marks"]))
```

```python
#Calculate Mean, Median, and Mode
import numpy as np
from scipy import stats

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

mean = np.mean(speed)
median = np.median(speed)
mode = stats.mode(speed, keepdims=False).mode

print("Mean =", mean)
print("Median =", median)
print("Mode =", mode)
```
