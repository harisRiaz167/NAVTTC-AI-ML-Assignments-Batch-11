Add Two Numbers
Question: Write a Python program to add two numbers (num1 = 1.5 and num2 = 6.3) and display the sum.
Explanation: This program stores two decimal numbers in variables, adds them using the + operator, and displays the result using print().

```python

# Add Two Numbers

num1 = 1.5
num2 = 6.3

result = num1 + num2

print("First number:", num1)
print("Second number:", num2)
print("Sum:", result)
```

Calculate the Square Root of a Number
Question: Write a Python program to calculate the square root of a number using num = 8.
Explanation: The math module provides mathematical functions. The math.sqrt() function calculates the square root of a non-negative real number.

```python

# Square Root of a Number

import math

num = 8

result = math.sqrt(num)

print("Number:", num)
print("Square root:", result)
```

Question: Find the square root of a real or complex number using Python's cmath module. Use num = 1 + 2j.
Explanation: A complex number contains a real part and an imaginary part. Python uses j to represent the imaginary unit. The cmath.sqrt() function can calculate square roots of complex numbers.

Convert Celsius to Fahrenheit
Question: Write a Python program to convert a temperature from Celsius to Fahrenheit using 37.5°C.
Explanation: The program uses the Celsius-to-Fahrenheit conversion formula.
Formula:
[
F = left(C \times \frac{9}{5}\right) + 32
]

Here, (C) represents Celsius and (F) represents Fahrenheit.

```python

#Celsius to Fahrenheit

celsius = 37.5

fahrenheit = (celsius * 9 / 5) + 32

print("Temperature in Celsius:", celsius)
print("Temperature in Fahrenheit:", fahrenheit)
```

Calculate the Area of a Triangle
Question: Write a Python program to calculate the area of a triangle using its three sides: a = 5, b = 6, and c = 7.
Explanation: When all three sides of a triangle are known, Heron's formula can be used to calculate its area.
First, calculate the semi-perimeter:
[
s = frac{a+b+c}{2}
]
Then calculate the area:
[
A = sqrt{s(s-a)(s-b)(s-c)}
]

```python

# Area of a Triangle Using Three Sides

import math

a = 5
b = 6
c = 7

s = (a + b + c) / 2

area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Side a:", a)
print("Side b:", b)
print("Side c:", c)
print("Semi-perimeter:", s)
print("Area of the triangle:", area)
```

Solve a Quadratic Equation
Question: Write a Python program to solve the quadratic equation \(ax^2 + bx + c = 0\) using the cmath module. Use a = 1, b = 5, and c = 6.
Explanation: A quadratic equation has the general form:
[
ax^2 + bx + c = 0
]

Its roots are calculated using the quadratic formula:
[
x = frac{-b \pm \sqrt{b^2-4ac}}{2a}
]

The expression (b^2-4ac) is called the discriminant. The cmath module allows the program to handle real and complex roots.

```python

# Solve a Quadratic Equation

import cmath

a = 1
b = 5
c = 6

discriminant = b**2 - 4*a*c

root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

print("First root:", root1)
print("Second root:", root2)
```

**Convert Kilometers to Miles**
Question: Write a Python program that takes a distance in kilometers from the user and converts it to miles using the conversion factor 0.621371.
Explanation: The program accepts a distance from the user, converts the input to a floating-point number, and multiplies it by the conversion factor.
Formula:
[
text{Miles} = text{Kilometers} \times 0.621371
]

```python

# Convert Kilometers to Miles

kilometers = float(input("Enter distance in kilometers: "))

miles = kilometers * 0.621371

print("Distance in kilometers:", kilometers)
print("Distance in miles:", miles)
```
