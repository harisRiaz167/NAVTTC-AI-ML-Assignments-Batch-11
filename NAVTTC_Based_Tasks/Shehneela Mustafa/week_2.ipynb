**Topic:** Python — Exception Handling

**Objective:**

To understand exception handling in Python and learn how to handle runtime errors using `try`, `except`, `else`, and `finally` blocks.

**Description:**

Exception handling is a technique used to handle errors that occur while a Python program is running. An exception may occur when a program tries to divide a number by zero, convert invalid text into an integer, or access a file that does not exist.

Python provides the following keywords for exception handling:

- **try:** Contains the code that may produce an exception.
- **except:** Handles an exception if one occurs.
- **else:** Executes when no exception occurs in the `try` block.
- **finally:** Executes whether an exception occurs or not.

Exception handling helps prevent a program from terminating unexpectedly and allows us to display meaningful error messages to the user.



```python
#  Exception Handling

num1 = 10
num2 = 2

try:
    result = num1 / num2

except ZeroDivisionError:
    print("Error: A number cannot be divided by zero.")

else:
    print("First number:", num1)
    print("Second number:", num2)
    print("Division result:", result)

finally:
    print("Program execution completed.")
```

# Week 2

**Topic:** Python — User-Defined Exception Handling

**Objective:**

To learn how to create and raise custom exceptions in Python to handle specific errors according to program requirements.

**Description:**

Python allows programmers to create their own exceptions, known as user-defined exceptions. These exceptions are useful when built-in exceptions do not adequately describe a particular error in a program.

A custom exception is generally created by defining a class that inherits from Python's built-in `Exception` class. The `raise` keyword is used to trigger the custom exception when a specific condition is met.

For example, a program that checks a person's age can raise a custom exception if the entered age is below 18. The exception can then be handled using a `try` and `except` block.

User-defined exceptions make programs easier to understand, maintain, and debug.



```python
#  User-Defined Exception Handling

class InvalidAgeError(Exception):
    pass


try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise InvalidAgeError("Age must be 18 or above.")

    print("You are eligible.")

except InvalidAgeError as error:
    print("Custom Exception:", error)

except ValueError:
    print("Error: Please enter a valid whole number.")

finally:
    print("Program execution completed.")
```
