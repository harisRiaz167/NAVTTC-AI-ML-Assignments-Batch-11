<h2>NumPy Array Slicing</h2>
<p><b>Objective:</b> Extract a range of elements from a NumPy array.</p>
<p><b>Description:</b> Slicing selects part of an array using start, stop, and step values. The start index is included, while the stop index is excluded.</p>
</div>

````python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

print("Elements from index 1 to 3:", arr[1:4])
print("First three elements:", arr[:3])
print("Every second element:", arr[::2])
````

**Expected output**

```text

Every second element: [10 30 50]
```

<p><b>Objective:</b> Understand and identify NumPy array data types.</p>

<h1>— Week 3</h1>
<p><b>Topic:</b> NumPy</p>
<h2>NumPy Copy and View</h2>
<p><b>Objective:</b> Understand the difference between copying and viewing an array.</p>
<p><b>Description:</b> A copy owns separate data, so changing the original does not change the copy. A view shares the underlying data, so changes to shared elements can appear in both.</p>
</div>

````python
import numpy as np

original = np.array([10, 20, 30, 40])
copied = original.copy()
viewed = original.view()

original[0] = 99

print("Original array:", original)
print("Copied array:", copied)
print("Viewed array:", viewed)
````

**Expected output**

```text
Original array: [99 20 30 40]
Copied array: [10 20 30 40]
Viewed array: [99 20 30 40]
```

<h2>NumPy Array Reshaping</h2>
<p><b>Objective:</b> Change the dimensions of an array.</p>
<p><b>Description:</b> The reshape() function changes an array's shape without changing its elements. The requested dimensions must accommodate the total number of elements.</p>
</div>

````python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
reshaped = arr.reshape(2, 3)

print("Original array:", arr)
print("Reshaped array:")
print(reshaped)
````

<h2>Joining NumPy Arrays</h2>
<p><b>Objective:</b> Combine two or more NumPy arrays.</p>
<p><b>Description:</b> The concatenate() function joins arrays along an existing axis. For one-dimensional arrays, it combines their elements into one array.</p>

<h2>Splitting NumPy Arrays</h2>
<p><b>Objective:</b> Divide a NumPy array into smaller arrays.</p>
<p><b>Description:</b> The split() function divides an array into a specified number of equal sub-arrays. The array length must be evenly divisible by the number of sections.</p>

<h2>Pandas Basics Tutorial</h2>
<p><b>Objective:</b> Learn the basic operations of the Pandas library.</p>
<p><b>Description:</b> Pandas is a Python library used for data analysis and manipulation. It provides Series and DataFrame structures and supports filtering, selecting columns, calculating statistics, and handling missing values.</p>

````python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [22, 25, 21, 24],
    "Marks": [85, 92, 78, 88]
}

df = pd.DataFrame(data)

print("Complete DataFrame:")
print(df)
print("\nFirst two rows:")
print(df.head(2))
print("\nAverage marks:", df["Marks"].mean())
````

<h1>— Week 3</h1>
<p><b>Topic:</b> Pandas</p>
<h2>Reading CSV Files with Pandas</h2>
<p><b>Objective:</b> Read and analyze a CSV file using Pandas.</p>
<p><b>Description:</b> CSV means Comma-Separated Values. Pandas read_csv() loads CSV data into a DataFrame, which can then be inspected and analyzed. This example creates a small CSV file before reading it.</p>
</div>

````python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [22, 25, 21],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)

new_df = pd.read_csv("students.csv")

print("Data read from CSV:")
print(new_df)
print("\nColumn names:")
print(new_df.columns.tolist())
print("\nSummary statistics:")
print(new_df["Marks"].describe())
````

**Note:** This code creates students.csv in the notebook's current working directory. Summary statistics are calculated from the Marks column.
