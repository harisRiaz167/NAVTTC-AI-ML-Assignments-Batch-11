### Setup — create `data.csv`
We generate a 169-row workout dataset (same shape as the W3Schools file) and save it as `data.csv`.

```python
rng = np.random.default_rng(42)
n = 169
duration = rng.choice([30, 45, 60, 60, 60, 60, 75, 90, 120, 180], size=n)
pulse    = rng.integers(90, 135, size=n)
maxpulse = pulse + rng.integers(10, 50, size=n)
calories = np.round(duration * rng.uniform(4.2, 6.8, size=n) + rng.normal(0, 20, size=n), 1)

raw = pd.DataFrame({"Duration": duration, "Pulse": pulse,
                    "Maxpulse": maxpulse, "Calories": calories})
raw.loc[[17, 52, 118], "Calories"] = np.nan      # a few empty cells
raw.to_csv("data.csv", index=False)
print("data.csv created with", len(raw), "rows")
```

### Task 1 — Inspect the problems first

```python
print("Empty cells per column:")
print(dirty.isnull().sum())
print()
print("Duplicated rows:", dirty.duplicated().sum())
print("Duplicate row index:", list(dirty[dirty.duplicated()].index))
```

## Part A — Cleaning Empty Cells
Empty cells can give wrong results during analysis. Options: remove the rows, or replace the empty values.

### Task 2 — Remove rows with empty cells: `dropna()`
`dropna()` returns a **new** DataFrame and leaves the original untouched.

```python
new_df = dirty.dropna()
print("Before:", dirty.shape, " After:", new_df.shape)
print("Rows removed:", len(dirty) - len(new_df))
new_df.isnull().sum()
```

### Task 3 — Replace empty values with a fixed value: `fillna()`
Here we fill every empty cell with `130`. Notice it also fills the empty `Date` in row 22 with `130` — that is why replacing in one column only (next task) is usually better.

```python
filled = dirty.fillna(130)
print(filled.isnull().sum().sum(), "empty cells left")
filled.loc[[18, 22, 28]]
```

### Task 4 — Replace empty values in one column only
Specify the column so other columns are unaffected.

```python
filled_col = dirty.fillna({"Calories": 130})
print("Row 18 and 28 Calories now:", filled_col.loc[[18, 28], "Calories"].tolist())
print("Date column still has", filled_col["Date"].isnull().sum(), "empty cell")
```

### Task 5 — Replace using Mean, Median or Mode
The common way to replace empty cells is with the **mean** (average), **median** (middle) or **mode** (most frequent) of the column.

```python
x_mean   = dirty["Calories"].mean()
x_median = dirty["Calories"].median()
x_mode   = dirty["Calories"].mode()[0]
print("Mean  :", round(x_mean, 2))
print("Median:", x_median)
print("Mode  :", x_mode)
```

```python
clean = dirty.copy()
clean = clean.fillna({"Calories": round(x_mean, 1)})     # fill with the mean
clean.loc[[18, 28], ["Calories"]]
```

## Part B — Cleaning Data of Wrong Format
Cells with data in the wrong format (e.g. mixed date formats) make analysis hard. Fix by converting all cells to the same format.

### Task 6 — Convert the `Date` column with `to_datetime()`
Row 22 has no date and row 26 is written as `20201226` instead of `2020/12/26`.

```python
print("Row 26 raw value:", repr(dirty.loc[26, "Date"]))
clean["Date"] = pd.to_datetime(clean["Date"], format="mixed")
print(clean["Date"].dtype)
clean.loc[[25, 26, 27], ["Date"]]
```

### Task 7 — Remove rows with an empty `Date`
After conversion, the empty date becomes `NaT`. We drop it using `subset`.

```python
print("Empty dates before:", clean["Date"].isnull().sum())
clean = clean.dropna(subset=["Date"])
print("Empty dates after :", clean["Date"].isnull().sum())
print("Shape:", clean.shape)
```

## Part C — Fixing Wrong Data
"Wrong data" is not always empty or badly formatted — it can simply be unrealistic. Row 7 says a workout lasted **450** minutes when it should be 45.

### Task 8 — Replace the wrong value

```python
print("Before:", clean.loc[7, "Duration"])
clean.loc[7, "Duration"] = 45
print("After :", clean.loc[7, "Duration"])
```

### Task 9 — Fix many wrong values with a loop
For larger data, set a boundary (e.g. Duration must be ≤ 120) and fix or drop everything outside it. Here we first check, then demonstrate on a copy using `drop()`.

```python
outliers = clean[clean["Duration"] > 120]
print("Rows with Duration > 120 remaining:", len(outliers))

# demonstration on a temporary copy with an injected bad value
demo = clean.copy()
demo.loc[3, "Duration"] = 999
for x in demo.index:
    if demo.loc[x, "Duration"] > 120:
        demo = demo.drop(x)          # or: demo.loc[x, "Duration"] = 120
print("Demo: dropped rows with Duration > 120 ->", demo.shape)
```

## Part D — Removing Duplicates
Duplicate rows are rows that have been registered more than once (rows 11 and 12 are identical).

### Task 10 — Find duplicates: `duplicated()`
Returns `True` for every row that is a duplicate of an earlier row.

```python
dups = clean.duplicated()
print("Number of duplicates:", dups.sum())
clean[dups]
```

### Task 11 — Remove duplicates: `drop_duplicates()`

```python
before = len(clean)
clean = clean.drop_duplicates()
print("Rows before:", before, "| after:", len(clean))
print("Duplicates left:", clean.duplicated().sum())
```

### Task 12 — Final check of the cleaned data

```python
clean = clean.reset_index(drop=True)
print("Empty cells:\n", clean.isnull().sum(), sep="")
print()
clean.info()
```

```python
clean.to_csv("clean_data.csv", index=False)
clean.head(10)
```

**Summary (Assignment 47):** `dropna()` / `fillna()` handled empty cells, `to_datetime()` fixed the date format, direct assignment fixed the unrealistic value, and `drop_duplicates()` removed repeated rows. The cleaned data is saved as `clean_data.csv` for the correlation assignment.

## Topic: Pandas – Data Correlation


### Task 1 — Load the cleaned data

```python
cdf = pd.read_csv("clean_data.csv")
print(cdf.shape)
cdf.head()
```

### Task 2 — Correlation matrix: `corr()`

```python
num = cdf[["Duration", "Pulse", "Maxpulse", "Calories"]]
num.corr().round(3)
```

### Task 3 — Interpret the result
- The diagonal is always **1.000** — every column perfectly correlates with itself.
- The matrix is symmetric (Duration↔Calories equals Calories↔Duration).

### Task 4 — Correlation of two specific columns
`Duration` vs `Calories` — we expect longer workouts to burn more calories; the coefficient tells us how strong that link is in this 30-row sample.

```python
r = num["Duration"].corr(num["Calories"])
print("Duration vs Calories:", round(r, 3))
```

### Task 5 — Correlation of every column with `Calories`, sorted

```python
num.corr()["Calories"].drop("Calories").sort_values(ascending=False).round(3)
```

### Task 6 — Classify each relationship with a helper function

```python
def strength(r):
    a = abs(r)
    if a >= 0.6:  level = "Good (strong)"
    elif a >= 0.3: level = "Moderate"
    else:          level = "Weak"
    return level + (" positive" if r > 0 else " negative" if r < 0 else "")

pairs = [("Duration", "Calories"), ("Duration", "Maxpulse"),
         ("Duration", "Pulse"), ("Pulse", "Maxpulse"), ("Pulse", "Calories")]
for a, b in pairs:
    r = num[a].corr(num[b])
    print(f"{a:9s} vs {b:9s}: r = {r:6.3f}  -> {strength(r)}")
```

### Task 7 — Alternative correlation methods
`corr()` uses Pearson by default; `spearman` and `kendall` are rank-based alternatives.

```python
print("Pearson :", round(num["Duration"].corr(num["Calories"], method="pearson"), 3))
print("Spearman:", round(num["Duration"].corr(num["Calories"], method="spearman"), 3))
print("Kendall :", round(num["Duration"].corr(num["Calories"], method="kendall"), 3))
```
