```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [22, 25, 23, 24, 26],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Peshawar"],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nFirst 3 Rows:")
print(df.head(3))

print("\nLast 2 Rows:")
print(df.tail(2))

print("\nDataFrame Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nColumn Names:")
print(df.columns)

print("\nName Column:")
print(df["Name"])

print("\nSelected Columns:")
print(df[["Name", "Marks"]])

print("\nStudents with Marks Greater than 85:")
print(df[df["Marks"] > 85])

print("\nStudent at Index 0:")
print(df.iloc[0])

print("\nStudent from Index 1 to 3:")
print(df.iloc[1:4])

df["Grade"] = ["B", "A", "C", "A", "B"]
print("\nAfter Adding Grade Column:")
print(df)

df = df.sort_values(by="Marks", ascending=False)
print("\nSorted by Marks:")
print(df)

print("\nAverage Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())

print("\nStudents Grouped by City:")
print(df.groupby("City")["Marks"].mean())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
data = [10, 20, 30, 40, 50]
series = pd.Series(data)

print("Pandas Series:")
print(series)

print("\nSeries with Custom Index:")
series2 = pd.Series(data, index=["a", "b", "c", "d", "e"])
print(series2)

print("\nAccessing Elements:")
print("First Element:", series[0])
print("Third Element:", series[2])
print("Element at Index 'b':", series2["b"])

print("\nSeries from Dictionary:")
marks = {"Ali": 85, "Ahmed": 90, "Sara": 78, "Ayesha": 95}
marks_series = pd.Series(marks)
print(marks_series)

print("\nSeries Information:")
print("Values:", marks_series.values)
print("Indexes:", marks_series.index)
print("Data Type:", marks_series.dtype)
print("Size:", marks_series.size)

print("\nSeries Operations:")
print("Sum:", marks_series.sum())
print("Mean:", marks_series.mean())
print("Maximum:", marks_series.max())
print("Minimum:", marks_series.min())

print("\nFiltering Series:")
print(marks_series[marks_series > 85])

print("\nAdding 5 to All Marks:")
print(marks_series + 5)
```

```python
data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [22, 25, 23, 24, 26],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Peshawar"],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)

print("DataFrame:")
print(df)

print("\nFirst 3 Rows:")
print(df.head(3))

print("\nDataFrame Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nName Column:")
print(df["Name"])

print("\nSelected Columns:")
print(df[["Name", "Marks"]])

print("\nFirst Row:")
print(df.iloc[0])

print("\nStudents with Marks Greater than 85:")
print(df[df["Marks"] > 85])

df["Grade"] = ["B", "A", "C", "A", "B"]
print("\nAfter Adding Grade Column:")
print(df)

df.loc[5] = ["Hassan", 21, "Multan", 80, "B"]
print("\nAfter Adding a New Row:")
print(df)

df = df.drop(columns=["Grade"])
print("\nAfter Removing Grade Column:")
print(df)

print("\nStatistical Summary:")
print(df.describe())

print("\nAverage Marks:", df["Marks"].mean())

print("\nSorted by Marks:")
print(df.sort_values(by="Marks", ascending=False))

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
    "Age": [22, 25, 23, 24, 26],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Peshawar"],
    "Marks": [85, 90, 78, 92, 88]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)

df = pd.read_csv("students.csv")

print("CSV File Data:")
print(df)

print("\nFirst 3 Rows:")
print(df.head(3))

print("\nLast 2 Rows:")
print(df.tail(2))

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataFrame Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nStudents with Marks Greater than 85:")
print(df[df["Marks"] > 85])

print("\nName and Marks Columns:")
print(df[["Name", "Marks"]])
```
