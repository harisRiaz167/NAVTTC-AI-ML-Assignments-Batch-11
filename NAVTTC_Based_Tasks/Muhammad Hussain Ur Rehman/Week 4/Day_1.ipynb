```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman", "Hassan"],
    "Age": [22, 25, 23, 24, 26, 21],
    "City": ["Lahore", "Karachi", "Islamabad", "Lahore", "Peshawar", "Karachi"],
    "Marks": [85, 90, 78, 92, 88, 75],
    "Salary": [30000, 45000, 35000, 50000, 40000, 28000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nAverage Marks:", df["Marks"].mean())
print("Maximum Marks:", df["Marks"].max())
print("Minimum Marks:", df["Marks"].min())
print("Median Marks:", df["Marks"].median())

print("\nStudents with Marks Greater than 85:")
print(df[df["Marks"] > 85])

print("\nStudents Sorted by Marks:")
print(df.sort_values(by="Marks", ascending=False))

print("\nAverage Marks by City:")
print(df.groupby("City")["Marks"].mean())

print("\nNumber of Students in Each City:")
print(df["City"].value_counts())

df["Result"] = df["Marks"].apply(lambda x: "Pass" if x >= 80 else "Fail")
print("\nStudents with Results:")
print(df)

print("\nAverage Salary:", df["Salary"].mean())

print("\nCorrelation Between Age, Marks, and Salary:")
print(df[["Age", "Marks", "Salary"]].corr())
```

```python
data = {
    "Name": ["Ali", "Ahmed", "Sara", "Ayesha", "Usman", "Hassan"],
    "Age": [22, 25, 23, 24, 26, 21],
    "Study_Hours": [2, 4, 3, 5, 4, 1],
    "Marks": [60, 75, 70, 90, 85, 50],
    "Salary": [30000, 45000, 35000, 50000, 40000, 28000]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nCorrelation Between Numerical Columns:")
print(df.corr(numeric_only=True))

print("\nCorrelation Between Study Hours and Marks:")
print(df["Study_Hours"].corr(df["Marks"]))

print("\nCorrelation Between Age and Salary:")
print(df["Age"].corr(df["Salary"]))

print("\nPositive Correlations:")
correlation_matrix = df.corr(numeric_only=True)
print(correlation_matrix[correlation_matrix > 0])

print("\nNegative Correlations:")
print(correlation_matrix[correlation_matrix < 0])
```
