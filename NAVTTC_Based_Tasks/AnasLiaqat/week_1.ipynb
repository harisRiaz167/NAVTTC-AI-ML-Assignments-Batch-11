```python
#Find the square root of a real or complex number using Python's cmath module (the example uses 1+2j).
import cmath

num = 1 + 2j
sqrt = cmath.sqrt(num)
print(f"The square root of {num} is {sqrt.real:.3f} + {sqrt.imag:.3f}j")
```

```python
#Write a Python program to convert a temperature from Celsius to Fahrenheit (the example uses 37.5°C).
celsius = 37.5
fahrenheit = (celsius * 9 / 5) + 32
print(f"{celsius} degrees Celsius is equal to {fahrenheit} degrees Fahrenheit")
```

```python
#Write a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).
a, b, c = 5, 6, 7
s = (a + b + c) / 2  # semi-perimeter
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print(f"The area of the triangle with sides {a}, {b}, {c} is {area:.2f}")
```

```python
#Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).
import cmath

a, b, c = 1, 5, 6
d = (b ** 2) - (4 * a * c)  # discriminant
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)
print(f"The solutions are {sol1} and {sol2}")
```

```python
#Complete the exercise on the list data type in Python.
my_list = [10, 20, 30, 40, 50]

print(my_list)               # whole list
print(my_list[0])            # indexing -> 10
print(my_list[-1])           # negative index -> 50
print(my_list[1:4])          # slicing -> [20, 30, 40]

my_list.append(60)           # add at end
my_list.insert(1, 15)        # insert at index 1
my_list.extend([70, 80])     # add multiple items
print(my_list)

my_list.remove(15)           # remove by value
my_list.pop()                # remove last item
print(my_list)

my_list[0] = 5               # update an item
print(my_list)

print(len(my_list))          # length
my_list.sort(reverse=True)   # sort descending
print(my_list)
my_list.reverse()
print(my_list)
print(30 in my_list)         # membership check
```

```python
#set data type
# Creating sets
s = {1, 2, 3, 4, 4, 2}          # duplicates are removed
print(s)                         # {1, 2, 3, 4}
empty = set()                    # {} would create a dictionary
print(type(empty))

# Sets are unordered, so they can't be indexed
print(3 in s)                    # membership check
print(len(s))

# Adding and removing
s.add(5)
s.update([6, 7, 8])
s.remove(8)                      # raises KeyError if the item is missing
s.discard(100)                   # no error if the item is missing
print(s.pop())                   # removes an arbitrary item
print(s)

# Set operations
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A | B)                     # union
print(A & B)                     # intersection
print(A - B)                     # difference
print(A ^ B)                     # symmetric difference
print(A.issubset(B), A.isdisjoint({10, 11}))

# Iterating and clearing
for x in A:
    print(x, end=" ")
print()
A.clear()
print(A)                         # set()
```

```python
#Complete the exercise on the dictionary data type in Python.
student = {"name": "Ali", "age": 21, "course": "AI"}
print(student)
print(student["name"])
print(student.get("grade", "N/A"))      # safe access with a default

# Add and update
student["age"] = 22
student["city"] = "Lahore"
student.update({"grade": "A"})

# Remove
student.pop("city")
del student["grade"]
print(student)

# Keys, values, items
print(student.keys())
print(student.values())
print(student.items())

# Iterate
for k, v in student.items():
    print(k, "->", v)

# Membership and length
print("name" in student, len(student))

# Dictionary comprehension
squares = {x: x ** 2 for x in range(1, 6)}
print(squares)

# Nested dictionary
people = {1: {"name": "Ali", "age": 21}, 2: {"name": "Sara", "age": 23}}
print(people[2]["name"])

# Copy and clear
copy = student.copy()
student.clear()
print(student, copy)
```
