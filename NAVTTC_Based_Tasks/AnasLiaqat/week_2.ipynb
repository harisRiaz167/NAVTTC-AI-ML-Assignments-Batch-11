```python
#Complete the exercise on user-defined functions in Python.
# 1. Simple function
def greet():
    print("Hello, welcome to Python!")

greet()
def add(a, b):
    return a + b

print("Sum =", add(5, 3))

def power(base, exp=2):
    return base ** exp

print(power(4))      # 16
print(power(2, 5))   # 32

# 4. Even or odd
def even_odd(n):
    return "Even" if n % 2 == 0 else "Odd"

print(even_odd(7))

def total(*nums):
    return sum(nums)

print(total(1, 2, 3, 4))
```

```python
#Complete the exercise on exception handling in Python.
# Basic try-except
try:
    num = int(input("Enter a number: "))
    print(10 / num)
except ZeroDivisionError:
    print("You can't divide by zero!")
except ValueError:
    print("Please enter a valid number.")
```

```python
#Complete the exercise on user-defined exception handling in Python.
class InvalidAgeError(Exception):
    def __init__(self, age, message="Age must be between 0 and 120"):
        self.age = age
        self.message = message
        super().__init__(f"{message}. Got: {age}")

def check_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError(age)
    print("Valid age:", age)

try:
    check_age(25)
    check_age(150)
except InvalidAgeError as e:
    print("Error:", e)
# Exception hierarchy
class BankError(Exception):
    pass

class InsufficientFundsError(BankError):
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"Balance {balance}, requested {amount}")
    return balance - amount

try:
    withdraw(100, 500)
except BankError as e:                     # catches the subclass too
    print("Bank error:", e)
```

```python
#Write a Python program to calculate the mean, median, and mode of a dataset.
import statistics
numbers = [10, 20, 20, 30, 40, 50, 70]

mean_value = statistics.mean(numbers)
median_value = statistics.median(numbers)
mode_value = statistics.mode(numbers)

print("Numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

```python
#Write a Python program to calculate the standard deviation of a dataset.

import statistics
numbers = [10, 20, 20, 30, 40, 50, 70]

std_deviation = statistics.pstdev(numbers)

print("Numbers:", numbers)
print("Standard Deviation:", std_deviation)
```
