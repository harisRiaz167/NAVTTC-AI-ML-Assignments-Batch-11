Implement linear regression in Python with scikit-learn.
Reference: https://stackabuse.com/linear-regression-in-python-with-scikit-learn/<br>
Objective: To implement linear regression using Python and Scikit-learn to predict a numerical value from input data.


```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([50, 55, 65, 70, 80])

model = LinearRegression()
model.fit(X, y)

prediction = model.predict([[6]])

print("Slope:", round(model.coef_[0], 2))
print("Intercept:", round(model.intercept_, 2))
print("Predicted marks for 6 hours:", round(prediction[0], 2))

y_pred = model.predict(X)
print("Mean Squared Error:", round(mean_squared_error(y, y_pred), 2))
print("R-squared:", round(r2_score(y, y_pred), 2))
```

Perform logistic regression.
Reference: https://towardsdatascience.com/logistic-regression-using-python-sklearn-numpy-mnist-handwriting-recognition-matplotlib-a6b31e2b166a
https://www.datacamp.com/community/tutorials/understanding-logistic-regression-python<br>
Objective: To implement logistic regression using Scikit-learn to classify data into two categories.


```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X = np.array([1, 2, 3, 4, 5, 6, 7, 8]).reshape(-1, 1)
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

model = LogisticRegression()

model.fit(X, y)

new_students = np.array([2, 7]).reshape(-1, 1)
predictions = model.predict(new_students)

print("Predictions:", predictions)
print("Accuracy:", round(accuracy_score(y, model.predict(X)), 2))

probabilities = model.predict_proba(new_students)
print("Probabilities (Fail, Pass):")
print(np.round(probabilities, 2))
```
