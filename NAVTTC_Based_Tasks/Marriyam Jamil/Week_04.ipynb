Complete the exercise on data analysis with Pandas.
Reference: https://www.w3schools.com/python/pandas_analyzing.asp<br>
Objective: To analyze data using Pandas functions such as head(), tail(), info(), describe(), and mean().


```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman"],
    "Age": [22, 21, 23, 20, 24],
    "Marks": [85, 90, 78, 95, 82]
}

df = pd.DataFrame(data)
print("First rows:")
print(df.head())

print("\nLast three rows:")
print(df.tail(3))

print("\nStatistical summary:")
print(df.describe())

print("\nAverage marks:", df["Marks"].mean())

print("Highest marks:", df["Marks"].max())

print("Lowest marks:", df["Marks"].min())
```

#### Assignment 47 — Week 4, Day 1
##### Topic: Pandas
Complete the exercise on Pandas data-cleaning techniques.
Reference: https://www.w3schools.com/python/pandas_cleaning.asp<br>
Objective: To identify and handle missing values, remove duplicate rows, and correct data in a DataFrame.


```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Sara", "Ayesha"],
    "Age": [22, 21, None, 21, 20],
    "Marks": [85, 90, 78, 90, None]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print("\nMissing values in each column:")
print(df.isnull().sum())

df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

df = df.drop_duplicates()

print("\nCleaned DataFrame:")
print(df)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

#### Assignment 48 — Week 4, Day 1
##### Topic: Pandas
Complete the exercise on data correlation with Pandas.
Reference: https://www.w3schools.com/python/pandas_correlations.asp<br>
Objective: To calculate the correlation between numerical columns in a DataFrame and understand how two variables are related.

```python
import pandas as pd
data = {
    "StudyHours": [1, 2, 3, 4, 5],
    "Marks": [50, 60, 65, 80, 90]
}
df = pd.DataFrame(data)
print("Student Data:")
print(df)
print("\nCorrelation Matrix:")
print(df.corr())
correlation = df["StudyHours"].corr(df["Marks"])
print("\nCorrelation between study hours and marks:", round(correlation, 2))
```
