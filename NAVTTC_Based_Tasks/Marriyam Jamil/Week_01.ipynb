#### Assignment 8 — Week 1, Day 3
##### Topic: Python
Write a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).


```python
import math
a = 5
b = 6
c = 7
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print("Semi-perimeter:", s)
print("Area of the triangle:", area)
```

#### Assignment 9 — Week 1, Day 3
##### Topic: Python
Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).


```python
import cmath
a = 1
b = 5
c = 6
d = (b ** 2) - (4 * a * c)
root1 = (-b + cmath.sqrt(d)) / (2 * a)
root2 = (-b - cmath.sqrt(d)) / (2 * a)
print("First root:", root1)
print("Second root:", root2)
```

#### Assignment 19 — Week 1, Day 5
##### Topic: Python
Complete the exercise on various looping techniques in Python.
Reference: https://www.geeksforgeeks.org/looping-techniques-python/<br>
Objective: Practice different ways of iterating over Python collections.

```python
# 1. Loop through a list
numbers = [10, 20, 30]
for n in numbers:
    print(n)

# 2. Loop using enumerate()
fruits = ["Apple", "Mango", "Banana"]
for index, fruit in enumerate(fruits):
    print(index, fruit)

# 3. Loop through two lists using zip()
names = ["Ali", "Sara"]
marks = [85, 90]
for name, mark in zip(names, marks):
    print(name, mark)

# 4. Loop in reverse order
for n in reversed(numbers):
    print(n)
```
