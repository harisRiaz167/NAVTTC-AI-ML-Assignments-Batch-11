#### Assignment 26 — Week 2, Day 1
##### Topic: Python
Complete the exercise on exception handling in Python.
Reference: https://www.programiz.com/python-programming/exception-handling<br>
Objective: Use try and except to handle errors when dividing two numbers.


```python
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))

    result = num1 / num2
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

except ValueError:
    print("Error: Please enter valid integers.")
```

#### Assignment 27 — Week 2, Day 1
##### Topic: Python
Complete the exercise on user-defined exception handling in Python.
Reference: https://www.programiz.com/python-programming/user-defined-exception<br>
Objective: Create a custom exception and raise it when a user's age is below 18.


```python
class InvalidAgeError(Exception):
    pass
try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise InvalidAgeError("Age must be 18 or above.")

    print("You are eligible.")

except InvalidAgeError as e:
    print("Error:", e)

except ValueError:
    print("Error: Please enter a valid integer.")
```

#### Assignment 49 — Week 2, Day 4
##### Topic: Statistics
Calculate mean, median, and mode.
Reference: https://www.w3schools.com/python/python_ml_mean_median_mode.asp<br>
Objective: Calculate the mean, median, and mode of a dataset using Python's statistics module.


```python
import statistics
data = [2, 4, 4, 6, 8, 10, 12]

mean_value = statistics.mean(data)
median_value = statistics.median(data)
mode_value = statistics.mode(data)

print("Data:", data)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```

#### Assignment 50 — Week 2, Day 5
##### Topic: Statistics
Calculate standard deviation.
Reference: https://www.w3schools.com/python/python_ml_standard_deviation.asp<br>
Objective: Calculate the standard deviation of a dataset using Python.

```python

import statistics
data = [2, 4, 4, 4, 5, 5, 7, 9]
std_deviation = statistics.pstdev(data)
print("Data:", data)
print("Standard deviation:", std_deviation)
```
