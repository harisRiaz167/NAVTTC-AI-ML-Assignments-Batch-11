#### Assignment 32 — Week 3, Day 4

Complete the exercise on NumPy array copying and viewing.
Reference: https://www.w3schools.com/python/numpy_copy_vs_view.asp<br>
Objective: Understand the difference between copying an array and creating a view of an array.


```python

import numpy as np

original = np.array([10, 20, 30, 40])
copy_arr = original.copy()
view_arr = original.view()
original[0] = 100

print("Original array:", original)
print("Copy array:", copy_arr)
print("View array:", view_arr)
```

#### Assignment 44 — Week 3, Day 5
##### Topic: Pandas
Complete the exercise on Pandas DataFrames.
Reference: https://www.w3schools.com/python/pandas_dataframes.asp<br>
Objective: To create a DataFrame, access columns and rows, and display selected data.

```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [22, 21, 23, 20],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)

print("Complete DataFrame:")
print(df)
print("\nStudent Names:")
print(df["Name"])
print("\nFirst two rows:")
print(df.head(2))
print("\nFirst row:")
print(df.iloc[0])
print("\nStudents scoring above 80:")
print(df[df["Marks"] > 80])
```

#### Assignment 45 — Week 3, Day 5
##### Topic: Pandas
Complete the exercise on opening/reading CSV files with Pandas.
Reference: https://www.w3schools.com/python/pandas_csv.asp<br>
Objective: To create a CSV file, read it using Pandas, and display its data.

```python
import pandas as pd
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [22, 21, 23, 20],
    "Marks": [85, 90, 78, 95]
}

df = pd.DataFrame(data)
df.to_csv("students.csv", index=False)
students = pd.read_csv("students.csv")
print("Student Data from CSV:")
print(students)
print("\nFirst two rows:")
print(students.head(2))
print("\nColumn names:")
print(students.columns.tolist())
```
