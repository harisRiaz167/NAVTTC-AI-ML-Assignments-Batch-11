## Assignment 47 — Week 4, Day 1
- Complete the exercise on Pandas data-cleaning techniques.
Reference: https://www.w3schools.com/python/pandas_cleaning.asp



```python
import pandas as pd
df = pd.read_csv('test.csv')

x = df["Calories"].median()

df.fillna({"Calories": x}, inplace=True)



df['Date'] = pd.to_datetime(df['Date'], format='mixed')


df.dropna(subset=['Date'], inplace = True)

df.loc[7, 'Duration'] = 45

for x in df.index:
  if df.loc[x, "Duration"] > 120:
    df.drop(x, inplace = True)


print(df.drop_duplicates())
print(df.to_string())
```

## Assignment 49 
- Calculate mean, median, and mode.
Reference: https://www.w3schools.com/python/python_ml_mean_median_mode.asp


```python
import numpy
from scipy import stats

speed = [99,86,87,88,111,86,103,87,94,78,77,85,86]

x = numpy.mean(speed)
y = stats.mode(speed)
z = numpy.median(speed)



print(f"Mean",x)
print(f"Mode",y.mode)
print(f"Median",z)
```
