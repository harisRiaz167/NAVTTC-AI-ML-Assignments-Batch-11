## Assignment 17 — Week 1, Day 5
- Complete the exercise on the break statement in Python.
Reference: https://www.geeksforgeeks.org/python-break-statement/


```python
a = [1, 3, 5, 7, 9, 11]
val = 3

for i in a:
    if i == val:
        print(f"Found at {i}")
        break
else:
    print(f"not found")

for i in range(10):
    print(i)
    if i == 6:
        break


c = 5

while True:
    print(c)
    c -= 1
    if c == 0:
        print("Countdown finished!")
        break  

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
val = 5
found = False

for r in matrix:
    for n in r:
        if n == val:
            print(f"{val} found!")
            found = True
            break  
    if found:
        break 
```

## Assignment 19 — Week 1, Day 5
- Complete the exercise on various looping techniques in Python.
Reference: https://www.geeksforgeeks.org/looping-techniques-python/


```python
words = ['The', 'Big', 'Bang']

for index, value in enumerate(words):
    print(index, value)


names = ['Leo', 'Kendall']
ages = (24, 27, 25)

for name, age in zip(names, ages):
    print(f"Name: {name}, Age: {age}")


p1 = {'Washington': 'First', 'Lincoln': 'Emancipator'}

for name, title in p1.items():
    print(name, title)



num = [1, 3, 5, 6,3,5]

print("Sorted list:")
for n in sorted(num):
    print(n, end=" ")

print("\nSorted list without duplicates:")
for n in sorted(set(num)):
    print(n, end=" ")

print("\n")
print("Original list")
num = [1, 3, 5, 6, 2, 1, 3,4,3,2]
print(num)
print("Reversed list")
for n in reversed(num):
    print(n, end=" ")
```
