## Task 1: Linear Regression

**Topic:** Machine Learning  
**Question:** Implement linear regression in Python using scikit-learn.

**Reference:** [Linear Regression in Python with scikit-learn](https://stackabuse.com/linear-regression-in-python-with-scikit-learn/)
 a small sample dataset to predict a student's score from the number of hours studied.

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Sample data: hours studied and scores
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([25, 35, 50, 60, 75])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Predict the score for 6 hours of study
prediction = model.predict([[6]])

print('Predicted score for 6 hours of study:', round(prediction[0], 2))
print('Model coefficient:', round(model.coef_[0], 2))
print('Model intercept:', round(model.intercept_, 2))
print('R-squared:', round(model.score(X, y), 3))
```

## Task 2: Logistic Regression

**Topic:** Machine Learning  
**Question:** Perform logistic regression in Python using scikit-learn.

**Reference:** [Logistic Regression using Python, scikit-learn, NumPy and MNIST](https://towardsdatascience.com/logistic-regression-using-python-sklearn-numpy-mnist-handwriting-recognition-matplotlib-a6b31e2b166a)

 a small sample dataset to classify whether a student passes based on hours studied.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# Sample data: hours studied
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
# 0 = Fail, 1 = Pass
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

# Create and train the model
model = LogisticRegression()
model.fit(X, y)

# Predict outcomes for new values
new_hours = np.array([[3.5], [6.5]])
predictions = model.predict(new_hours)

for hours, result in zip(new_hours.flatten(), predictions):
    status = 'Pass' if result == 1 else 'Fail'
    print(f'{hours} hours of study: {status}')

# Evaluate the model on the sample training data
train_predictions = model.predict(X)
print('Training accuracy:', round(accuracy_score(y, train_predictions) * 100, 2), '%')
print('Confusion matrix:')
print(confusion_matrix(y, train_predictions))
```

## Conclusion

- Linear Regression predicts a numerical value.
- Logistic Regression predicts a class, such as Pass or Fail.
- These examples use small sample datasets to demonstrate the basic scikit-learn workflow.
