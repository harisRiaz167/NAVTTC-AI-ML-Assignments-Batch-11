**Reference:** [https://www.w3schools.com/python/numpy_array_reshape.asp](https://www.w3schools.com/python/numpy_array_reshape.asp)

## Task 4
**Topic:** NumPy  
**Task statement:** Complete the exercise on NumPy array iteration.  
**Reference:** [https://www.w3schools.com/python/numpy_array_iterating.asp](https://www.w3schools.com/python/numpy_array_iterating.asp)


```python
# Iterate a 1-D array
for x in np.array([1, 2, 3]):
    print(x)

# Iterate a 2-D array (row by row)
a2 = np.array([[1, 2, 3], [4, 5, 6]])
for row in a2:
    print(row)

# Iterate every scalar of a 2-D array
for row in a2:
    for x in row:
        print(x, end=" ")
print()

# nditer: iterate all elements regardless of dimensions
a3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
for x in np.nditer(a3):
    print(x, end=" ")
print()

# nditer with data type change
for x in np.nditer(np.array([1, 2, 3]), flags=['buffered'], op_dtypes=['S']):
    print(x)

# ndenumerate: get the index too
for idx, x in np.ndenumerate(a2):
    print(idx, x)
```

## Task 5
**Topic:** NumPy  
**Task statement:** Complete the exercise on NumPy random techniques.  
**Reference:** [https://www.w3schools.com/python/numpy_random.asp](https://www.w3schools.com/python/numpy_random.asp)



```python
np.random.seed(42)   # reproducible results

print("Random int:", np.random.randint(100))
print("Random float:", np.random.rand())
print("1-D random ints:", np.random.randint(100, size=5))
print("2-D random ints:\n", np.random.randint(100, size=(3, 5)))
print("2-D random floats:\n", np.random.rand(3, 5))

# choice from a list
print("Choice:", np.random.choice([3, 5, 7, 9]))
print("Choice 2-D:\n", np.random.choice([3, 5, 7, 9], size=(3, 5)))

# Data distribution with probabilities (must sum to 1)
print(np.random.choice([3, 5, 7, 9], p=[0.1, 0.3, 0.6, 0.0], size=10))

# shuffle (in place) and permutation (returns copy)
arr = np.array([1, 2, 3, 4, 5])
np.random.shuffle(arr)
print("Shuffled:", arr)
print("Permutation:", np.random.permutation(np.array([1, 2, 3, 4, 5])))
```

## Task 9
**Topic:** NumPy  
**Task statement:** Complete the exercise on NumPy array copying and viewing.  
**Reference:** [https://www.w3schools.com/python/numpy_copy_vs_view.asp](https://www.w3schools.com/python/numpy_copy_vs_view.asp)



```python

arr = np.array([1, 2, 3, 4, 5])
c = arr.copy()
arr[0] = 42
print("original:", arr)
print("copy:    ", c)


arr = np.array([1, 2, 3, 4, 5])
v = arr.view()
arr[0] = 42
print("original:", arr)
print("view:    ", v)

# Changes made through the view affect the original too
v[1] = 99
print("original after view change:", arr)

# Check ownership with .base
print("copy.base:", c.base)
print("view.base is arr:", v.base is arr)
```

##  Task 16
**Topic:** Pandas  
**Task statement:** Complete the exercise on Pandas Series.  
**Reference:** [https://www.w3schools.com/python/pandas_series.asp](https://www.w3schools.com/python/pandas_series.asp)


```python
# Create a Series from a list
a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)

# Access by label (index)
print("First value:", myvar[0])

# Custom labels
myvar = pd.Series(a, index=["x", "y", "z"])
print(myvar)
print("Value at label 'y':", myvar["y"])

# Series from a dictionary (keys become labels)
calories = {"day1": 420, "day2": 380, "day3": 390}
s = pd.Series(calories)
print(s)

# Select only some items from the dictionary
print(pd.Series(calories, index=["day1", "day2"]))
```
