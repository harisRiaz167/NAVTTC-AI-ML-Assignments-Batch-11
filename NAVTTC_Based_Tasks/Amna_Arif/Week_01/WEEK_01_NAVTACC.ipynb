## Q6. Area of a triangle
Write a Python program to find the area of a triangle using its three sides (`a = 5`, `b = 6`, `c = 7`) and the semi-perimeter formula.

```python
a = 5
b = 6
c = 7


s = (a + b + c) / 2

# area using Heron's formula
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
print('The area of the triangle is', area)
```

## Q7. Solve a quadratic equation
Solve the quadratic equation `ax² + bx + c = 0` using the `cmath` module (`a = 1`, `b = 5`, `c = 6`).

```python
import cmath

a = 1
b = 5
c = 6

# discriminant
d = (b ** 2) - (4 * a * c)

# two solutions
sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print('The solutions are', sol1, 'and', sol2)
```

## Q11. Nested if statement
Run and examine the nested-if example using `i = 10` and the nested conditions `i < 10` and `i < 12`.

```python
i = 10

if i == 10:
    if i < 10:
        print('i is smaller than 10')
    if i < 12:
        print('i is smaller than 12 too')
    else:
        print('i is greater than 12')
```

**Observation:** `i == 10` is True, so we enter the outer block. `i < 10` is False (skipped), but `i < 12` is True, so that message prints.

## Q12. if-elif-else statement
Run and examine the if-elif-else example using `i = 20` and the conditions `i == 10`, `i == 15`, and `i == 20`.

```python
i = 20

if i == 10:
    print('i is 10')
elif i == 15:
    print('i is 15')
elif i == 20:
    print('i is 20')
else:
    print('i is not present')
```

**Observation:** Python checks conditions from top to bottom and runs the first one that is True (`i == 20`).

## Q13. for loops
Complete the exercise on `for` loops in Python.  
Reference: https://www.geeksforgeeks.org/python-for-loops/

```python
# Loop over a list
fruits = ['apple', 'banana', 'mango']
for fruit in fruits:
    print(fruit)

# Loop using range()
print('Numbers from 1 to 5:')
for i in range(1, 6):
    print(i)

# Loop over a string
for ch in 'Python':
    print(ch)
```

## Q14. while loops
Complete the exercise on `while` loops in Python.

```python
count = 1

while count <= 5:
    print('Count is', count)
    count += 1

print('Loop finished')
```

## Q15. break statement
Complete the exercise on the `break` statement in Python.  
Reference: https://www.geeksforgeeks.org/python-break-statement/

```python
# Stop the loop when the number 5 is reached
for i in range(1, 10):
    if i == 5:
        print('Found 5, breaking the loop')
        break
    print(i)
```

## Q16. continue statement
Complete the exercise on the `continue` statement in Python.  
Reference: https://www.geeksforgeeks.org/python-continue-statement/

```python
# Skip the number 5 and continue with the rest
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

## Q17. Looping techniques
Complete the exercise on various looping techniques in Python.  
Reference: https://www.geeksforgeeks.org/looping-techniques-python/

```python
# 1. enumerate() - get index and value
colors = ['red', 'green', 'blue']
for index, color in enumerate(colors):
    print(index, color)

# 2. zip() - loop over two lists together
names = ['Ali', 'Sara', 'Ahmed']
marks = [85, 90, 78]
for name, mark in zip(names, marks):
    print(name, 'scored', mark)

# 3. sorted() - loop in sorted order
for num in sorted([5, 2, 9, 1]):
    print(num)

# 4. reversed() - loop in reverse order
for num in reversed([1, 2, 3]):
    print(num)

# 5. items() - loop over a dictionary
student = {'name': 'Amna', 'class': 'C2C'}
for key, value in student.items():
    print(key, ':', value)
```
