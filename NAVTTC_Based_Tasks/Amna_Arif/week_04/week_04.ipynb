## Task 1

**Topic:** NumPy

**Task statement:** Complete the exercise on creating NumPy arrays using Python.

```python
import numpy as np

arr0 = np.array(42)                          # 0-D
arr1 = np.array([1, 2, 3, 4, 5])             # 1-D
arr2 = np.array([[1, 2, 3], [4, 5, 6]])      # 2-D
arr3 = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])  # 3-D

for a in (arr0, arr1, arr2, arr3):
    print(a, "-> ndim =", a.ndim)

# Array with 5 dimensions
arr5 = np.array([1, 2, 3, 4], ndmin=5)
print(arr5, "ndim =", arr5.ndim)
```

## Task 3

**Topic:** NumPy

**Task statement:** Complete the exercise on NumPy array reshaping.

```python
arr = np.arange(1, 13)

print(arr.reshape(4, 3))        # 1-D -> 2-D
print(arr.reshape(2, 3, 2))     # 1-D -> 3-D
print(arr.reshape(2, 2, -1))    # unknown dimension
print(arr.reshape(4, 3).reshape(-1))   # flatten back to 1-D
print(arr.reshape(2, 3, 2).base is not None)   # reshape returns a view
```

## Task 4

**Topic:** NumPy

**Task statement:** Complete the exercise on NumPy array iteration.

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])

print("Row by row:")
for x in arr:
    print(x)

print("Each element:")
for x in arr:
    for y in x:
        print(y, end=" ")
print()

print("nditer:")
for x in np.nditer(arr):
    print(x, end=" ")
print()

print("nditer with dtype change:")
for x in np.nditer(arr, flags=['buffered'], op_dtypes=['S']):
    print(x, end=" ")
print()

print("ndenumerate:")
for idx, x in np.ndenumerate(arr):
    print(idx, x)
```

## Task 9

**Topic:** NumPy

**Task statement:** Complete the exercise on NumPy array copying and viewing.

```python
arr = np.array([1, 2, 3, 4, 5])
c = arr.copy()
arr[0] = 42
print("Original:", arr)
print("Copy    :", c)          # unaffected

arr = np.array([1, 2, 3, 4, 5])
v = arr.view()
arr[0] = 42
print("Original:", arr)
print("View    :", v)          # affected

print("copy owns data:", c.base is None)
print("view base is original:", v.base is arr)
```

## Task 20

**Topic:** Pandas

**Task statement:** Complete the exercise on Pandas data-cleaning techniques.

```python
dirty = pd.DataFrame({
    "Duration": [60, 60, 45, 45, None, 450, 30],
    "Date": ["2020/12/01", "2020/12/02", "2020/12/03", "2020/12/03", "2020/12/05", "20201206", "2020/12/07"],
    "Calories": [409.1, 479.0, 340.0, 340.0, 300.0, 380.0, None],
})

# 1) Empty cells
print(dirty.dropna())                                 # remove rows
filled = dirty.fillna({"Calories": dirty["Calories"].mean()})   # replace with mean
filled["Duration"] = filled["Duration"].fillna(filled["Duration"].median())

# 2) Wrong format
filled["Date"] = pd.to_datetime(filled["Date"], format="mixed")

# 3) Wrong data
filled.loc[filled["Duration"] > 120, "Duration"] = 120

# 4) Duplicates
print(filled.duplicated())
filled = filled.drop_duplicates()
print(filled)
```
