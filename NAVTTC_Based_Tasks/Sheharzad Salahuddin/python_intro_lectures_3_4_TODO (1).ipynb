# Lecture 3 — File Handling & OOP Foundations

## 1. File handling

Programs often need to store information beyond the current run.

Python can read and write files using `open()`.

A common pattern is:

```python
with open("file.txt", "r") as file:
    content = file.read()
```

The `with` statement automatically closes the file afterwards.

Common modes:

| Mode | Meaning |
|---|---|
| `"r"` | read |
| `"w"` | write / overwrite |
| `"a"` | append |
| `"x"` | create a new file |

````python
from pathlib import Path

path = Path("example.txt")

path.write_text("Python is fun!\n", encoding="utf-8")
print(path.read_text(encoding="utf-8"))
````

### Writing multiple lines

`write()` stores text exactly as you provide it.

````python
with open("students.txt", "w", encoding="utf-8") as file:
    file.write("Aisha\n")
    file.write("Bilal\n")
    file.write("Hira\n")

with open("students.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

### Reading line by line

A file object is iterable, so you can use a `for` loop.

````python
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())
````

### Appending

Append mode adds new content without deleting the old content.

````python
with open("students.txt", "a", encoding="utf-8") as file:
    file.write("Omar\n")

with open("students.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

## 2. Exception handling

An **exception** is an error that occurs while the program is running.

Example:

```python
10 / 0
```

would raise a `ZeroDivisionError`.

We can handle expected errors with:

```python
try:
    ...
except SomeError:
    ...
```

````python
try:
    number = int("hello")
except ValueError:
    print("That was not a valid integer.")
````

### `else` and `finally`

- `else` runs if no exception occurred
- `finally` runs no matter what

````python
try:
    number = int("25")
except ValueError:
    print("Invalid number")
else:
    print("Conversion worked:", number)
finally:
    print("Finished attempting conversion.")
````

### Raising your own exceptions

Use `raise` when your function detects an invalid situation.

````python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative.")
    return age

try:
    print(set_age(-5))
except ValueError as error:
    print(error)
````

For example:

- `Student` → class

A class lets us keep related **data** and **behavior** together.

## 4. Constructors and instance variables

`__init__()` runs when an object is created.

`self` refers to the current object.

Instance variables belong to each individual object.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

aisha = Student("Aisha", 88)
bilal = Student("Bilal", 72)

print(aisha.name, aisha.score)
print(bilal.name, bilal.score)
````

## 5. Instance methods

Methods are functions defined inside a class.

An instance method usually works with data stored on `self`.

````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def passed(self):
        return self.score >= 70

aisha = Student("Aisha", 88)

print(aisha.passed())
````

## 6. Class variables

A class variable is shared by all instances.

Use instance variables for information that differs per object.

Use class variables for information shared by the class.

````python
class Student:
    school = "Python Academy"

    def __init__(self, name):
        self.name = name

aisha = Student("Aisha")
bilal = Student("Bilal")

print(aisha.school)
print(bilal.school)
print(Student.school)
````

### Class methods

A class method receives the class as `cls`.

It is commonly used to:
- work with class-level state
- create alternate constructors

````python
class Student:
    count = 0

    def __init__(self, name):
        self.name = name
        Student.count += 1

    @classmethod
    def how_many(cls):
        return cls.count

Student("Aisha")
Student("Bilal")

print(Student.how_many())
````

## 7. Destructors

`__del__()` is called when Python destroys an object.

You will see it in teaching material, but in real programs you should **not rely on it for important cleanup**.

Context managers such as `with open(...)` are more reliable for resources.

````python
class Demo:
    def __init__(self, name):
        self.name = name
        print(f"Created {self.name}")

    def __del__(self):
        # Exact timing is controlled by Python.
        pass

demo = Demo("example")
````

## 8. Public, protected, and private members

Python does not enforce access levels in exactly the same way as languages such as Java or C++.

By convention:

- `name` → public
- `_name` → protected-like convention
- `__name` → name-mangled

A leading underscore means:

> "This is internal; please do not use it directly unless necessary."

````python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance
        self.__pin = 1234

account = Account("Ali", 5000)

print(account.owner)
print(account._balance)

# print(account.__pin)  # Would raise AttributeError
````

### Name mangling

Names beginning with two underscores are internally changed by Python.

This reduces accidental access but is **not a security feature**.

````python
print(account.__dict__)
````

# Lecture 4 — Advanced Object-Oriented Programming

## 10. Inheritance

Inheritance lets one class reuse and extend another class.

- **Parent / base class** → general behavior
- **Child / subclass** → specialized behavior

````python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def bark(self):
        return "Woof!"

dog = Dog()

print(dog.speak())
print(dog.bark())
````

### `super()`

Use `super()` to call behavior from the parent class.

````python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, score):
        super().__init__(name)
        self.score = score

student = Student("Aisha", 88)
print(student.name, student.score)
````

## 11. Common inheritance patterns

### Multilevel inheritance

A → B → C

````python
class LivingThing:
    pass

class Animal(LivingThing):
    pass

class Dog(Animal):
    pass

dog = Dog()

print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
print(isinstance(dog, LivingThing))
````

### Hierarchical inheritance

Multiple child classes inherit from the same parent.

````python
class Vehicle:
    def move(self):
        return "Moving"

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

print(Car().move())
print(Bike().move())
````

### Multiple inheritance

A class can inherit from more than one parent.

Use this carefully because it can make class relationships harder to understand.

````python
class Camera:
    def take_photo(self):
        return "Photo taken"

class Phone:
    def call(self):
        return "Calling..."

class Smartphone(Camera, Phone):
    pass

device = Smartphone()

print(device.take_photo())
print(device.call())
````

## 12. Method Resolution Order (MRO)

With multiple inheritance, Python needs rules for deciding **where to look for a method**.

That order is called the **Method Resolution Order**.

````python
class A:
    def identify(self):
        return "A"

class B(A):
    def identify(self):
        return "B"

class C(A):
    def identify(self):
        return "C"

class D(B, C):
    pass

print(D().identify())
print(D.mro())
````

## 13. Object Relationships

### Association

Association means two independent objects interact.

Example: a `Teacher` teaches a `Student`.

Neither object necessarily owns the other.

````python
class Student:
    def __init__(self, name):
        self.name = name

class Teacher:
    def teach(self, student):
        return f"Teaching {student.name}"

student = Student("Aisha")
teacher = Teacher()

print(teacher.teach(student))
````

### Aggregation

Aggregation is a "has-a" relationship where the contained object can exist independently.

Example:
- A `Team` has `Player` objects.
- Players can exist even if the team object disappears.

````python
class Player:
    def __init__(self, name):
        self.name = name

class Team:
    def __init__(self, players):
        self.players = players

player1 = Player("Ali")
player2 = Player("Sara")

team = Team([player1, player2])

print([player.name for player in team.players])
````

### Composition

Composition is a stronger "has-a" relationship.

The outer object creates and owns the inner object.

Example:
- A `Car` creates and owns an `Engine`.

````python
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()

    def start(self):
        return self.engine.start()

car = Car()
print(car.start())
````

## 14. Polymorphism

Polymorphism means:

> Different objects can respond to the same operation in different ways.

This lets us write code that cares about **what an object can do**, rather than its exact class.

````python
class Dog:
    def speak(self):
        return "Woof"

class Cat:
    def speak(self):
        return "Meow"

animals = [Dog(), Cat()]

for animal in animals:
    print(animal.speak())
````

### Dynamic polymorphism through overriding

A subclass can replace a parent method with its own implementation.

````python
class Employee:
    def calculate_pay(self):
        return 0

class SalariedEmployee(Employee):
    def calculate_pay(self):
        return 100000

class HourlyEmployee(Employee):
    def calculate_pay(self):
        return 20 * 160

employees = [SalariedEmployee(), HourlyEmployee()]

for employee in employees:
    print(employee.calculate_pay())
````

## 15. Magic / dunder methods

Names such as `__init__`, `__str__`, and `__len__` are called **dunder methods**.

They let your objects work naturally with Python syntax and built-in functions.

````python
class Course:
    def __init__(self, title, students):
        self.title = title
        self.students = students

    def __str__(self):
        return self.title

    def __len__(self):
        return len(self.students)

course = Course("Python", ["Aisha", "Bilal", "Hira"])

print(course)
print(len(course))
````

## 16. Operator overloading

Dunder methods can define what operators mean for your own classes.

For example, `__add__()` defines what `+` means.

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

v1 = Vector(2, 3)
v2 = Vector(4, 1)

print(v1 + v2)
````

## 17. Abstract classes and methods

An abstract class defines a **required interface** for subclasses.

Python provides this through the `abc` module.

If a subclass does not implement the required abstract method, Python will prevent it from being instantiated.

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

rectangle = Rectangle(4, 5)
print(rectangle.area())
````

## 18. Empty classes

## 19. Data classes

For classes whose main job is storing data, `@dataclass` can remove repetitive code.

Python automatically creates methods such as `__init__()` and a useful representation.

````python
from dataclasses import dataclass

@dataclass
class StudentRecord:
    name: str
    score: int
    active: bool = True

student = StudentRecord("Aisha", 88)

print(student)
````

### Exercise 7 — Polymorphism

Create `Circle` and `Square` classes.

Each should have an `area()` method.

Then place both objects in a list and print each area using one loop.

````python
import math

# Step 1: Create a general Shape class
class Shape:
    def area(self):
        pass  # A general shape doesn't have an area formula yet

# Step 2: Create a Circle class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    # Circle calculates area using its radius
    def area(self):
        return math.pi * (self.radius ** 2)

# Step 3: Create a Square class
class Square(Shape):
    def __init__(self, side):
        self.side = side

    # Square calculates area using its side length
    def area(self):
        return self.side ** 2

# Step 4: Use Polymorphism in a loop
shapes = [Circle(5), Square(4)]

for s in shapes:
    print(f"The area is: {s.area():.2f}")
````


prints:

```text
PKR 500
```

````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        return Money(self.amount + other.amount)

    def __str__(self):
        return f"PKR {self.amount}"


m1 = Money(500)
m2 = Money(700)

print(m1)       
print(m1 + m2)  
````

# Mini Project — Simple Library System

Build a small object-oriented library system.

## Requirements

### `Book`
Each book should have:
- title
- author
- `is_available`, initially `True`

Methods:
- `borrow()` — mark unavailable
- `return_book()` — mark available
- `__str__()` — show title and author

### `Member`
Each member should have:
- name
- a list of borrowed books

Methods:
- `borrow_book(book)`
- `return_book(book)`

### `Library`
The library should:
- contain a list of books
- have `add_book(book)`
- have `show_available_books()`

### Behavior

A member may borrow a book only if it is available.

This mini-project uses:

- classes and objects
- constructors
- instance variables
- methods
- composition / aggregation
- conditionals
- lists
- loops
- dunder methods

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

    def borrow(self):
        if self.is_available:
            self.is_available = False
            return True
        return False

    def return_book(self):
        self.is_available = True

    def __str__(self):
        return f"{self.title} by {self.author}"


class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.borrow():
            self.borrowed_books.append(book)
            print(f"{self.name} borrowed '{book.title}'.")
        else:
            print(f"'{book.title}' is not available.")

    def return_book(self, book):
        if book in self.borrowed_books:
            book.return_book()
            self.borrowed_books.remove(book)
            print(f"{self.name} returned '{book.title}'.")
        else:
            print(f"{self.name} has not borrowed '{book.title}'.")


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_available_books(self):
        print("\nAvailable Books:")

        for book in self.books:
            if book.is_available:
                print(f"- {book}")


# Create books
book1 = Book("1984", "George Orwell")
book2 = Book("The Hobbit", "J.R.R. Tolkien")
book3 = Book("Clean Code", "Robert C. Martin")

# Create library
library = Library()

# Add books to library
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

# Create member
member = Member("Sheharzad")

# Show available books
library.show_available_books()

# Borrow a book
member.borrow_book(book1)

# Show available books again
library.show_available_books()

# Try to borrow the same book
member2 = Member("Ali")
member2.borrow_book(book1)

# Return the book
member.return_book(book1)

# Show available books again
library.show_available_books()
````
