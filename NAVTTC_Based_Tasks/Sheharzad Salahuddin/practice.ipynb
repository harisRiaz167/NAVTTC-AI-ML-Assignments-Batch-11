```python
vocab = {
    "[PAD]": 0,
    "[UNK]" : 1,
    "i" : 2,
    "love" : 3,
    "this" : 4,
    "film" : 5
}

sentence = "I love this film"
words = sentence.split()
encoded = [vocab.get(word, vocab["[UNK]"]) for word in words]

max_length = 6
encoded = encoded + [vocab["[PAD]"] * (max_length - len(encoded))]

print(encoded)
```
