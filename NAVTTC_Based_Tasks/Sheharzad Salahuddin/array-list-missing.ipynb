```python
def missing_number(L):
    n = len(L)
    num = n * (n + 1) // 2
    num2 = sum(L)
    return num - num2


print(missing_number([0, 1, 3]))
# print(missing_number([0, 1, 2]))
# print(missing_number([1, 2, 3, 5]))
# print(missing_number([4, 4]))

print(missing_number([6]))
```

```python
import numpy as np

x1, x2 = 4, 1

z = -2.5 + 0.6 * x1 - 0.8 * x2

print(f"Z = {z:.2f}")

p = 1 / (1 + np.exp(-z))

print(f"p = {p: .4f}")

for t in [0.50, 0.70]:
    pred = int(p > t)
    print(f"threshold {t:.2f} -> class {pred}")
```
