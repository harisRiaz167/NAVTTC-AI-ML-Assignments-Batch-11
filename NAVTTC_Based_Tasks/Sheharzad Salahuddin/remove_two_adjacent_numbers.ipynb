# Remove Adjacent Duplicates

Given a string `s`, repeatedly remove two adjacent characters if they are the same. Continue until no more adjacent duplicates can be removed. Return the final string.

**Approach (stack):** Walk through the string. If the current character matches the top of the stack, pop it (the pair cancels out); otherwise push it. Removing a pair can expose a new pair, and the stack handles that automatically because the next character is compared against the newly exposed top.

- Time: O(n)
- Space: O(n)

```python
def remove_duplicates(s: str) -> str:
    stack = []
    for ch in s:
        if stack and stack[-1] == ch:
            stack.pop()
        else:
            stack.append(ch)
    return "".join(stack)
```

```python
tests = [
    ("abbaca", "ca"),
    ("azxxzy", "ay"),
    ("aabbcc", ""),
    ("abccba", ""),
    ("abc", "abc"),
    ("a", "a"),
    ("", ""),
    ("112233", ""),
    ("1223", "13"),
]

for s, expected in tests:
    result = remove_duplicates(s)
    status = "PASS" if result == expected else "FAIL"
    print(f"{status}: {s!r} -> {result!r} (expected {expected!r})")
```
