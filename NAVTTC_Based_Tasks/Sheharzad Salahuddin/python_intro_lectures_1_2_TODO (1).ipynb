# Lecture 1 — Python Fundamentals & Data Structures

## 1. Values, expressions, and statements

A **value** is a piece of data.

Examples:
- `42`
- `3.14`
- `"hello"`
- `True`

An **expression** combines values and operators to produce a new value.

```python
2 + 3
```

A **statement** tells Python to do something.

```python
score = 10
```

Here, `score = 10` stores a value in a variable.

````python
print(42)
print(2 + 3)

score = 10
print(score)
````

### Predict before running

What do you think each expression will produce?

````python
print(10 + 5)
print(10 - 5)
print(10 * 5)
print(10 / 5)
````

## 2. Numbers, Booleans, and strings

Python values have **types**.

Three important beginner types are:

| Type | Example | Meaning |
|---|---|---|
| `int` | `25` | whole number |
| `float` | `25.5` | decimal number |
| `bool` | `True` | true/false value |
| `str` | `"Python"` | text |

Use `type()` to ask Python what type a value has.

````python
print(type(25))
print(type(25.5))
print(type(True))
print(type("Python"))
````

## 3. Variables

A variable is a **name that refers to a value**.

Think of it as a label.

```python
age = 18
```

The name is `age`; the value is `18`.

Variable names:
- can contain letters, numbers, and `_`
- cannot start with a number
- are case-sensitive: `score` and `Score` are different
- cannot use reserved keywords such as `if`, `for`, or `class`

````python
name = "Ayesha"
age = 18
height = 1.67
is_student = True

print(name)
print(age)
print(height)
print(is_student)
````

### Variables can change

The `=` symbol in Python means **assign**.

It does not mean mathematical equality.

````python
score = 10
print("Before:", score)

score = score + 5
print("After:", score)
````

## 8. Data Structures

### Lists

A list stores multiple values **in order**.

Lists are **mutable**, which means they can be changed.

````python
fruits = ["apple", "banana", "mango"]

print(fruits)
print(fruits[0])

fruits.append("orange")
print(fruits)

fruits[1] = "strawberry"
print(fruits)
````

### Tuples

A tuple is also ordered, but it is usually used for values that should **not change**.

Tuples are **immutable**.

````python
point = (10, 20)

print(point)
print(point[0])
print(point[1])
````

### Dictionaries

A dictionary stores **key-value pairs**.

Instead of asking for item `0`, you usually ask for an item by name.

````python
student = {
    "name": "Hassan",
    "age": 19,
    "score": 88
}

print(student["name"])
print(student["score"])

student["score"] = 91
student["city"] = "Lahore"

print(student)
````

### Sets

A set stores **unique values**.

Sets are useful when duplicates do not matter.

````python
topics = {"python", "math", "python", "physics"}

print(topics)

topics.add("statistics")
print(topics)
````

# Lecture 2 — Control Flow, Loops & Functions

## 9. Conditional execution

Programs become useful when they can make decisions.

The basic structure is:

```python
if condition:
    ...
elif another_condition:
    ...
else:
    ...
```

Python uses **indentation** to show which code belongs to each branch.

````python
temperature = 34

if temperature > 35:
    print("Very hot")
elif temperature > 25:
    print("Warm")
else:
    print("Cool")
````

### Multiple conditions

Logical operators let us combine conditions.

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("You may enter.")
else:
    print("Entry denied.")
````

### Nested conditionals

A conditional can exist inside another conditional.

Use nesting when the second decision only matters after the first one.

````python
logged_in = True
is_admin = False

if logged_in:
    print("Welcome!")

    if is_admin:
        print("Admin panel enabled.")
    else:
        print("Standard account.")
else:
    print("Please log in.")
````

### Ternary expression

A ternary expression is a short one-line conditional.

```python
value_if_true if condition else value_if_false
```

Use it for **small, simple decisions**.

````python
age = 17

status = "Adult" if age >= 18 else "Minor"
print(status)
````

## 10. `break`, `continue`, and `pass`

- `break` — stop the loop completely
- `continue` — skip the rest of this iteration
- `pass` — do nothing; useful as a placeholder

````python
for number in range(1, 8):
    if number == 5:
        break
    print(number)
````

Here, `continue` skips even numbers.

````python
for number in range(1, 8):
    if number % 2 == 0:
        continue
    print(number)
````

`pass` is useful when you want to create a block now and fill it in later.

````python
if True:
    pass

print("The program continues.")
````

## 11. `while` loops

A `while` loop repeats **while a condition is true**.

Be careful: if the condition never becomes false, you create an infinite loop.

````python
count = 1

while count <= 7:
    print(count)
    count = count + 1
````

## 12. `for` loops

A `for` loop is usually used to visit items in a collection.

````python
subjects = ["Math", "Physics", "Python"]

for subject in subjects:
    print(subject)
````

### `range()`

`range()` is commonly used when you want a sequence of integers.

`range(5)` produces values from `0` up to, but not including, `5`.

````python
for number in range(5):
    print(number)
````

### `enumerate()`

`enumerate()` gives both:
- the index
- the value

This is cleaner than manually keeping a counter.

````python
students = ["Aisha", "Bilal", "Hira"]

for index, student in enumerate(students):
    print(index, student)
````

Use `start=1` when you want human-friendly numbering.

````python
students = ["Aisha", "Bilal", "Hira"]

for number, student in enumerate(students, start=1):
    print(f"{number}. {student}")
````

## 13. Nested loops

A nested loop is a loop inside another loop.

The inner loop finishes all of its iterations for **each** outer-loop iteration.

````python
for row in range(1, 4):
    for column in range(1, 4):
        print(f"row={row}, column={column}")
````

## 14. List comprehensions

A list comprehension is a compact way to build a list.

Traditional version:

````python
squares = []

for number in range(1, 6):
    squares.append(number ** 2)

print(squares)
````

The same idea as a list comprehension:

````python
squares = [number ** 2 for number in range(1, 6)]

print(squares)
````

You can also add a condition.

````python
even_squares = [
    number ** 2
    for number in range(1, 11)
    if number % 2 == 0
]

print(even_squares)
````

## 15. Iterables and iterators

An **iterable** is something you can loop over.

Examples:
- list
- tuple
- string
- dictionary
- set

An **iterator** is the object that gives items **one at a time**.

You usually do not create iterators manually because a `for` loop handles this for you.

But we can see the mechanism:

````python
colors = ["red", "green", "blue"]

iterator = iter(colors)

print(next(iterator))
print(next(iterator))
print(next(iterator))
````

### What a `for` loop is roughly doing

This:

```python
for color in colors:
    print(color)
```

is conceptually similar to repeatedly calling `next()` on an iterator until there are no items left.

You do **not** need to memorize the internal mechanism yet. The important idea is:

> **Iterable = something we can iterate over. Iterator = object that produces the next item.**

## 16. Functions

A function packages reusable behavior.

```python
def function_name(parameters):
    ...
    return result
```

Use functions when you notice yourself doing the same kind of task repeatedly.

````python
def greet(name):
    return f"Hello, {name}!"

message = greet("Noor")
print(message)
````

### Parameters and arguments

In:

```python
def greet(name):
```

`name` is a **parameter**.

In:

```python
greet("Noor")
```

`"Noor"` is an **argument**.

````python
def add(a, b):
    return a + b

result = add(4, 7)
print(result)
````

### Keyword arguments

You can pass arguments by parameter name.

This often makes function calls easier to read.

````python
def create_profile(name, age, city):
    return f"{name}, age {age}, from {city}"

print(create_profile(age=20, city="Lahore", name="Zain"))
````

## 17. Variable scope

A variable created inside a function is usually **local** to that function.

A variable created outside a function is usually **global**.

Prefer local variables when possible. They make code easier to reason about.

````python
course_name = "Python"   # global

def show_course():
    room = "Lab 2"          # local
    print(course_name)
    print(room)

show_course()

# print(room)  # This would cause an error because room is local.
````

## 18. Lambda expressions

A lambda is a small anonymous function.

Normal function:

````python
def square(x):
    return x ** 2

print(square(5))
````

Equivalent lambda:

````python
square = lambda x: x ** 2

print(square(5))
````

## 19. `map()` and `filter()`

`map()` applies a function to every item.

`filter()` keeps items that satisfy a condition.

In modern Python, list comprehensions are often easier to read, but `map()` and `filter()` are still important concepts.

````python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))
evens = list(filter(lambda x: x % 2 == 0, numbers))

print(squared)
print(evens)
````

## 20. Inner / nested functions

A function can be defined inside another function.

This is useful when the inner helper only makes sense inside the outer function.

````python
def format_student(name, score):
    def format_score(value):
        return f"{value}%"
    return format_score
    return f"{name}: {format_score(score)}"

print(format_student("Amna", 87))
````

# Mini Project — Student Result Analyzer

This combines many ideas from Lectures 1 and 2.

We have a list of student dictionaries:

```python
students = [
    {"name": "Aisha", "score": 88},
    {"name": "Bilal", "score": 67},
    {"name": "Hira", "score": 94},
    {"name": "Omar", "score": 73},
]
```

Your program should:

1. Create a function `get_grade(score)`
2. Return:
   - `A` for 90+
   - `B` for 80–89
   - `C` for 70–79
   - `Needs Improvement` below 70
3. Loop over all students
4. Print a numbered report such as:

```text
1. Aisha — 88 — B
2. Bilal — 67 — Needs Improvement
...
```

5. Count how many students scored at least 70
6. Print the final count

````python
students = [
    {"name": "Aisha", "score": 88},
    {"name": "Bilal", "score": 67},
    {"name": "Hira", "score": 94},
    {"name": "Omar", "score": 73},
]

# TODO 1: write get_grade(score)
def get_grade(score):
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"

    return grade


# TODO 2: create a counter for passing students
passing_count = 0


# TODO 3: loop over students using enumerate(..., start=1)
for number, student in enumerate(students, start=1):

    name = student["name"]
    score = student["score"]
    grade = get_grade(score)

    # TODO 4: print each student's number, name, score, and grade
    print(number, name, score, grade)

    # TODO 5: count students who scored at least 70
    if score >= 70:
        passing_count += 1


print("Students scoring at least 70:", passing_count)
````
