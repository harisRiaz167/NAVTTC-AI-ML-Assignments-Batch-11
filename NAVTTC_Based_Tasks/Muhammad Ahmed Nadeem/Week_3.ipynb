### Task 3: Data Types & Explicit Precision Casting (`astype`)
**Objective:** Inspect numeric representations and cast floating point numbers to integer and boolean formats.

```python
# Task 3: Data Type Conversions
sensor_readings = np.array([2.45, 4.89, 0.00, 6.15])
int_cast = sensor_readings.astype(np.int32)
bool_cast = sensor_readings.astype(bool)

print(f"Float Array:     {sensor_readings} | dtype: {sensor_readings.dtype}")
print(f"Cast to Int32:   {int_cast} | dtype: {int_cast.dtype}")
print(f"Cast to Boolean: {bool_cast} | dtype: {bool_cast.dtype}")
```

### Task 4: Memory Management: Copy vs. View Deep-Dive
**Objective:** Demonstrate data mutation differences between NumPy shallow views and independent memory copies using `.base`.

```python
# Task 4: Copy vs View Memory Mechanics
base_array = np.array([1, 2, 3, 4])
shallow_view = base_array.view()
deep_copy = base_array.copy()

# Mutate view -> Mutates base
shallow_view[0] = 777
print(f"Base Array after View Mutation: {base_array}")
print(f"View references base array?      {shallow_view.base is not None}")

# Mutate copy -> Does NOT mutate base
deep_copy[1] = 999
print(f"Base Array after Copy Mutation: {base_array}")
print(f"Copy references base array?      {deep_copy.base is not None}")
```

### Task 5: Matrix Reshaping & Boolean Indexing (`where`)
**Objective:** Restructure a 1D vector into a 2D matrix, flatten back to vector, and perform filtered searches using `np.where()`.

```python
# Task 5: Reshape and Where Filtering
flat_data = np.arange(10, 70, 5)
matrix_3x4 = flat_data.reshape(3, 4)
reconstructed = matrix_3x4.flatten()

print("Reshaped 2D Matrix (3x4):\n", matrix_3x4)
print("Reconstructed 1D:", reconstructed)

# Filtering with np.where
filtered = matrix_3x4[matrix_3x4 > 35]
print("Values exceeding threshold (>35):", filtered)
```
