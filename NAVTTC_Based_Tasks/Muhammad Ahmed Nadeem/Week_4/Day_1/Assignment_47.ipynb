```python
import pandas as pd, numpy as np
df = pd.DataFrame({'A': [1, np.nan, 3, 3], 'B': [10, 20, 30, 30]})
df = df.drop_duplicates()
df['A'] = df['A'].fillna(df['A'].median())
print("Cleaned DataFrame:\n", df)
```
