### Task 1: Tabular Dataset Ingestion & Structural Inspection
**Objective:** Construct a representative corporate dataset, ingest into a Pandas DataFrame, and inspect dimensions, summary statistics, and missing data points.

```python
# Task 1: Dataset Exploration
import pandas as pd
import numpy as np

records = {
    'EmployeeID': [501, 502, 503, 504, 505, 506, 501],
    'Age': [26, 32, np.nan, 44, 28, 23, 26],
    'Salary_PKR': [120000, 185000, 240000, 165000, 195000, 110000, 120000],
    'Performance_Score': [85, 74, 92, np.nan, 80, 67, 85],
    'Branch': ['Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Karachi', 'Rawalpindi', 'Lahore']
}

df = pd.DataFrame(records)
print("DataFrame Preview (head 5):\n", df.head())
print(f"\nDimensions: {df.shape}")
print("Descriptive Summary:\n", df.describe())
```

---
### Task 2: Data Cleaning, Deduplication & Median Imputation
**Objective:** Detect missing values, eliminate duplicate rows with `drop_duplicates()`, and impute numerical nulls using median values to preserve data integrity.

```python
# Task 2: Cleaning and Imputation Pipeline
print("Missing Counts per Column:\n", df.isnull().sum())

# Deduplication
df_cleaned = df.drop_duplicates().copy()

# Median Imputation
med_age = df_cleaned['Age'].median()
med_score = df_cleaned['Performance_Score'].median()

df_cleaned['Age'] = df_cleaned['Age'].fillna(med_age)
df_cleaned['Performance_Score'] = df_cleaned['Performance_Score'].fillna(med_score)

print("\nCleaned Dataset (Deduplicated & Median Imputed):\n", df_cleaned)
```

---
### Task 3: Correlation Matrix Analysis & Multi-Group Aggregations
**Objective:** Compute pairwise Pearson correlations across numerical columns and perform branch-wise group analysis using `groupby()`.

```python
# Task 3: Correlation and GroupBy Analysis
num_fields = ['Age', 'Salary_PKR', 'Performance_Score']
corr_tbl = df_cleaned[num_fields].corr()
print("Pearson Correlation Matrix:\n", corr_tbl.round(6))

branch_summary = df_cleaned.groupby('Branch')['Salary_PKR'].agg(['mean', 'std'])
print("\nBranch-Wise Salary Analytics:\n", branch_summary)
```
