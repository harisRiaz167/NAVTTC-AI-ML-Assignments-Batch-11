### Assignment 47 (Week 4, Day 1)
**Topic:** `PandasComplete the exercise on Pandas data-cleaning techniques.`  
**Description:** PandasComplete the exercise on Pandas data-cleaning techniques.  
**Reference:** [https://www.w3schools.com/python/pandas_cleaning.asp](https://www.w3schools.com/python/pandas_cleaning.asp)


```python
import pandas as pd, numpy as np
df = pd.DataFrame({'A': [1, np.nan, 3, 3], 'B': [10, 20, 30, 30]})
df = df.drop_duplicates()
df['A'] = df['A'].fillna(df['A'].median())
print("Cleaned DataFrame:\n", df)
```
