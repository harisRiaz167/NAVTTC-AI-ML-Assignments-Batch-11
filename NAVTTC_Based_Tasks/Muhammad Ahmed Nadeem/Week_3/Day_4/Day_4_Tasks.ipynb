### Assignment 32 (Week 3, Day 4)
**Topic:** `NumPyComplete the exercise on NumPy array copying and viewing.`  
**Description:** NumPyComplete the exercise on NumPy array copying and viewing.  
**Reference:** [https://www.w3schools.com/python/numpy_copy_vs_view.asp](https://www.w3schools.com/python/numpy_copy_vs_view.asp)


```python
import numpy as np
orig = np.array([1, 2, 3])
cp = orig.copy()
vw = orig.view()
orig[0] = 99
print(f"Copy: {cp} (unaffected) | View: {vw} (updated)")
```
