### Assignment 27 (Week 2, Day 1)
**Topic:** `PythonComplete the exercise on user-defined exception handling in Python.`  
**Description:** PythonComplete the exercise on user-defined exception handling in Python.  
**Reference:** [https://www.programiz.com/python-programming/user-defined-exception](https://www.programiz.com/python-programming/user-defined-exception)


```python
class CustomAgeError(Exception): pass
def verify_age(age):
    if age < 18: raise CustomAgeError("Underage!")
    return "Eligible"
try:
    verify_age(15)
except CustomAgeError as e:
    print("Caught custom exception:", e)
```
