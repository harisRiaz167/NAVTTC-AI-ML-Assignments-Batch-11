### Task 1: Modular Function Design & Performance Evaluation
**Objective:** Build a clean, reusable Python function that accepts mandatory student details, arbitrary assessment marks using `*scores`, an optional pass mark threshold, and metadata via `**kwargs`.

```python
# Task 1: Reusable Function with *args and **kwargs

def evaluate_candidate(student_name: str, *scores: float, passing_mark: float = 50.0, **meta_info) -> dict:
    """
    Calculates the cumulative assessment score and assign a letter grade.
    """
    total_score = sum(scores)
    count = len(scores)
    avg_score = (total_score / count) if count > 0 else 0.0
    
    if avg_score >= 85:
        grade = 'A'
    elif avg_score >= 70:
        grade = 'B'
    elif avg_score >= passing_mark:
        grade = 'C'
    else:
        grade = 'F'
        
    return {
        'name': student_name,
        'assessments_count': count,
        'average': round(avg_score, 2),
        'grade': grade,
        'passed': avg_score >= passing_mark,
        'meta': meta_info
    }

# Function Execution
report = evaluate_candidate(
    "Muhammad Ahmed Nadeem", 88.0, 94.0, 82.0, 94.0,
    passing_mark=50.0,
    batch="Batch 11", domain="AI/ML Specialization"
)

print("=" * 16 + " STUDENT PERFORMANCE REPORT " + "=" * 16)
print(f"Candidate Name:    {report['name']}")
print(f"Assessments Taken: {report['assessments_count']}")
print(f"Overall Average:   {report['average']:.2f}%")
print(f"Letter Grade:      {report['grade']}")
print(f"Status:            {'Passed' if report['passed'] else 'Failed'}")
print(f"Batch Details:     {report['meta']}")
print("=" * 60)
```

### Task 2: Robust Built-in Exception Trapping
**Objective:** Create a secure calculation pipeline that executes divisions safely, intercepting `ZeroDivisionError`, `ValueError`, and `TypeError` with custom fallback responses.

```python
# Task 2: Defensive Numerical Division Handler

def calculate_ratio(numerator, denominator):
    try:
        num = float(numerator)
        den = float(denominator)
        result = num / den
    except ZeroDivisionError:
        print("[Case 2 Error] ZeroDivisionError: Zero cannot serve as a divisor.")
        return None
    except (ValueError, TypeError) as err:
        print(f"[Input Error] Non-numeric input provided: {err}")
        return None
    else:
        print(f"[Case 1] {num} divided by {den} = {result:.2f}")
        return result
    finally:
        print("--> Transaction log recorded successfully.")

# Executing test scenarios
calculate_ratio(250, 5)
calculate_ratio(250, 0)
```

### Task 3: Domain-Specific Custom Exception Classes
**Objective:** Build a domain-specific custom exception `AccountBalanceError` that derives from Python's base `Exception`, with validation rules for financial transactions.

```python
# Task 3: Custom Exception Implementation

class AccountBalanceError(Exception):
    """Raised when a withdrawal exceeds available account liquidity."""
    def __init__(self, requested: float, available: float):
        self.requested = requested
        self.available = available
        self.message = f"Insufficient funds. Requested: PKR {requested}, Available: PKR {available}"
        super().__init__(self.message)

class BankAccount:
    def __init__(self, initial_balance: float = 10000.0):
        self.balance = initial_balance
        
    def withdraw(self, amount: float):
        if amount > self.balance:
            raise AccountBalanceError(requested=amount, available=self.balance)
        self.balance -= amount
        print(f"Withdrawal of PKR {amount} approved. Remaining balance: PKR {self.balance}")

# Testing custom exception
account = BankAccount(10000.0)
try:
    account.withdraw(3000.0)
    account.withdraw(9000.0)
except AccountBalanceError as err:
    print(f"[Custom Exception Caught] AccountBalanceError: {err.message}")
```

### Task 4: Descriptive Central Tendency Statistics
**Objective:** Compute the statistical measures of central tendency: Mean ($\mu$), Median, and Mode for a sample distribution using first-principles logic.

```python
# Task 4: Mean, Median, and Mode from First Principles
import statistics

data_points = [14, 18, 14, 22, 25, 28, 14, 32, 35, 38]

# 1. Mean Calculation
mean_val = sum(data_points) / len(data_points)

# 2. Median Calculation
sorted_points = sorted(data_points)
n = len(sorted_points)
if n % 2 == 1:
    median_val = sorted_points[n // 2]
else:
    median_val = (sorted_points[(n // 2) - 1] + sorted_points[n // 2]) / 2.0

# 3. Mode Calculation
mode_val = statistics.mode(data_points)

print(f"Dataset: {data_points}")
print(f"Calculated Mean:   {mean_val:.2f}")
print(f"Calculated Median: {median_val:.2f}")
print(f"Calculated Mode:   {mode_val}")
```

### Task 5: Statistical Dispersion (Variance & Standard Deviation)
**Objective:** Compute statistical dispersion to measure data spread using population variance ($\sigma^2$) and standard deviation ($\sigma$).

```python
# Task 5: Variance and Standard Deviation
import math
import statistics

data_points = [14, 18, 14, 22, 25, 28, 14, 32, 35, 38]
mu = sum(data_points) / len(data_points)

# Sum of squared deviations
variance_pop = sum((x - mu) ** 2 for x in data_points) / len(data_points)
std_dev_pop = math.sqrt(variance_pop)
std_dev_sample = statistics.stdev(data_points)

print(f"Dataset: {data_points}")
print(f"Population Variance (σ²):   {variance_pop:.4f}")
print(f"Population Std Dev (σ):     {std_dev_pop:.4f}")
print(f"Sample Standard Deviation:  {std_dev_sample:.4f}")
```
