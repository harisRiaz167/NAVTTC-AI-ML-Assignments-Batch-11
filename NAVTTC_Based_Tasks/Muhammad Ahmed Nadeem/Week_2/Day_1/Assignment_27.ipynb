### Problem Statement
PythonComplete the exercise on user-defined exception handling in Python.


```python
class CustomAgeError(Exception): pass
def verify_age(age):
    if age < 18: raise CustomAgeError("Underage!")
    return "Eligible"
try:
    verify_age(15)
except CustomAgeError as e:
    print("Caught custom exception:", e)
```
