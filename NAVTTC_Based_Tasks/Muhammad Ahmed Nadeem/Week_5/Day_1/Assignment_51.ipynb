### Problem Statement
Machine LearningImplement linear regression in Python with scikit-learn.


```python
from sklearn.linear_model import LinearRegression
import numpy as np
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([50, 60, 70, 80, 90])
reg = LinearRegression().fit(X, y)
print(f"Linear Regression: Slope={reg.coef_[0]:.2f}, Intercept={reg.intercept_:.2f}")
print(f"R² Score: {reg.score(X, y):.4f}")
```
