### Problem Statement
PythonTake two numbers as input from the user, convert them to float, add them, and display the sum.


```python
def add_floats(val1=14.5, val2=25.3):
    total = float(val1) + float(val2)
    print(f"Input 1: {val1}, Input 2: {val2}")
    print(f"Sum of floats: {total}")
    return total
add_floats()
```
