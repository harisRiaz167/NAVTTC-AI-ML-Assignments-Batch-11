### Problem Statement
PythonWrite a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).


```python
import math
a, b, c = 5.0, 6.0, 7.0
s = (a + b + c) / 2.0
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print(f"Sides: a={a}, b={b}, c={c} | Semi-perimeter: {s}")
print(f"Triangle Area (Heron's Formula): {area:.4f}")
```
