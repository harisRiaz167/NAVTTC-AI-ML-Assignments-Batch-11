### Assignment 8 (Week 1, Day 3)
**Topic:** `PythonWrite a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).`  
**Description:** PythonWrite a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).  



```python
import math
a, b, c = 5.0, 6.0, 7.0
s = (a + b + c) / 2.0
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
print(f"Sides: a={a}, b={b}, c={c} | Semi-perimeter: {s}")
print(f"Triangle Area (Heron's Formula): {area:.4f}")
```

### Assignment 9 (Week 1, Day 3)
**Topic:** `PythonSolve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).`  
**Description:** PythonSolve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).  



```python
import cmath
a, b, c = 1.0, 5.0, 6.0
d = (b**2) - (4*a*c)
r1 = (-b + cmath.sqrt(d)) / (2*a)
r2 = (-b - cmath.sqrt(d)) / (2*a)
print(f"Equation: {a}x² + {b}x + {c} = 0")
print(f"Discriminant: {d} -> Roots: {r1.real}, {r2.real}")
```
