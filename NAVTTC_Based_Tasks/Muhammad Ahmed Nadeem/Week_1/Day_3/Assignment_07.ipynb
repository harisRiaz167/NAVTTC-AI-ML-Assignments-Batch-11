### Problem Statement
PythonWrite a Python program to convert a temperature from Celsius to Fahrenheit (the example uses 37.5°C).


```python
celsius = 37.5
fahrenheit = (celsius * 9.0 / 5.0) + 32.0
print(f"Celsius: {celsius}°C -> Fahrenheit: {fahrenheit}°F")
```
