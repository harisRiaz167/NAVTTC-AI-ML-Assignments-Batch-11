PythonFind the square root of a real or complex number using Python's cmath module (the example uses 1+2j).


```python
import cmath
c_val = 1 + 2j
sqrt_c = cmath.sqrt(c_val)
print(f"Complex Number: {c_val}")
print(f"Square Root via cmath.sqrt(): {sqrt_c}")
```
