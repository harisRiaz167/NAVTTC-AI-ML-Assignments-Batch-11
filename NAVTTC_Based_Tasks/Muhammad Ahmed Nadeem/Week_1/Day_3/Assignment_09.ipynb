### Problem Statement
PythonSolve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).


```python
import cmath
a, b, c = 1.0, 5.0, 6.0
d = (b**2) - (4*a*c)
r1 = (-b + cmath.sqrt(d)) / (2*a)
r2 = (-b - cmath.sqrt(d)) / (2*a)
print(f"Equation: {a}x² + {b}x + {c} = 0")
print(f"Discriminant: {d} -> Roots: {r1.real}, {r2.real}")
```
