### Problem Statement
PythonTake a distance in kilometers from the user and convert it to miles using the conversion factor shown in the PDF (0.621371).


```python
km = 100.0
conv = 0.621371
miles = km * conv
print(f"Distance: {km} km = {miles:.4f} miles (factor: {conv})")
```
