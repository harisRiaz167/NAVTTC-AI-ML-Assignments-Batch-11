### Problem Statement
PythonRun and examine the provided nested-if example using i = 10 and nested conditions i < 10 and i < 12.


```python
i = 10
print(f"Variable i = {i}")
if i == 10:
    print("Outer if passed: i == 10")
    if i < 12:
        print("Nested if 1 passed: i < 12")
    if i < 10:
        print("This will not print")
    else:
        print("Nested else passed: i is not less than 10")
```
