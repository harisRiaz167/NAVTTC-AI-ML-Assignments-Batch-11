PythonRun and examine the provided if-elif-else example using i = 20 and conditions i == 10, i == 15, and i == 20.


```python
i = 20
print(f"Variable i = {i}")
if i == 10:
    print("i is 10")
elif i == 15:
    print("i is 15")
elif i == 20:
    print("i is 20 (Matched elif branch)")
else:
    print("i is not present")
```
