**Reference:** [https://www.geeksforgeeks.org/python-break-statement/](https://www.geeksforgeeks.org/python-break-statement/)


```python
print("Finding number 7 in list [1, 3, 5, 7, 9]:")
for n in [1, 3, 5, 7, 9]:
    if n == 7:
        print(f"  Found {n}! Terminating with break.")
        break
```

### Assignment 18 (Week 1, Day 5)
**Topic:** `PythonComplete the exercise on the continue statement in Python.`  
**Description:** PythonComplete the exercise on the continue statement in Python.  
**Reference:** [https://www.geeksforgeeks.org/python-continue-statement/](https://www.geeksforgeeks.org/python-continue-statement/)


```python
print("Skipping even numbers using continue:")
for n in range(1, 7):
    if n % 2 == 0:
        continue
    print(f"  Odd: {n}")
```
