### Task 5: Temperature Scale Conversion (Celsius to Fahrenheit)
**Problem Statement:**  
Write a Python program to convert a given Celsius temperature ($37.5^\circ\text{C}$) into degrees Fahrenheit using the standard conversion formula.

**Formula:**
$$
F = \left(C \times \frac{9}{5}\right) + 32
$$
where $C$ denotes Celsius and $F$ denotes Fahrenheit.

```python
# Task 5: Convert Celsius to Fahrenheit

celsius = 37.5
fahrenheit = (celsius * 9 / 5) + 32

print("Temperature in Celsius:", celsius)
print("Temperature in Fahrenheit:", fahrenheit)
```

### Task 6: Triangle Area Calculation Using Heron's Formula
**Problem Statement:**  
Write a Python program to determine the area of a triangle given the lengths of its three sides: $a = 5$, $b = 6$, and $c = 7$ using Heron's Formula.

**Mathematical Formulation:**
1. Semi-perimeter calculation:
$$
s = \frac{a + b + c}{2}
$$
2. Area computation:
$$
A = \sqrt{s(s - a)(s - b)(s - c)}
$$

```python
# Task 6: Area of a Triangle Using Heron's Formula
import math

a = 5
b = 6
c = 7

# Calculate semi-perimeter
s = (a + b + c) / 2

# Heron's formula for triangle area
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Side a:", a)
print("Side b:", b)
print("Side c:", c)
print("Semi-perimeter:", s)
print("Area of the triangle:", area)
```

### Task 7: Quadratic Equation Solver
**Problem Statement:**  
Write a Python program to solve the quadratic equation $ax^2 + bx + c = 0$ for coefficients $a = 1, b = 5, c = 6$ using the quadratic formula and the `cmath` module.

**Mathematical Formulation:**
The roots are calculated using the quadratic formula:
$$
x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}
$$
where $\Delta = b^2 - 4ac$ is the discriminant.

```python
# Task 7: Solve Quadratic Equation (ax^2 + bx + c = 0)
import cmath

a = 1
b = 5
c = 6

# Calculate discriminant
discriminant = b**2 - 4*a*c

# Calculate roots
root1 = (-b + cmath.sqrt(discriminant)) / (2*a)
root2 = (-b - cmath.sqrt(discriminant)) / (2*a)

print("First root:", root1)
print("Second root:", root2)
```

### Task 8: Metric Distance Conversion (Kilometers to Miles)
**Problem Statement:**  
Write a Python program that takes a distance in kilometers from the user and converts it to miles using the standard conversion factor $0.621371$.

**Formula:**
$$
\text{Miles} = \text{Kilometers} \times 0.621371
$$

```python
# Task 8: Convert Kilometers to Miles

kilometers = float(input("Enter distance in kilometers: "))
miles = kilometers * 0.621371

print("Distance in kilometers:", kilometers)
print("Distance in miles:", miles)
```
