#Write a Python program to find the area of a triangle using its three sides (the example uses a=5, b=6, c=7 and the semi-perimeter formula).

```python
a = 5
b = 6
c = 7
s = (a + b + c) / 2
area = (s * (s-a) * (s-b) * (s-c)) ** 0.5
print("Area of triangle is:", area)
```

#Solve the quadratic equation ax² + bx + c = 0 using the cmath module (the example uses a=1, b=5, c=6).

```python
import cmath
a = 1
b = 5
c = 6

d = (b**2) - (4*a*c)

sol1 = (-b - cmath.sqrt(d)) / (2*a)
sol2 = (-b + cmath.sqrt(d)) / (2*a)
print("The solutions are:", sol1, "and", sol2)
```
