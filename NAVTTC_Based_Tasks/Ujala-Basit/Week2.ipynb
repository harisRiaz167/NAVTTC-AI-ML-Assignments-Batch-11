#Complete the exercise on user-defined exception handling in Python.
Reference: https://www.programiz.com/python-programming/user-defined-exception

```python
class InvalidAgeException(Exception):
    pass

age = 15

try:
    if age < 18:
        raise InvalidAgeException
    else:
        print("Eligible to vote")
except InvalidAgeException:
    print("Exception occurred: Age must be at least 18.")
```

#Calculate mean, median, and mode.
Reference: https://www.w3schools.com/python/python_ml_mean_median_mode.asp

```python
import numpy as np
from scipy import stats

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

x = np.mean(speed)
y = np.median(speed)
z = stats.mode(speed, keepdims=True)

print("Mean:", x)
print("Median:", y)
print("Mode:", z.mode[0])
```

#Calculate standard deviation.
