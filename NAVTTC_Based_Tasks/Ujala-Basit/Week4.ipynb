#Complete the exercise on NumPy array copying and viewing.
Reference: https://www.w3schools.com/python/numpy_copy_vs_view.asp

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
y = arr.view()

print("Copy:", x)
print("View:", y)
```
